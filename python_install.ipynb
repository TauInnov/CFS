# התקנת פייתון על המחשב האישי


את כל תרגילי הקורס אפשר להריץ ישירות מהדפדפן דרך [Google Colab](https://colab.research.google.com/), בלי להתקין דבר. עם זאת, מומלץ להתקין פייתון גם על המחשב האישי — כדי לעבוד גם בלי אינטרנט, וכדי להתרגל לסביבת עבודה מקומית שתשמש אתכם גם בהמשך הלימודים.

בעמוד זה מוסבר כיצד להתקין פייתון עם כל הספריות הדרושות לקורס (NumPy, Matplotlib, Pandas ועוד), עבור **Windows** ועבור **Mac**.


## ההמלצה: Anaconda

הדרך הפשוטה ביותר להתקין פייתון היא באמצעות [Anaconda](https://www.anaconda.com/download) — חבילה חינמית שכוללת יחד עם פייתון גם את כל הספריות המדעיות הדרושות לקורס (NumPy, Pandas, Matplotlib ועוד) וגם את Jupyter Notebook, כך שאין צורך להתקין כל ספרייה בנפרד.


## התקנה ב-Windows

1. היכנסו לכתובת [anaconda.com/download](https://www.anaconda.com/download) ולחצו על **Download**. יש להזין כתובת מייל, ולאחר מכן יישלח קישור להורדה (אפשר גם לדלג ולהוריד ישירות).
2. ודאו שההורדה היא עבור **Windows** (הגרסה המוצעת כברירת מחדל בדרך כלל מתאימה).
3. פתחו את קובץ ההתקנה שהורד (`Anaconda3-...-Windows-x86_64.exe`) ועברו על אשף ההתקנה:
   - בשלב "Select Installation Type" בחרו **Just Me** (מומלץ, לא דורש הרשאות מנהל).
   - אפשר להשאיר את נתיב ההתקנה כברירת המחדל.
   - בשלב **Advanced Installation Options** אין חובה לסמן "Add Anaconda to my PATH environment variable" — לא מומלץ על ידי המתקין עצמו, ופותחים סביבת עבודה ייעודית (**Anaconda Prompt**, ראו בהמשך) במקום.
   - לחצו **Install** והמתינו לסיום (ההתקנה עשויה לקחת מספר דקות).

### פתיחת טרמינל ב-Windows
לאחר ההתקנה, פתחו את תפריט **התחל (Start)**, הקלידו `Anaconda Prompt` ולחצו Enter. ייפתח חלון שורת פקודה (טרמינל) שבו פייתון וכל הספריות זמינים ישירות — זו הדרך המומלצת להריץ פקודות פייתון ב-Windows.

### הרצת נוטבוק
בתוך Anaconda Prompt הקלידו:
```
jupyter notebook
```
ולחצו Enter. ייפתח חלון דפדפן עם ממשק Jupyter, שבו ניתן ליצור ולהריץ קובצי נוטבוק (`.ipynb`).

לחלופין, אפשר לפתוח את **Anaconda Navigator** (מתפריט Start) — ממשק גרפי שממנו אפשר להפעיל את Jupyter Notebook בלחיצת כפתור, בלי להשתמש בטרמינל כלל.


## התקנה ב-Mac

1. בדקו איזה מעבד יש במחשב שלכם: תפריט אפל (הסמל ) בפינה הימנית העליונה ← **About This Mac**. אם כתוב "Apple M1/M2/M3/..." יש לכם מעבד Apple Silicon; אם כתוב "Intel" יש לכם מעבד Intel.
2. היכנסו לכתובת [anaconda.com/download](https://www.anaconda.com/download) והורידו את קובץ ההתקנה עבור **macOS** בהתאם למעבד שלכם (Apple Silicon או Intel), בפורמט **`.pkg`** (מומלץ, מתקין גרפי).
3. פתחו את קובץ ה-`.pkg` שהורד ועברו על אשף ההתקנה בלחיצות **Continue** ← **Agree** ← **Install**, עד לסיום ההתקנה. ייתכן שתידרשו להזין את סיסמת המשתמש שלכם במחשב.

### פתיחת טרמינל ב-Mac
לחצו **Cmd + Space** לפתיחת Spotlight, הקלידו `Terminal` ולחצו Enter. (אפשר גם למצוא את Terminal תחת Applications ← Utilities.) לאחר ההתקנה, Anaconda מוסיפה את עצמה אוטומטית לטרמינל — אם תפתחו טרמינל חדש (חשוב: חלון **חדש**, לא כזה שכבר היה פתוח לפני ההתקנה) אמורה להופיע התוספת `(base)` בתחילת השורה.

### הרצת נוטבוק
בטרמינל הקלידו:
```
jupyter notebook
```
ולחצו Enter. ייפתח חלון דפדפן עם ממשק Jupyter, שבו ניתן ליצור ולהריץ קובצי נוטבוק (`.ipynb`).

לחלופין, אפשר לפתוח את **Anaconda Navigator** (מתוך Launchpad או Applications) — ממשק גרפי שממנו אפשר להפעיל את Jupyter Notebook בלחיצת כפתור, בלי להשתמש בטרמינל כלל.


## בדיקת ההתקנה

כדי לוודא שההתקנה הצליחה, פתחו טרמינל (Anaconda Prompt ב-Windows, Terminal ב-Mac) והריצו:
```
python --version
```
אמור להופיע מספר גרסה (למשל `Python 3.12.4`). אם התקבלה שגיאה שהפקודה לא מזוהה — ודאו שאתם משתמשים ב-Anaconda Prompt (ב-Windows) או שפתחתם חלון טרמינל חדש לאחר ההתקנה (ב-Mac).
