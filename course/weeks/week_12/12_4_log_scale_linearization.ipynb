## סקאלה לוגריתמית ולינאריזציה

מסלול הזריקה אינו דורש סולם לוגריתמי — הערכים בו נמצאים כולם באותו סדר גודל. אבל כשגודל פיזיקלי משתנה על פני **כמה סדרי גודל**, סולם ליניארי "דוחס" את הערכים הקטנים לכדי קו אחד צמוד לציר, והגרף הופך חסר תועלת. הפתרון הוא סולם לוגריתמי:

```python
ax.set_xscale("log")
ax.set_yscale("log")
```

אפשר להפעיל את שניהם יחד (log-log), רק אחד מהם, או אף אחד — תלוי בטווח הערכים בכל ציר בנפרד.

בחלקו השני של העמוד הרעיון של סולם לוגריתמי מוביל ל**לינאריזציה**: הפיכת קשר לא-ליניארי לקו ישר, שאליו אפשר להתאים קו עם ‎`linear_fit`‎.

### דוגמה: טווח הזריקה כתלות במהירות ההתחלתית

הטווח נתון על ידי $R(v_0) = v_0^2 \sin(2\theta)/g$ — תלות **ריבועית** ב-$v_0$. אם נסרוק $v_0$ על פני טווח רחב (למשל בין 1 ל-1000 מ/ש, כתרחיל מתמטי גרידא, לא ערכים פיזיקליים סבירים לזריקת יד), $R$ ישתנה על פני כמה סדרי גודל, ונשווה תצוגה ליניארית מול log-log.

```````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
theta = np.radians(45)
v0_range = 10**np.linspace(0, 3, 200)   # 1 ... 1000, log-spaced
R_range = v0_range**2 * np.sin(2 * theta) / g

fig, axs = plt.subplots(1, 2, figsize=(10, 4))

axs[0].plot(v0_range, R_range)
axs[0].set_title("Linear scale")
axs[0].set_xlabel(r"$v_0$ (m/s)")
axs[0].set_ylabel("R (m)")

axs[1].plot(v0_range, R_range)
axs[1].set_xscale("log")
axs[1].set_yscale("log")
axs[1].set_title("Log-log scale")
axs[1].set_xlabel(r"$v_0$ (m/s)")
axs[1].set_ylabel("R (m)")

plt.show()
```````

בגרף הליניארי (משמאל) כמעט כל המידע על ערכי $v_0$ הקטנים "נעלם" צמוד לציר. בגרף ה-log-log (מימין) רואים קו ישר — תוצאה של חוק חזקה ($R\propto v_0^2$). השיפוע של הקו הישר הוא החזקה, 2: כשה-$v_0$ גדל פי 10, הטווח גדל פי $10^2 = 100$.

חשוב: הביטוי ‎`10**np.linspace(0, 3, 200)`‎ פורש 200 **מעריכים** במרווחים שווים בין 0 ל-3, והחזקה הופכת אותם לערכים בין $10^0$ ל-$10^3$. כך הנקודות פרוסות באופן אחיד **בסולם הלוגריתמי**, ולא באופן אחיד בין 1 ל-1000.

### דוגמה נוספת: דעיכה בסולם לוגריתמי בציר אחד

בגוף המתקרר מסעיף 8.1, ההפרש מטמפרטורת החדר, $T - 20$, קטן פי $(1-k)$ בכל דקה. בסולם לוגריתמי **בציר y בלבד** דעיכה כזו נראית כקו ישר. זה ההבדל מהדוגמה הקודמת: חוק חזקה נראה ישר בסולם log-log, ודעיכה מעריכית נראית ישרה כשרק ציר y לוגריתמי.

```````python
T0, k = 80.0, 0.1                # initial temperature (°C), cooling constant
minutes = np.arange(0, 61)       # 0, 1, ..., 60 min
gap = (T0 - 20) * (1 - k)**minutes   # T - 20: distance from the room (°C)

fig, axs = plt.subplots(1, 2, figsize=(10, 4))
axs[0].plot(minutes, gap)
axs[0].set_title("Linear scale")
axs[0].set_xlabel("t (min)")
axs[0].set_ylabel("T - 20 (°C)")
axs[1].plot(minutes, gap)
axs[1].set_yscale("log")
axs[1].set_title("Log scale on y only")
axs[1].set_xlabel("t (min)")
axs[1].set_ylabel("T - 20 (°C)")
plt.show()
```````

### לינאריזציה: דעיכה מעריכית דרך ln

‎`linear_fit`‎ יודעת להתאים רק **קווים ישרים**. אבל הרבה תופעות פיזיקליות אינן ליניאריות -- למשל **בליעה** של קרינה דרך חומר, לפי חוק באר-למברט: $I(x) = I_0 e^{-\mu x}$, כאשר $I$ עוצמת הקרינה, $x$ עובי הבולם, $\mu$ מקדם הבליעה.

הרעיון: מפעילים ‎`ln`‎ על שני האגפים, $\ln I = \ln I_0 - \mu x$, ומתקבל **קו ישר** ב-$x$ מול $\ln I$. שיפוע הקו הוא $-\mu$, החיתוך הוא $\ln I_0$. זו "לינאריזציה" -- הופכים בעיה לא-ליניארית לליניארית על ידי שינוי משתנה. מאותה סיבה דעיכת הגוף המתקרר נראתה ישרה בסולם לוגריתמי בציר y: סולם כזה ממקם כל ערך לפי הלוגריתם שלו.

הנתונים: עובי הבולם ‎`x_abs`‎ והעוצמה המדודה ‎`I_meas`‎ בכל עובי. הם נוצרו מהמודל עם $\mu = 0.35$ ו-$I_0 = 120$, ועם רעש מדידה, כדי שאפשר יהיה להשוות את תוצאת ההתאמה לערכים האמיתיים. בסולם לוגריתמי בציר y הנקודות קרובות לקו ישר.

```````python
import numpy as np
import matplotlib.pyplot as plt

# Measured intensities: the model I0*exp(-mu*x) with I0 = 120, mu = 0.35,
# plus measurement noise
rng = np.random.default_rng(3)
x_abs = np.arange(9, dtype=float)
I_true = 120.0 * np.exp(-0.35 * x_abs)
I_meas = I_true + rng.normal(0, 3.0, size=len(x_abs))
```````

```````python
I0_true = 120.0   # the values used to make the data
mu_true = 0.35
print(np.round(I_meas, 2))

fig, ax = plt.subplots()
ax.plot(x_abs, I_meas, "o")
ax.set_yscale("log")
ax.set_xlabel("Absorber thickness x (arb. units)")
ax.set_ylabel("Intensity I (arb. units)")
plt.show()
```````

### לינאריזציה והתאמה

```````python
def linear_fit(x, y):
    x_bar, y_bar = x.mean(), y.mean()
    m = np.sum((x - x_bar) * (y - y_bar)) / np.sum((x - x_bar)**2)
    b = y_bar - m * x_bar
    return m, b

y_lin = np.log(I_meas)          # ln(I)
m, b = linear_fit(x_abs, y_lin)

mu_fit = -m
I0_fit = np.exp(b)
print(f"mu משוחזר = {mu_fit:.3f}  (אמיתי: {mu_true})")
print(f"I0 משוחזר = {I0_fit:.1f}  (אמיתי: {I0_true})")
```````

### באג נפוץ: ln על ערך שלילי או אפס

אם יש בנתונים ערך שלילי (למשל אחרי חיסור רקע אגרסיבי מדי) או אפס, ‎`np.log`‎ **לא זורק שגיאה** -- הוא מחזיר ‎`nan`‎ (עם ‎`RuntimeWarning`‎ שקל לפספס), וממשיך הלאה בשקט. אם לא בודקים במפורש, ה-‎`nan`‎ הזה מזהם את ‎`linear_fit`‎ כולה -- וכל התוצאה (‎`m`‎, ‎`b`‎) הופכת ל-‎`nan`‎, בלי שום שגיאה שמסבירה מדוע.

```````python
I_meas_bad = I_meas.copy()
I_meas_bad[-1] = I_meas_bad[-1] - 12.0   # לדוגמה: חיסור רקע שגוי הוציא ערך שלילי

y_lin_bad = np.log(I_meas_bad)   # RuntimeWarning, לא Exception
print(y_lin_bad)
print("יש nan בנתונים:", np.isnan(y_lin_bad).any())

m_bad, b_bad = linear_fit(x_abs, y_lin_bad)
print(f"m עם ה-nan בפנים: {m_bad}")   # nan מזהם הכל
```````
