## קריאת קוד התאמה קיים

לעיתים קרובות מקבלים קוד ניתוח שכבר כתוב: ממחברת של שותף לעבודה, מניתוח קודם או מסוכן בינה מלאכותית. לפני שמשתמשים במספרים שהוא מדפיס, קוראים אותו ומבינים מה הוא מחשב. סדר קריאה שימושי:

1. **הנתונים**: מאין הם מגיעים, ואילו עמודות נבחרות.
2. **הפונקציות**: מה כל פונקציה מקבלת ומה היא מחזירה. כדאי להציב בראש קלט קטן, למשל שלוש נקודות.
3. **החישוב הראשי**: באיזו שורה מחושב כל גודל שמודפס.
4. **הפלט**: האם המספרים סבירים, למשל שיפוע קרוב לצפי התאורטי ו-χ² מצומצם קרוב ל-1.

הסקריפט שבעמוד זה מתאים קו ישר לטווח הממוצע מול $\sin(2\theta)$, כמו בסעיפים 12.1 עד 12.5, אבל הוא כתוב מעט אחרת. שני הפרמטרים שמורים במערך אחד, ‎`A`‎: ‎`A[0]`‎ הוא החיתוך $a_0$, ו-‎`A[1]`‎ הוא השיפוע $a_1$. העמודות נבחרות לפי מיקום, עם ‎`iloc`‎: ‎`table.iloc[:, 1]`‎ פירושו כל השורות, והעמודה במקום 1, כלומר העמודה השנייה (הספירה מתחילה מ-0). מספרי השורות מופיעים משמאל לקוד.

``````python
# Mean range against sin(2*theta): straight-line fit
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

table = pd.DataFrame({"angle_deg": [15, 30, 45, 60, 75],
                      "R_mean": [21.06, 34.54, 41.31, 38.46, 20.96],
                      "R_err": [0.60, 0.98, 1.14, 2.41, 0.45]})

def linear(A, x):
    return A[0] + A[1] * x

def fit_line(x, y):
    x_bar = x.mean()
    y_bar = y.mean()
    a1 = ((x - x_bar) * (y - y_bar)).sum() / ((x - x_bar)**2).sum()
    a0 = y_bar - a1 * x_bar
    return np.array([a0, a1])

x = np.sin(2 * np.radians(table.iloc[:, 0]))
y = table.iloc[:, 1]
dy = table.iloc[:, 2]

A = fit_line(x, y)
chi2 = (((y - linear(A, x)) / dy)**2).sum()
chi2_red = chi2 / (len(x) - 2)
print(f"a0 = {A[0]:.2f}, a1 = {A[1]:.2f}, chi2_red = {chi2_red:.2f}")
``````

המשך הסקריפט מצייר את הנקודות עם קווי שגיאה ואת קו ההתאמה:

``````python
x_line = np.linspace(0, 1, 100)
fig, ax = plt.subplots()
ax.errorbar(x, y, yerr=dy, fmt="o", capsize=3, label="Data")
ax.plot(x_line, linear(A, x_line), label="Fit")
ax.set_xlabel("sin(2θ)")
ax.set_ylabel("Mean range (m)")
ax.legend()
plt.show()
``````

### קריאה לפי הסדר

שלושה משלבי הקריאה, על הסקריפט הזה:

1. **הנתונים.** בשורות 6 עד 8 נבנית טבלה קטנה, שורה לכל זווית, עם שלוש עמודות: הזווית במעלות, הטווח הממוצע והשגיאה שלו (הערכים של סעיף 12.1, מעוגלים). בשורות 20 עד 22 ‎`iloc`‎ בוחר את העמודות לפי מיקום: ‎`x`‎ מחושב מהעמודה במקום 0, ‎`y`‎ הוא העמודה במקום 1, ו-‎`dy`‎ היא העמודה במקום 2.
2. **הפונקציות.** ‎`linear`‎ מקבלת מערך של שני פרמטרים ומחזירה את ערכי הקו בכל נקודה של ‎`x`‎. ‎`fit_line`‎, בשורות 13 עד 18, היא נוסחת הריבועים הפחותים מסעיף 12.3, כתובה במפורש. היא מחזירה את שני הפרמטרים במערך אחד, והחיתוך ראשון.
3. **הפלט.** השיפוע $a_1 \approx 41.3$ מטר קרוב לערך התאורטי 40.8, החיתוך $a_0 \approx 0.44$ מטר קרוב ל-0, ו-χ² המצומצם, כ-1.33, קרוב ל-1. המספרים סבירים.

השלב שנותר, החישוב הראשי, הוא נושא השאלות שבהמשך.

**שאלה 3.** בעותק הבא של הסקריפט שונתה שורה אחת. הוא מדפיס את אותם $a_0$ ו-$a_1$, אבל χ² המצומצם שלו גדול פי מאות. מצאו את השורה שהשתנתה, והסבירו מדוע $a_0$ ו-$a_1$ לא השתנו.

``````python
# Mean range against sin(2*theta): straight-line fit
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

table = pd.DataFrame({"angle_deg": [15, 30, 45, 60, 75],
                      "R_mean": [21.06, 34.54, 41.31, 38.46, 20.96],
                      "R_err": [0.60, 0.98, 1.14, 2.41, 0.45]})

def linear(A, x):
    return A[1] + A[0] * x

def fit_line(x, y):
    x_bar = x.mean()
    y_bar = y.mean()
    a1 = ((x - x_bar) * (y - y_bar)).sum() / ((x - x_bar)**2).sum()
    a0 = y_bar - a1 * x_bar
    return np.array([a0, a1])

x = np.sin(2 * np.radians(table.iloc[:, 0]))
y = table.iloc[:, 1]
dy = table.iloc[:, 2]

A = fit_line(x, y)
chi2 = (((y - linear(A, x)) / dy)**2).sum()
chi2_red = chi2 / (len(x) - 2)
print(f"a0 = {A[0]:.2f}, a1 = {A[1]:.2f}, chi2_red = {chi2_red:.2f}")
``````

`````{admonition} פתרון
:class: dropdown, tip
השורה שהשתנתה היא שורה 11: ‎`return A[1] + A[0] * x`‎. הפונקציה מחליפה בין החיתוך לשיפוע, ולכן מחזירה את הערכים של קו אחר לגמרי, עם חיתוך של כ-41 ושיפוע של כ-0.4. ‎`fit_line`‎ אינה קוראת ל-‎`linear`‎, ולכן $a_0$ ו-$a_1$ מחושבים ומודפסים נכון. χ², לעומת זאת, משתמש ב-‎`linear(A, x)`‎ בשורה 25, ולכן השאריות גדולות מאוד. גם קו ההתאמה בגרף היה עובר רחוק מהנקודות. χ² מצומצם רחוק מאוד מ-1 הוא סימן לבדוק את הקוד, ולא רק את המדידות.
`````
