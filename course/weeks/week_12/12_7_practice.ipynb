שתי שאלות שמשלבות כמה מהכלים של סעיף 9.7 ושל שבוע 12 יחד: סטטיסטיקה תיאורית, הולכת שגיאות, התאמה ליניארית וטיב התאמה. לכל שאלה יש תא "עבודה" ריק ופתרון מלא בהמשך -- מומלץ לנסות לבד לפני שפותחים את הפתרון.

``````python
import numpy as np
import pandas as pd

g = 9.8
df = pd.read_csv("lab_measurements.csv")
df_clean = df.dropna()

def linear_fit(x, y):
    x_bar, y_bar = x.mean(), y.mean()
    m = np.sum((x - x_bar) * (y - y_bar)) / np.sum((x - x_bar)**2)
    b = y_bar - m * x_bar
    return m, b
``````

`````{admonition} פתרון
angle2 = 60.0
sub2 = df_clean[df_clean["angle_deg"] == angle2]
x2 = sub2["run_id"].to_numpy(dtype=float)
y2 = sub2["range_measured"].to_numpy(dtype=float)
sigma2 = y2.std(ddof=1)

m2, b2 = linear_fit(x2, y2)
resid2 = y2 - (m2*x2 + b2)
chi2_2 = np.sum((resid2/sigma2)**2)
reduced_chi2_2 = chi2_2 / (len(x2) - 2)

print(f"שיפוע: {m2:.3f}")
print(f"chi^2_nu: {reduced_chi2_2:.3f}")
```
פרשנות: כאן ‎`sigma_y`‎ הוא הפיזור של המדידות סביב הממוצע שלהן, ו-‎`chi^2_nu`‎ יוצא קרוב ל-1. כלומר, השאריות סביב הקו גדולות בערך כמו הפיזור סביב הממוצע: הקו אינו מסביר חלק ניכר מהפיזור. אין עדות למגמה שיטתית עם ‎`run_id`‎, וההתפזרות מוסברת על ידי רעש מדידה. מגמה אמיתית הייתה מקטינה את השאריות סביב הקו, ו-‎`chi^2_nu`‎ היה יוצא קטן בהרבה מ-1.
`````
