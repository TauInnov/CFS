def ranges(v0, angles_deg, g):
    """Range (m) at each angle in angles_deg (degrees); v0 in m/s."""

### משימה 2: חיזוי הפלט

**בלי להריץ**: מה ידפיסו שורות 3, 4 ו-5? כתבו כל שורה כפי ש-NumPy מדפיס אותה.

``````python
t = np.array([0.0, 0.5, 1.0, 1.5, 2.0])   # time (s)
y = 10.0 * t - 5.0 * t**2                  # height (m), with g = 10
print(y)
print(t[y >= 3])
print(np.sum(y > 0))
``````

### משימה 3: מצאו את השגיאות

מישהו כתב גרסה משלו ל-‎`heights`‎ ממשימה 1. הקריאה בשורה 5 צריכה להדפיס ‎`[5.]`‎ (בזריקה אנכית: $10^2/20 = 5$ מטר), אבל היא מדפיסה ‎`[399.61501726]`‎. קראו את הקוד לפני שאתם מריצים אותו: יש בו שתי שגיאות. מצאו אותן, תקנו אותן, והסבירו כל אחת במשפט אחד.

``````python
def heights_v2(v0, angles_deg, g):
    """Maximum height (m) at each angle (degrees); v0 in m/s."""
    return (v0 * np.sin(angles_deg))**2 / 2 * g

print(heights_v2(10.0, np.array([90.0]), 10.0))
``````

2. **קדימות פעולות:** ‎`/ 2 * g`‎ מחלק ב-2 ואז **כופל** ב-‎`g`‎. צריך ‎`/ (2 * g)`‎.
