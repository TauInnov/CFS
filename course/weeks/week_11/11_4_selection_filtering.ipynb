## בחירה, סינון ומיון: עמודות, שורות, תנאים בוליאניים ו-sort_values

```{note}
בעמוד הזה בוחרים עמודה ב-‎`df["col"]`‎ ושורות ב-‎`df[mask]`‎, התבנית שתפגשו הכי הרבה בקוד pandas אמיתי. בדפים אחרים של השבוע מופיע גם ‎`df.loc[...]`‎: הוא בוחר שורה לפי **התווית** שלה, ואפשר לתת לו גם שם עמודה, למשל ‎`df.loc[3, "angle_deg"]`‎. לעומתו, ‎`df.iloc[...]`‎ (בהמשך העמוד) בוחר לפי **מיקום**.
```

- **עמודה:** ‎`df["col"]`‎.
- **שורות לפי תנאי (בדיוק כמו מסכות בוליאניות על מערך, מסעיף 7.7):** ‎`df[df["col"] > 5]`‎.
- **שורות ועמודות לפי מיקום:** ‎`df.iloc[0]`‎, ‎`df.iloc[:3, 3]`‎.
- **מיון:** ‎`df.sort_values("col")`‎ (או ‎`ascending=False`‎ לסדר יורד).
- **ערכים חסרים:** ‎`df.isna()`‎ ו-‎`df.dropna()`‎.

``````python
import pandas as pd
df = pd.read_csv("launch_log.csv")
``````

### דוגמה: זריקות בזווית ‎45°‎ בלבד, ממוינות לפי טווח

``````python
mask_45 = df["angle_deg"] == 45
runs_45 = df[mask_45]
runs_45_sorted = runs_45.sort_values("range_measured", ascending=False)
runs_45_sorted
``````

‎`df["angle_deg"] == 45`‎ בעצמו הוא ‎`Series`‎ של ‎`True`‎/‎`False`‎ — בדיוק כמו מסכה על מערך NumPy. ‎`df[mask]`‎ משאיר רק את השורות שבהן המסכה ‎`True`‎. אין הבדל מושגי מהמסכות שראינו בסעיף 7.7 — רק שכאן המסכה בוחרת **שורות שלמות**, לא איברים בודדים.

### באג נפוץ: תנאים מחוברים בלי סוגריים

בפייתון רגיל, ‎`a == 45 and b > 30`‎ עובד. ב-pandas, כשמחברים שני תנאים בוליאניים על ‎`Series`‎, **חובה** להשתמש ב-‎`&`‎/‎`|`‎ (לא ‎`and`‎/‎`or`‎), **וחובה** לעטוף כל תנאי בסוגריים — אחרת סדר הפעולות (קדימויות) משתבש, בדיוק כמו במסכות על מערכים בסעיף 7.7.

בתא הבא השורה השגויה עטופה ב-‎`try`‎/‎`except`‎: כאשר היא נכשלת, ‎`except`‎ תופס את השגיאה ומדפיס את סוגה ואת ההודעה שלה, כך שהמחברת ממשיכה לרוץ.

``````python
try:
    bad = df[df["angle_deg"] == 45 & df["range_measured"] > 30]   # חסרים סוגריים!
except Exception as e:
    print(type(e).__name__, "-", e)
``````

‎`&`‎ בעל קדימות גבוהה יותר מ-‎`==`‎ בפייתון, אז השורה הזו מנסה לבצע קודם ‎`45 & df["range_measured"]`‎ — שגיאה. הגרסה הנכונה:

``````python
good = df[(df["angle_deg"] == 45) & (df["range_measured"] > 30)]
good
``````

### בחירה לפי מיקום: iloc

‎`df.iloc[i, j]`‎ בוחר לפי **מקום** בטבלה, כמו אינדקס של מערך NumPy: מספר השורה ומספר העמודה, החל מ-0. ‎`df.iloc[0]`‎ היא השורה הראשונה, ו-‎`df.iloc[:3, 3]`‎ הן שלוש השורות הראשונות של העמודה הרביעית, ‎`range_measured`‎. אחרי סינון, ‎`iloc[0]`‎ היא השורה הראשונה שנשארה, גם אם ה-‎`index`‎ שלה אינו 0.

``````python
print(df.iloc[0])        # first row: one value per column
print()
print(df.iloc[:3, 3])    # rows 0 to 2 of column 3 (range_measured)
``````

### ערכים חסרים: NaN ו-dropna

כשמדידה נכשלת, התא שלה ביומן נשאר ריק. ‎`read_csv`‎ קורא תא ריק כ-‎`NaN`‎ (Not a Number): ערך מיוחד מסוג ‎`float`‎ שפירושו "אין כאן מדידה". בדוגמה הבאה יומן קצר של ארבע זריקות, שנכתב ישירות בקוד כמו ‎`df_bad`‎ בסעיף 11.3. בזריקה 2 חסר הטווח:

``````python
import io

# run 2 failed, so its range field is empty
log = io.StringIO("run_id,range_measured\n1,39.5\n2,\n3,40.0\n4,40.5\n")
runs = pd.read_csv(log)
runs
``````

‎`NaN`‎ מופיע במקום הערך החסר. שני כלים מטפלים בו:

- **‎`runs.isna()`‎** מחזיר טבלה של ‎`True`‎/‎`False`‎, עם ‎`True`‎ בכל תא חסר. כמו במסכה על מערך, ‎`runs.isna().sum()`‎ סופר את ערכי ה-‎`True`‎ בכל עמודה, כלומר את הערכים החסרים.
- **‎`runs.dropna()`‎** מחזיר טבלה חדשה בלי השורות שיש בהן ‎`NaN`‎. כמו ‎`sort_values`‎, הוא אינו משנה את ‎`runs`‎ עצמו, ולכן שומרים את התוצאה במשתנה.

ערך חסר משפיע גם על הממוצע: pandas מדלג על ‎`NaN`‎ ב-‎`mean()`‎ וב-‎`sum()`‎ בלי אזהרה. לכן הממוצע כאן מחושב על שלוש מדידות בלבד, אף שבטבלה ארבע שורות:

``````python
print(runs.isna().sum())
print(runs["range_measured"].mean())   # NaN is skipped: 3 values

runs_clean = runs.dropna()
print(len(runs), len(runs_clean))
``````
