# שבוע 9 — צירים: מערכים רב-ממדיים

השבוע עוברים ממערך עם ציר אחד למערך עם כמה צירים. הדוגמה ממשיכה את מסלול הזריקה משבועות 7 ו-8, הפעם עם **כמה ריצות של אותה זריקה**: מערך בצורה (ריצה, זמן, קואורדינטה). את העבודה עם מערכים כאלה מארגנים ארבעה כללים, שמוצגים בתחילת סעיף 9.1.

כמו בשבוע 7, הכלים של השבוע הם תרגום של מבנים מוכרים מהחצי הראשון:

1. **לולאות מקוננות** ← מערך רב-ממדי ו-broadcasting: כל ציר הוא גודל פיזיקלי, כמו ריצה, זמן או קואורדינטה (9.1–9.5).
2. **מצבר בלולאה** ← צמצום לפי ציר: ממוצע, סכום או מקסימום לאורך ציר אחד (9.6).

בסוף השבוע, בסעיף 9.7, שני הרעיונות נפגשים במדידות חוזרות: מערך בצורה (זווית, חזרה), וממוצע ושגיאת ממוצע לאורך ציר החזרות.

בסוף השבוע תדעו:
- לקרוא ולפרש ‎`shape`‎ של מערך רב-ממדי לפי המשמעות הפיזיקלית של כל ציר
- לפרוס מערכים לפי צירים (‎`a[0]`‎, ‎`a[:, 2]`‎)
- להבדיל בין ‎`reshape`‎ (סידור מחדש) ל-‎`transpose`‎ (החלפת צירים)
- לחבר מערכים עם ‎`concatenate`‎
- להתאים צירים בין מערכים בעזרת broadcasting ו-‎`None`‎
- לצמצם מערך לפי ציר נבחר (‎`mean`‎, ‎`std`‎, ‎`sum`‎, ‎`max`‎, ‎`argmax`‎)
- לחשב ממוצע, סטיית תקן ושגיאת ממוצע של מדידות חוזרות
