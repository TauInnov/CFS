## np.array ו-concatenate — לאחד ריצות

בסעיף 9.1 בנינו את ‎`Y`‎ בלולאה מפורשת. בפועל, כשיש כבר כמה מערכים חד-ממדיים נפרדים (למשל, ריצה שנמדדה בנפרד בכל פעם), משלבים אותם באחת משתי דרכים:

- **‎`np.array([a1, a2, ...])`‎** — כמו שהוא הופך רשימה של מספרים למערך, ‎`np.array`‎ הופך רשימה של מערכים באותה צורה למערך עם **ציר חדש**: כל מערך נעשה שורה. אם כל ‎`a_i`‎ בצורה ‎`(60,)`‎, ו-5 מהם, מתקבל ‎`(5, 60)`‎.
- **‎`np.concatenate([a1, a2, ...])`‎** — מחבר לאורך ציר **קיים**, בלי ליצור ציר חדש. שני מערכים בצורה ‎`(60,)`‎ נותנים מערך אחד בצורה ‎`(120,)`‎.

ההבדל הזה בדיוק: ‎`np.array`‎ מוסיף ממד, ‎`concatenate`‎ לא.

``````python
import numpy as np

g = 9.8
theta = np.radians(45)
v0_nom = 20.0
t_flight_nom = 2 * v0_nom * np.sin(theta) / g
t = np.linspace(0, t_flight_nom, 60)   # the same time grid for every run

def height(v0, t):
    """Height (m) of one run with launch speed v0 (m/s), at the times t (s)."""
    return v0 * np.sin(theta) * t - 0.5 * g * t**2
``````

### דוגמה: איחוד ריצות נפרדות למערך אחד

נניח שכל ריצה נמדדה בנפרד (כמו קובץ נתונים משלה), וקיבלנו 5 מערכי גובה נפרדים. ‎`np.array`‎ מאחד אותם למערך "ריצה × זמן" אחד.

``````python
v0_runs = np.array([18.0, 19.5, 20.0, 20.5, 22.0])   # launch speeds (m/s)

y0 = height(v0_runs[0], t)
y1 = height(v0_runs[1], t)
y2 = height(v0_runs[2], t)
y3 = height(v0_runs[3], t)
y4 = height(v0_runs[4], t)

Y = np.array([y0, y1, y2, y3, y4])
print("Y.shape:", Y.shape)   # (5, 60)
``````

### דוגמה: הוספת ריצה חדשה עם ‎`concatenate`‎

בוצעה ריצה שישית. במקום לבנות הכול מחדש, מוסיפים אותה לאורך ציר הריצות הקיים.

``````python
v0_new = 21.0                 # the launch speed of the sixth run (m/s)
y_new = height(v0_new, t)

# y_new[None, :] has shape (1, 60); None is explained in 9.6
Y_extended = np.concatenate([Y, y_new[None, :]])
print("Y_extended.shape:", Y_extended.shape)       # (6, 60)
``````

### באג נפוץ: ‎`np.array`‎ במקום ‎`concatenate`‎ (או להפך)

‎`np.array([Y, y_new])`‎ **לא** יעבוד ישירות כאן — ‎`Y`‎ בצורה ‎`(5, 60)`‎ ו-‎`y_new`‎ בצורה ‎`(60,)`‎ אינם באותה צורה, וזו בדיוק הדרישה של ‎`np.array`‎. השגיאה כאן **כן** מופיעה (‎`ValueError`‎), בניגוד לבאגי הצירים השקטים שראינו — אבל כדאי לזהות אותה מייד: הפתרון הוא ‎`concatenate`‎ עם ‎`[None, :]`‎, לא ‎`np.array`‎.

### נסו בעצמכם

לכל ריצה יש שתי קואורדינטות, $x$ ו-$y$. הקוד בתא הבא שם אותן כשתי שורות עם ‎`np.array`‎, ואז הופך את השורות לעמודות עם ‎`.T`‎ (סעיף 9.3).

**לפני שמריצים:** כתבו מה תדפיס שורת ההדפסה. כמה שורות יהיו, וכמה מספרים בכל שורה?

``````python
x = np.array([0.0, 1.0, 2.0])   # x (m) at three times
y = np.array([0.0, 1.5, 2.0])   # y (m) at the same times
print(np.array([x, y]).T)
``````

`````{admonition} פתרון
:class: dropdown, tip
```text
[[0.  0. ]
 [1.  1.5]
 [2.  2. ]]
```
שלוש שורות, אחת לכל זמן, ובכל שורה שני מספרים: ה-$x$ וה-$y$ של אותו זמן. ‎`np.array([x, y])`‎ לבדו נותן את $x$ ואת $y$ כשתי שורות, בצורה ‎`(2, 3)`‎. ‎`.T`‎ מחליף שורות ועמודות, ולכן הצורה היא ‎`(3, 2)`‎, וציר הקואורדינטה נמצא בסוף.
`````

### תרגול עצמי

בתא הבא ‎`Y2`‎ מכיל שתי ריצות, ו-‎`c`‎ ריצה שלישית.

**לפני שמריצים:** מה תדפיס השורה הראשונה? ומה יקרה בשורה האחרונה?

``````python
a = np.array([0.0, 1.5, 2.0])
b = np.array([0.0, 2.5, 4.0])
Y2 = np.array([a, b])                 # two runs: shape (2, 3)
c = np.array([0.0, 3.5, 6.0])         # a third run
print(np.concatenate([Y2, c[None, :]]))
print(np.array([Y2, c]))
``````

`````{admonition} פתרון
:class: dropdown, tip
השורה הראשונה מדפיסה שלוש ריצות, אחת בכל שורה:
```text
[[0.  1.5 2. ]
 [0.  2.5 4. ]
 [0.  3.5 6. ]]
```
‎`c[None, :]`‎ הופך את ‎`c`‎ לצורה ‎`(1, 3)`‎, ו-‎`concatenate`‎ מוסיף אותו לאורך ציר הריצות הקיים. השורה האחרונה נעצרת בשגיאה:
```text

```
‎`np.array`‎ דורש מערכים באותה צורה, ו-‎`Y2`‎ בצורה ‎`(2, 3)`‎, בעוד ‎`c`‎ בצורה ‎`(3,)`‎.
`````
