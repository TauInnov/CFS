### פריט 1: חיזוי הפלט

כדור נזרק ישר למעלה פעמיים, במהירויות שונות. **בלי להריץ**, כתבו מה תדפיס כל אחת משלוש שורות ההדפסה.

``````python
import numpy as np

v0_runs = np.array([10.0, 20.0])   # launch speed of each throw (m/s)
t = np.array([0.0, 1.0, 2.0])      # times (s)
g = 10.0                           # m/s^2

Y = v0_runs[:, None] * t - 0.5 * g * t**2   # height (m): (throw, time)
print(Y)
print(Y.mean(axis=0))
print(Y.max(axis=1))
``````

- ‎`Y`‎ בצורה ‎`(2, 3)`‎: שורה לכל זריקה ועמודה לכל זמן. בזריקה הראשונה הגובה בזמן 1 הוא ‎`10 - 5 = 5`‎, ובזמן 2 הוא ‎`20 - 20 = 0`‎.
- ‎`mean(axis=0)`‎ מעלים את ציר הזריקות: ממוצע לכל זמן, למשל ‎`(0 + 20) / 2 = 10`‎ בזמן 2.
- ‎`max(axis=1)`‎ מעלים את ציר הזמן: הגובה המרבי של כל זריקה.

### פריט 3: באגים בלולאת עדכון

הפונקציה ‎`simulate_height`‎ אמורה לחשב, צעד אחרי צעד, את המהירות והגובה של כדור שנזרק ישר למעלה, כמו הלולאה של סעיף 9.10. הפונקציה ‎`landing_index`‎ אמורה להחזיר את הצעד הראשון שבו הכדור מתחת לקרקע, או ‎`-1`‎ אם אין צעד כזה. בקוד יש ארבעה באגים: שלושה ב-‎`simulate_height`‎ ואחד ב-‎`landing_index`‎.

הקוד, כפי שקיבלתם אותו:

``````python
import numpy as np

def simulate_height(v0, g, dt, n_steps):
    """Velocity (m/s) and height (m) of a ball thrown up, step by step."""
    v = np.zeros(n_steps)
    y = np.zeros(n_steps)
    for i in range(n_steps):
        v[i + 1] = v[i] - g * dt
        y[i + 1] = y[i + 1] + v[i] * dt
    return v, y

def landing_index(y):
    """First step with the ball below the ground, or -1 if there is none."""
    for i in range(len(y)):
        if y[i] <= 0:
            return i
    return -1

v, y = simulate_height(10.0, 10.0, 0.1, 4)
print(v)
print(y)
print(landing_index(np.array([0.0, 1.0, 0.5, -0.5])))
``````

**א.** לפני שמחפשים באגים: מה **צריכה** להחזיר ‎`simulate_height(10.0, 10.0, 0.1, 4)`‎? מלאו טבלה של ‎`i`‎, ‎`v[i]`‎ ו-‎`y[i]`‎, כמו בסעיף 9.10.

**ב.** הריצה הראשונה נעצרת בשגיאה:

```text
IndexError: index 4 is out of bounds for axis 0 with size 4
```

השגיאה מצביעה על שורה 8. באיזה ערך של ‎`i`‎ היא קרתה? באיזו שורה הבאג, ומה התיקון?

**ג.** אחרי התיקון, הריצה השנייה מדפיסה:

```text
[ 0. -1. -2. -3.]
[ 0.   0.  -0.1 -0.2]
0
```

השוו לטבלה מסעיף א. מהו הערך הראשון ששונה, ומה התיקון?

**ד.** אחרי התיקון השני, הריצה השלישית מדפיסה:

```text

[0.  1.  0.9 0.8]

```

המהירויות נכונות עכשיו. איזו שורה מחשבת את הגובה לא נכון, ומה התיקון?

**ה.** ‎`landing_index`‎ מחזירה 0, אף שהכדור עובר מתחת לקרקע רק בצעד 3. מצאו את הבאג ותקנו אותו. מה תחזיר הפונקציה המתוקנת על הגבהים מסעיף א?

לכל תיקון, הסבירו במשפט אחד מה היה לא נכון.

def simulate_height(v0, g, dt, n_steps):
    """Velocity (m/s) and height (m) of a ball thrown up, step by step."""

    y = np.zeros(n_steps)
    v[0] = v0
    for i in range(n_steps - 1):
        v[i + 1] = v[i] - g * dt
        y[i + 1] = y[i] + v[i] * dt
    return v, y

def landing_index(y):
    """First step with the ball below the ground, or -1 if there is none."""
    for i in range(len(y)):
        if y[i] < 0:
            return i
    return -1

v, y = simulate_height(10.0, 10.0, 0.1, 4)
print(v)                                                 # [10.  9.  8.  7.]
print(y)                                                 # [0.  1.  1.9 2.7]
print(landing_index(np.array([0.0, 1.0, 0.5, -0.5])))    # 3
print(landing_index(y))                                  # -1
