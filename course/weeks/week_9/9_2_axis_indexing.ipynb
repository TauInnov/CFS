```````python
import numpy as np

g = 9.8
theta = np.radians(45)
v0_nom = 20.0                       # nominal launch speed (m/s)
# measured launch speed of each run (m/s):
v0_runs = np.array([18.0, 19.5, 20.0, 20.5, 22.0])
n_runs, n_t = 5, 60
t_flight_nom = 2 * v0_nom * np.sin(theta) / g
t = np.linspace(0, t_flight_nom, n_t)   # the same time grid for all runs

Y = np.zeros((n_runs, n_t))
X = np.zeros((n_runs, n_t))
for i in range(n_runs):
    Y[i] = v0_runs[i] * np.sin(theta) * t - 0.5 * g * t**2
    X[i] = v0_runs[i] * np.cos(theta) * t
XY = np.zeros((n_runs, n_t, 2))   # run, time, coordinate
XY[:, :, 0] = X                    # coordinate 0: x
XY[:, :, 1] = Y                    # coordinate 1: y
```````

### דוגמה

```````python
run2_full_trajectory = Y[1]          # the second run (index 1), at every time
all_runs_at_t2 = Y[:, 2]             # every run, at the third time (index 2)
all_x_coords = XY[:, :, 0]           # x only, of every run at every time

print("run2_full_trajectory.shape:", run2_full_trajectory.shape)
print("all_runs_at_t2.shape:", all_runs_at_t2.shape)
print("all_x_coords.shape:", all_x_coords.shape)
```````

### באג נפוץ: ‎`a[0]`‎ מול ‎`a[:, 0]`‎

‎`Y[0]`‎ ו-‎`Y[:, 0]`‎ נראים דומים אך שונים לגמרי פיזיקלית: ‎`Y[0]`‎ הוא **כל מסלול הריצה הראשונה**, בעוד ‎`Y[:, 0]`‎ הוא **הגובה ההתחלתי של כל הריצות** (שכולן מתחילות מ-$y=0$). הקוד רץ בלי שגיאה בשני המקרים, אבל מחזיר משהו אחר לגמרי.
