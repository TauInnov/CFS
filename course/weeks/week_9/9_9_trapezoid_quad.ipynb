## 9.9 — אינטגרציה: trapezoid (מדידות) ו-quad (פונקציה אנליטית)

(המשך הסרטון מ-9.8.) `cumsum` נתן קירוב גס לאינטגרל. ל-SciPy יש שני כלים מדויקים יותר, לשני מצבים שונים:

- **`scipy.integrate.trapezoid(y, x)`** — מקבל **מדידות בדידות** ($y$ בנקודות $x$), ומחשב את השטח שמתחתן בשיטת הטרפזים. מתאים לנתוני ניסוי.
- **`scipy.integrate.quad(f, a, b)`** — מקבל **פונקציה אנליטית** (callable) וגבולות אינטגרציה, ומחשב את האינטגרל שלה בשיטה נומרית מדויקת בהרבה. מתאים כשיש נוסחה סגורה.

הכלל המעשי מהצעת הקורס: קריאה ל-`trapezoid` או ל-`quad` היא שימוש בכלי מוכן — לא שיטה נומרית שלומדים לבנות (זה בקורס ההמשך).

``````python
import numpy as np
from scipy import integrate

g = 9.8
theta = np.radians(45)
v0 = 20.0
``````

### דוגמה: טווח מתוך "מדידות" מהירות, עם `trapezoid`

יש לנו מדגם של $v_x(t)$ (בדידות, כמו נתוני ניסוי), ורוצים למצוא כמה מרחק הצטבר — האינטגרל של המהירות על הזמן.

``````python
t_flight = 2 * v0 * np.sin(theta) / g
t_samples = np.linspace(0, t_flight, 25)     # מדגם דליל יחסית, כמו במדידה אמיתית
vx_samples = v0 * np.cos(theta) * np.ones_like(t_samples)

R_trapz = integrate.trapezoid(vx_samples, t_samples)
R_theory = v0**2 * np.sin(2 * theta) / g

print("trapezoid:", R_trapz)
print("תיאוריה:  ", R_theory)
``````

### דוגמה: אותו טווח, עם `quad` על הנוסחה האנליטית

הפעם יש לנו את $v_x(t)=v_0\cos\theta$ כפונקציה ממש (לא מדגם), אז אפשר להעביר אותה ישירות ל-`quad`.

``````python
def vx_func(t):
    return v0 * np.cos(theta)

R_quad, err_estimate = integrate.quad(vx_func, 0, t_flight)
print("quad:", R_quad, "  (שגיאת קירוב משוערת:", err_estimate, ")")
``````

`quad` מחזיר **זוג**: הערך המשוער, והערכת שגיאת הקירוב שלו — כדאי תמיד לזרוק עין על הרכיב השני, גם בלי לנתח אותו לעומק.
