## צמצום לפי ציר: mean, std, sum, max, argmax (כלל 2)

הכלל השני מארבעת הכללים:

```{admonition} כלל 2 — axis הוא הציר שנעלם
:class: tip
‎`axis=k`‎ בפעולת צמצום (‎`mean`‎, ‎`std`‎, ‎`sum`‎, ‎`max`‎, ‎`argmax`‎...) הוא הציר **שנעלם** מהתוצאה. ‎`mean(axis=0)`‎ על מערך ‎`(5, 60)`‎ מחזיר ‎`(60,)`‎ — ממוצע **על פני הריצות**, ציר הזמן נשאר.
```

זה בדיוק הכלי שמאפשר לענות על שאלות כמו "מה הגובה הממוצע בכל רגע?" לעומת "מה הגובה המרבי שכל ריצה הגיעה אליו?" — אותו מערך, שני צירי צמצום שונים.

``````python
import numpy as np

g = 9.8
theta = np.radians(45)
v0_nom = 20.0
v0_runs = np.array([18.0, 19.5, 20.0, 20.5, 22.0])   # launch speeds (m/s)
n_runs, n_t = 5, 60
t_flight_nom = 2 * v0_nom * np.sin(theta) / g
t = np.linspace(0, t_flight_nom, n_t)

Y = v0_runs[:, None] * np.sin(theta) * t - 0.5 * g * t**2   # shape (5, 60)
``````

### דוגמה: ממוצע וסטיית תקן על פני הריצות

הגובה הממוצע בכל נקודת זמן על פני 5 הריצות, והפיזור סביבו, מחושבים עם ‎`axis=0`‎ (ציר הריצות נעלם, ציר הזמן נשאר). זו גם ההכנה לסעיף 9.7, שעוסק במדידות חוזרות.

סטיית התקן (‎`std`‎) מודדת כמה הערכים מתפזרים סביב הממוצע שלהם: בערך, המרחק הטיפוסי של ערך מהממוצע. בסעיף 9.7 נראה כיצד מחשבים אותה, מתי משתמשים בה, ומדוע במדידות חוזרות מוסיפים ל-‎`std`‎ את הפרמטר ‎`ddof=1`‎, שאינו מופיע בתא שלמטה. בגרף שמתחת לתא: הממוצע בקו מלא, והממוצע פחות סטיית תקן ועוד סטיית תקן בקווים מקווקווים.

``````python
y_mean = Y.mean(axis=0)   # shape (60,): the mean over the runs, at every time
y_std = Y.std(axis=0)     # shape (60,): the spread over the runs
print(y_mean.shape, y_std.shape)
``````

``````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot(t, y_mean, color="tab:blue", label="Mean")
ax.plot(t, y_mean - y_std, color="tab:blue", linestyle="--",
        label="Mean ± standard deviation")
ax.plot(t, y_mean + y_std, color="tab:blue", linestyle="--")
ax.set_xlabel("t (s)")
ax.set_ylabel("y (m)")
ax.legend()
plt.show()
``````

### באג נפוץ: לשכוח ‎`axis`‎

בלי ‎`axis=...`‎, פעולת הצמצום פועלת על **כל המערך יחד**, ומחזירה מספר בודד — לא וקטור. ‎`Y.mean()`‎ (בלי axis) מחזיר גובה ממוצע אחד, שמערבב רגעים ליד הקרקע, בהתחלה ובסוף, עם רגעים ליד השיא, ומערבב גם בין הריצות — תוצאה בלי שום פירוש פיזיקלי ברור.

``````python
scalar_mean = Y.mean()   # all 300 values together: rarely what you want
print(scalar_mean, type(scalar_mean))
``````

``````python
Y2 = np.array([[0.0, 4.0, 6.0],
               [0.0, 2.0, 2.0]])   # (run, time)
print(Y2.sum(axis=1))
print(Y2.mean(axis=0))
``````

`````{admonition} פתרון
- ‎`sum(axis=1)`‎ מעלים את ציר הזמן: סכום אחד לכל ריצה, ‎`0 + 4 + 6 = 10`‎ ו-‎`0 + 2 + 2 = 4`‎.
- ‎`mean(axis=0)`‎ מעלים את ציר הריצות: ממוצע אחד לכל זמן, למשל ‎`(4 + 2) / 2 = 3`‎.
`````
