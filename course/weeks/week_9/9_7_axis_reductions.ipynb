## צמצום לפי ציר: mean, std, sum, argmin, argmax (כלל 2)

הכלל השני מארבעת הכללים:

```{admonition} כלל 2 — axis הוא הציר שנעלם
:class: tip
‎`axis=k`‎ בפעולת צמצום (‎`mean`‎, ‎`std`‎, ‎`sum`‎, ‎`argmin`‎, ‎`argmax`‎...) הוא הציר **שנעלם** מהתוצאה. ‎`mean(axis=0)`‎ על מערך ‎`(5, 200)`‎ מחזיר ‎`(200,)`‎ — ממוצע **על פני הריצות**, ציר הזמן נשאר.
```

זה בדיוק הכלי שמאפשר לענות על שאלות כמו "מה הגובה הממוצע בכל רגע?" לעומת "מה הגובה המרבי שכל ריצה הגיעה אליו?" — אותו מערך, שני צירי צמצום שונים.

``````python
import numpy as np

g = 9.8
theta = np.radians(45)
v0_nom = 20.0
v0_runs = np.array([18.0, 19.5, 20.0, 20.5, 22.0])   # launch speeds (m/s)
n_runs, n_t = 5, 60
t_flight_nom = 2 * v0_nom * np.sin(theta) / g
t = np.linspace(0, t_flight_nom, n_t)

Y = v0_runs[:, None] * np.sin(theta) * t - 0.5 * g * t**2   # shape (5, 60)
``````

### דוגמה: ממוצע וסטיית תקן על פני הריצות

לכל נקודת זמן, מה הגובה הממוצע על פני 5 הריצות, ומה הפיזור סביבו? — זה בדיוק ‎`axis=0`‎ (ציר הריצות נעלם, ציר הזמן נשאר). זו גם ההכנה הישירה לניתוח שגיאות בשבוע 11.

סטיית התקן (‎`std`‎) מודדת כמה הערכים מתפזרים סביב הממוצע שלהם: בערך, המרחק הטיפוסי של ערך מהממוצע. בסעיף 11.2 נראה איך מחשבים אותה ומתי משתמשים בה. בגרף שמתחת לתא: הממוצע בקו מלא, והממוצע פחות סטיית תקן ועוד סטיית תקן בקווים מקווקווים.

``````python
y_mean = Y.mean(axis=0)   # shape (60,): the mean over the runs, at every time
y_std = Y.std(axis=0)     # shape (60,): the spread over the runs
print(y_mean.shape, y_std.shape)
``````

``````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot(t, y_mean, color="tab:blue", label="ממוצע")
ax.plot(t, y_mean - y_std, color="tab:blue", linestyle="--",
        label="ממוצע ± סטיית תקן")
ax.plot(t, y_mean + y_std, color="tab:blue", linestyle="--")
ax.set_xlabel("t (s)")
ax.set_ylabel("y (m)")
ax.legend()
plt.show()
``````

### דוגמה: הגובה המרבי שכל ריצה הגיעה אליו

השאלה ההפוכה: לכל ריצה בנפרד, מהו הגובה המרבי לאורך הזמן שלה? כאן ציר הזמן הוא זה שנעלם — ‎`axis=1`‎.

``````python
max_height_per_run = Y.max(axis=1)        # shape (5,): one per run
time_index_of_max = np.argmax(Y, axis=1)  # shape (5,): where it happened

print("גובה מרבי בכל ריצה:", max_height_per_run)
print("אינדקס הזמן של השיא בכל ריצה:", time_index_of_max)
``````

### באג נפוץ: לשכוח ‎`axis`‎

בלי ‎`axis=...`‎, פעולת הצמצום פועלת על **כל המערך יחד**, ומחזירה מספר בודד — לא וקטור. ‎`Y.mean()`‎ (בלי axis) מחזיר גובה ממוצע אחד, שמערבב זמנים מוקדמים (קרובים לקרקע) עם זמנים מאוחרים — תוצאה בלי שום פירוש פיזיקלי ברור.

``````python
scalar_mean = Y.mean()   # all 300 values together: rarely what you want
print(scalar_mean, type(scalar_mean))
``````

### נסו בעצמכם

בתא הבא יש שתי ריצות קצרות, עם 3 נקודות זמן בכל אחת.

**לפני שמריצים:** כתבו מה תדפיס כל אחת משתי שורות ההדפסה. איזה ציר נעלם בכל אחת?

``````python
Y2 = np.array([[0.0, 4.0, 6.0],
               [0.0, 2.0, 2.0]])   # (run, time)
print(Y2.sum(axis=1))
print(Y2.mean(axis=0))
``````

`````{admonition} פתרון
:class: dropdown, tip
```text
[10.  4.]
[0. 3. 4.]
```
- ‎`sum(axis=1)`‎ מעלים את ציר הזמן: סכום אחד לכל ריצה, ‎`0 + 4 + 6 = 10`‎ ו-‎`0 + 2 + 2 = 4`‎.
- ‎`mean(axis=0)`‎ מעלים את ציר הריצות: ממוצע אחד לכל זמן, למשל ‎`(4 + 2) / 2 = 3`‎.
`````
