# שבוע 13 — בקרת גרסאות עם Git
בשבוע האחרון של הקורס נלמד כלי שהוא לא עוד ספרייה כמו NumPy או pandas, אלא הרגל עבודה: **Git**, מערכת בקרת הגרסאות הנפוצה בעולם.
כל פיזיקאי שעובד עם קוד מכיר את הבעיות: קבצים בשמות כמו ‎`sim_v2_final_REAL.py`‎, שינוי "קטן" שמשבש קוד עובד בלי אפשרות לחזור אחורה, או שני שותפי מעבדה שמוחקים זה את שינויי זה.
Git פותר את זה: הוא שומר **היסטוריה** מלאה של הקוד שלכם, תמונת מצב אחרי תמונת מצב, כל אחת מתועדת ובת-שחזור.
שני רעיונות מהחצי הראשון של הקורס עוזרים כאן: **רשימת היסטוריה** (שבוע 4) הופכת לסדרת commit-ים, שכל אחד מהם הוא תמונת מצב שנוספת לסוף ההיסטוריה; ו**שמות ואובייקטים** (שבוע 5): ‎`HEAD`‎ ושם של branch הם שמות שמצביעים על commit, כמו משתנה שמצביע על אובייקט.
השבוע לא יכסה את Git כולו, אלא את מה שנדרש ברוב העבודה היומיומית של מדען שעובד לבד או בזוג: לשמור היסטוריה, לקרוא אותה, לגבות אותה ל-GitHub, ולפתור קונפליקט מיזוג אחד.
בסוף השבוע תדעו:
- מדוע בקרת גרסאות פותרת בעיה מוכרת, וליצור repo ולשמור היסטוריה עם ‎`git init`‎, ‎`add`‎, ‎`commit`‎
- לקרוא היסטוריה קיימת ולאתר בה באגים עם ‎`git log`‎, ‎`git diff`‎, ‎`git status`‎
- לגבות ולשתף קוד עם GitHub — ‎`remote`‎, ‎`push`‎, ‎`clone`‎, ‎`pull`‎ — ולהחליט מה לא עולה אליו עם ‎`.gitignore`‎
- להסתעף עם branch, ולפתור merge conflict אחד בביטחון
