## 10.4 — בחירה, סינון ומיון: עמודות, שורות, תנאים בוליאניים ו-sort_values

```{note}
ל-pandas יש כלים עשירים יותר לבחירה (`loc`, `iloc`) — מעבר להיקף הקורס הזה. כאן מתמקדים בגישה הבסיסית: `df["col"]` לעמודה, ו-`df[mask]` לשורות — התבנית שתפגשו הכי הרבה בקוד pandas אמיתי.
```

- **עמודה:** `df["col"]`.
- **שורות לפי תנאי (בדיוק כמו מסכות בוליאניות על מערך, מסעיף 7.8):** `df[df["col"] > 5]`.
- **מיון:** `df.sort_values("col")` (או `ascending=False` לסדר יורד).

``````python
import pandas as pd
df = pd.read_csv("launch_log.csv")
``````

### דוגמה: זריקות בזווית 45° בלבד, ממוינות לפי טווח

``````python
mask_45 = df["angle_deg"] == 45
runs_45 = df[mask_45]
runs_45_sorted = runs_45.sort_values("range_measured", ascending=False)
runs_45_sorted
``````

`df["angle_deg"] == 45` בעצמו הוא `Series` של `True`/`False` — בדיוק כמו מסכה על מערך NumPy. `df[mask]` משאיר רק את השורות שבהן המסכה `True`. אין הבדל מושגי מהמסכות שראינו בסעיף 7.8 — רק שכאן המסכה בוחרת **שורות שלמות**, לא איברים בודדים.

### באג נפוץ: תנאים מחוברים בלי סוגריים

בפייתון רגיל, `a == 45 and b > 30` עובד. ב-pandas, כשמחברים שני תנאים בוליאניים על `Series`, **חובה** להשתמש ב-`&`/`|` (לא `and`/`or`), **וחובה** לעטוף כל תנאי בסוגריים — אחרת סדר הפעולות (קדימויות) משתבש, בדיוק כמו הבאג מסעיף 7.10.

``````python
try:
    bad = df[df["angle_deg"] == 45 & df["range_measured"] > 30]   # חסרים סוגריים!
except Exception as e:
    print(type(e).__name__, "-", e)
``````

`&` בעל קדימות גבוהה יותר מ-`==` בפייתון, אז השורה הזו מנסה לבצע קודם `45 & df["range_measured"]` — שגיאה. הגרסה הנכונה:

``````python
good = df[(df["angle_deg"] == 45) & (df["range_measured"] > 30)]
good
``````
