## 10.6 — describe

`df.describe()` הוא כלי הבדיקה המהיר ביותר לטבלה שלמה: הוא מחשב, לכל עמודה **מספרית**, שמונה סטטיסטיקות בבת אחת — `count`, `mean`, `std`, `min`, שלושה אחוזונים (`25%`, `50%`, `75%`), ו-`max`.

``````python
import pandas as pd
df = pd.read_csv("launch_log.csv")
``````

### דוגמה: תמונת מצב מהירה על כל הטבלה

``````python
df.describe()
``````

שימו לב ש-`run_id` מופיע ב-`describe()` למרות שהוא לא באמת גודל פיזיקלי — הוא מספרי (`int64`), אז `describe()` כולל אותו אוטומטית. חשוב לזכור **לא** לפרש את הממוצע/סטיית התקן שלו כאילו יש להם משמעות פיזיקלית.

### דוגמה: describe אחרי groupby

אפשר לשלב את שני הכלים: `describe()` בתוך כל קבוצה בנפרד, כדי לראות בבת אחת את הפיזור בכל זווית.

``````python
df.groupby("angle_deg")["range_measured"].describe()
``````

### באג נפוץ: `describe()` "שוכח" עמודת טקסט

אם לעמודה יש טיפוס טקסטואלי (`object`/`str`, למשל בגלל הבאג שראינו בסעיף 10.3 — טקסט חורג שהחליק לתוך עמודה מספרית), `describe()` **פשוט לא מציג אותה** בפלט הרגיל, בלי אזהרה. קל לפספס עמודה שלמה בלי לשים לב.

``````python
import io
bad_csv = io.StringIO("angle_deg,v0_measured\n15,20.3\n15,19.0?\n15,20.8\n")
df_bad = pd.read_csv(bad_csv)
print(df_bad.describe())   # v0_measured נעלמה מהפלט - כי היא טקסטואלית, לא מספרית!
``````
