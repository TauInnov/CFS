### פריט 1: באגים בלולאת עדכון

הפונקציה ‎`simulate_height`‎ אמורה לחשב, צעד אחרי צעד, את המהירות והגובה של כדור שנזרק ישר למעלה, כמו הלולאה של סעיף 10.1. הפונקציה ‎`landing_index`‎ אמורה להחזיר את הצעד הראשון שבו הכדור מתחת לקרקע, או ‎`-1`‎ אם אין צעד כזה. בקוד יש ארבעה באגים: שלושה ב-‎`simulate_height`‎ ואחד ב-‎`landing_index`‎.

הקוד, כפי שקיבלתם אותו:

``````python
import numpy as np

def simulate_height(v0, g, dt, n_steps):
    """Velocity (m/s) and height (m) of a ball thrown up, step by step."""
    v = np.zeros(n_steps)
    y = np.zeros(n_steps)
    for i in range(n_steps):
        v[i + 1] = v[i] - g * dt
        y[i + 1] = y[i + 1] + v[i] * dt
    return v, y

def landing_index(y):
    """First step with the ball below the ground, or -1 if there is none."""
    for i in range(len(y)):
        if y[i] <= 0:
            return i
    return -1

v, y = simulate_height(10.0, 10.0, 0.1, 4)
print(v)
print(y)
print(landing_index(np.array([0.0, 1.0, 0.5, -0.5])))
``````

**א.** לפני שמחפשים באגים: מה **צריכה** להחזיר ‎`simulate_height(10.0, 10.0, 0.1, 4)`‎? מלאו טבלה של ‎`i`‎, ‎`v[i]`‎ ו-‎`y[i]`‎, כמו בסעיף 10.1.

**ב.** הריצה הראשונה נעצרת בשגיאה:

```text

```

השגיאה מצביעה על שורה 8. באיזה ערך של ‎`i`‎ היא קרתה? באיזו שורה הבאג, ומה התיקון?

**ג.** אחרי התיקון, הריצה השנייה מדפיסה:

השוו לטבלה מסעיף א. מהו הערך הראשון ששונה, ומה התיקון?

**ד.** אחרי התיקון השני, הריצה השלישית מדפיסה:


המהירויות נכונות עכשיו. איזו שורה מחשבת את הגובה לא נכון, ומה התיקון?

**ה.** ‎`landing_index`‎ מחזירה 0, אף שהכדור עובר מתחת לקרקע רק בצעד 3. מצאו את הבאג ותקנו אותו. מה תחזיר הפונקציה המתוקנת על הגבהים מסעיף א?

### פריט 2: שאלה מלאה — מצאו את שלוש הטעויות

גוף נע לאורך קו ישר, ומיקומו ‎`x`‎ נמדד בזמנים ‎`t`‎. הפונקציה ‎`velocity`‎ אמורה להחזיר את המהירות בין כל שתי מדידות סמוכות, והפונקציה ‎`distance_so_far`‎ את המרחק שהגוף עבר מההתחלה ועד כל מדידה. הפלט הצפוי כתוב בהערות של שתי שורות ההדפסה.

בקוד שלוש טעויות. **קראו אותו בלי להריץ**: מצאו כל טעות לפי מספר השורה, תקנו אותה, והסבירו במשפט אחד מה היה לא נכון. ציינו גם איזו טעות מתגלה ראשונה כשמריצים את הקוד.

``````python
import numpy as np

def velocity(t, x):
    """Velocity (m/s) between consecutive measurements."""
    return np.diff(t) / np.diff(x)

def distance_so_far(t, x):
    """Distance (m) traveled from the start up to each measurement."""
    v = velocity(t, x)
    dt = t[1] - t[0]
    steps = abs(v) * dt
    return np.concatenate([0.0, np.cumsum(steps)])

t = np.array([0.0, 1.0, 2.0, 4.0])   # time (s)
x = np.array([0.0, 4.0, 6.0, 8.0])   # position (m)
print(velocity(t, x))          # expected: [4. 2. 1.]
print(distance_so_far(t, x))   # expected: [0. 4. 6. 8.]
``````

`````{admonition} פתרון
:class: dropdown, tip
1. **שורה 5:** המהירות היא הפרש המיקומים חלקי הפרש הזמנים, ולא להפך. התיקון: ‎`return np.diff(x) / np.diff(t)`‎.
2. **שורה 12:** ‎`np.concatenate`‎ מחבר מערכים, ו-‎`0.0`‎ הוא מספר בודד ולא מערך. התיקון: ‎`np.concatenate([[0.0], np.cumsum(steps)])`‎.
3. **שורות 10–11:** הזמנים לא במרווחים שווים (המרווח האחרון הוא 2 שניות), ו-‎`t[1] - t[0]`‎ הוא רק המרווח הראשון. כל צעד צריך את המרווח שלו. התיקון: ‎`steps = abs(v) * np.diff(t)`‎, ואז השורה של ‎`dt`‎ מיותרת.

**הסדר שבו הטעויות מתגלות:** שורת ההדפסה הראשונה מדפיסה ‎`[0.25 0.5  1.  ]`‎ במקום ‎`[4. 2. 1.]`‎; זו טעות 1. השורה השנייה נעצרת בשגיאה של טעות 2:

```text
ValueError: zero-dimensional arrays cannot be concatenated
```

השגיאה הזו מסתירה את טעות 3: רק אחרי שמתקנים את טעויות 1 ו-2 הפלט הוא ‎`[0. 4. 6. 7.]`‎, והמספר האחרון עדיין שגוי.

הקוד המתוקן:

```python
def velocity(t, x):
    """Velocity (m/s) between consecutive measurements."""
    return np.diff(x) / np.diff(t)

def distance_so_far(t, x):
    """Distance (m) traveled from the start up to each measurement."""
    v = velocity(t, x)
    steps = abs(v) * np.diff(t)
    return np.concatenate([[0.0], np.cumsum(steps)])

print(velocity(t, x))          # [4. 2. 1.]
print(distance_so_far(t, x))   # [0. 4. 6. 8.]
`````
