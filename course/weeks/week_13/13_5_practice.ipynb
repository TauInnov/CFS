### פריט 1: שאלת Git מדורגת

קיבלתם מעמית ריפו עם סקריפט קצר, ‎`speed.py`‎, שמחשב את המהירות הממוצעת $v = d/t$ של כדור שעבר 7 מטרים ב-2 שניות. הסקריפט רץ בלי שגיאה, אבל התוצאה שהוא מדפיס חשודה. הפקודה ‎`git log --oneline`‎ מדפיסה:

``````python
# Builds the repository of item 1 in a temporary folder.
# Fixed names and dates keep the commit hashes the same in every build.
import os
import subprocess
import tempfile

os.chdir(tempfile.mkdtemp(prefix="git_item1_"))
for who in ("AUTHOR", "COMMITTER"):
    os.environ[f"GIT_{who}_NAME"] = "Colleague"
    os.environ[f"GIT_{who}_EMAIL"] = "colleague@example.com"

versions = [
    ("d = 7.0   # distance (m)\nt = 2.0   # time (s)\nv = d / t\n"
     'print(f"v = {v} m/s")\n', "חישוב מהירות ממוצעת"),
    ("d = 7   # distance (m)\nt = 2   # time (s)\nv = d // t\n"
     'print(f"v = {v} m/s")\n', "פישוט: הסרת נקודות עשרוניות מיותרות"),
    ("d = 7   # distance (m)\nt = 2   # time (s)\nv = d // t\n"
     'print(f"average speed: v = {v} m/s")\n', "שיפור הודעת ההדפסה"),
]
subprocess.run(["git", "init", "-q", "-b", "main"], check=True)
subprocess.run(["git", "config", "color.ui", "false"], check=True)
for i, (text, message) in enumerate(versions):
    with open("speed.py", "w") as f:
        f.write(text)
    date = f"2026-11-01T10:{20 * i:02d}:00+02:00"
    os.environ["GIT_AUTHOR_DATE"] = date
    os.environ["GIT_COMMITTER_DATE"] = date
    subprocess.run(["git", "add", "speed.py"], check=True)
    subprocess.run(["git", "commit", "-q", "-m", message], check=True)

hashes = !git log --format=%h
first, second = hashes[-1], hashes[-2]
!git log --oneline
``````

הפקודה ‎`git diff 1530822 d01b58b -- speed.py`‎, שמשווה את שני ה-commit-ים הראשונים, מדפיסה:

``````python
!git diff {first} {second} -- speed.py
``````

### פריט 2: שאלה מלאה — מצאו את שלוש הטעויות

גוף חם מתקרר בחדר שהטמפרטורה בו 20 מעלות. הפונקציה ‎`cool`‎ אמורה להחזיר את הטמפרטורה דקה אחת אחר כך, $T - k(T - 20)$. הפונקציה ‎`cooling_curve`‎ אמורה למלא, צעד אחרי צעד, מערך של הטמפרטורות בדקות ‎`0, 1, ..., n - 1`‎, והפונקציה ‎`first_minute_below`‎ אמורה להחזיר, בעזרת מסכה, את הדקה הראשונה שבה הטמפרטורה נמוכה מ-‎`limit`‎. הפלט הצפוי כתוב בהערות של שתי שורות ההדפסה.

בקוד שלוש טעויות. **קראו אותו בלי להריץ**: מצאו כל טעות לפי מספר השורה, תקנו אותה, והסבירו במשפט אחד מה היה לא נכון. ציינו גם איזו טעות מתגלה ראשונה כשמריצים את הקוד, ואילו טעויות היא מסתירה.

``````python
import numpy as np

def cool(T, k):
    """Temperature (C) one minute later."""
    return T - k * T - 20

def cooling_curve(T0, k, n):
    """Temperature (C) at minutes 0, 1, ..., n - 1."""
    T = np.zeros(n)
    T[0] = T0
    for i in range(1, n + 1):
        T[i] = cool(T[i - 1], k)
    return T

def first_minute_below(T, limit):
    """First minute with T below limit."""
    below = np.where(T <= limit)[0]
    return below[0]

T = cooling_curve(100.0, 0.5, 6)
print(T)   # expected: [100.   60.   40.   30.   25.   22.5]
print(first_minute_below(T, 30.0))   # expected: 4
``````

`````{admonition} פתרון
:class: dropdown, tip
1. **שורה 11:** ‎`range(1, n + 1)`‎ מגיע עד ‎`i = 6`‎, ואז ‎`T[i]`‎ הוא ‎`T[6]`‎, שאין לו מקום במערך באורך 6. התיקון: ‎`for i in range(1, n):`‎.
2. **שורה 5:** בלי סוגריים, הכפל ‎`k * T`‎ מחושב קודם ו-20 מוחסר אחריו, במקום להכפיל ב-‎`k`‎ את ההפרש ‎`T - 20`‎. התיקון: ‎`return T - k * (T - 20)`‎.
3. **שורה 17:** התנאי ‎`T <= limit`‎ מתקיים גם בדקה שבה הטמפרטורה שווה בדיוק ל-‎`limit`‎, והפונקציה צריכה את הדקה הראשונה שבה הטמפרטורה נמוכה ממנו. התיקון: ‎`below = np.where(T < limit)[0]`‎.

**הסדר שבו הטעויות מתגלות:** הריצה הראשונה נעצרת בשגיאה של טעות 1, עוד לפני שורת ההדפסה הראשונה:

```text
IndexError: index 6 is out of bounds for axis 0 with size 6
```

השגיאה מוצגת בשורה 12, אבל התיקון בשורה 11. השגיאה הזו מסתירה את שתי הטעויות האחרות: אחרי התיקון הראשון, שורת ההדפסה הראשונה מדפיסה

```text
[100.     30.     -5.    -22.5   -31.25  -35.625]
```

והשורה השנייה מדפיסה 1. הטמפרטורה יורדת מתחת ל-20 מעלות, טמפרטורת החדר, וזה אינו אפשרי לגוף שמתקרר בחדר: זו טעות 2. רק אחרי שמתקנים גם אותה הטמפרטורות נכונות, והשורה השנייה מדפיסה 3 במקום 4: בדקה 3 הטמפרטורה היא בדיוק 30, ואינה נמוכה מ-30. זו טעות 3.

הקוד המתוקן:

```python
import numpy as np

def cool(T, k):
    """Temperature (C) one minute later."""
    return T - k * (T - 20)

def cooling_curve(T0, k, n):
    """Temperature (C) at minutes 0, 1, ..., n - 1."""
    T = np.zeros(n)
    T[0] = T0
    for i in range(1, n):
        T[i] = cool(T[i - 1], k)
    return T

def first_minute_below(T, limit):
    """First minute with T below limit."""
    below = np.where(T < limit)[0]
    return below[0]

T = cooling_curve(100.0, 0.5, 6)
print(T)   # [100.   60.   40.   30.   25.   22.5]
print(first_minute_below(T, 30.0))   # 4
```
`````
