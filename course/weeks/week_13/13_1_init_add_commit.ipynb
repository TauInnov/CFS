## בקרת גרסאות: git init, add, commit

עד עכשיו, כשרציתם לחזור אחורה בקוד, השתמשתם ב-Ctrl+Z, או שמרתם עותקים ידניים של קובץ עם שמות כמו ‎`sim_v2.py`‎, ‎`sim_v2_final.py`‎, ‎`sim_v2_final_REAL.py`‎. שתי השיטות האלה נשברות בדיוק כשהכי צריך אותן:

- **Ctrl+Z** נעלם ברגע שסוגרים את הקובץ, ולא עוזר כלל אם השינוי הבעייתי בוצע לפני שבוע.
- **תיקיות ‎`v1`‎/‎`v2`‎/‎`final`‎** לא מתעדות *מדוע* כל גרסה נוצרה, קל לאבד עקבות אחרי איזה קובץ באמת "הכי מעודכן", ואי אפשר בקלות לראות מה בדיוק השתנה בין שתי גרסאות.

**בקרת גרסאות** (Version Control) פותרת את זה אחרת: היא שומרת **היסטוריה** של תמונות מצב (snapshots) של הקוד שלכם, כל אחת מתויגת בהודעה שמסבירה מה השתנה ומדוע. אפשר לחזור לכל נקודה בהיסטוריה, להשוות בין שתי נקודות, ולדעת בדיוק מה קרה שם — גם אחרי שנה. **Git** היא התוכנה הנפוצה ביותר לבקרת גרסאות, וזו המערכת שנלמד השבוע.

### תמונות מצב ואזור ה-staging

ב-Git, כל תמונת מצב כזו נקראת **commit**: תמונה שלמה של כל הקבצים שאתם עוקבים אחריהם ברגע נתון, ולא רשימת שינויים (diff). כל commit נשמר לצמיתות בהיסטוריית הפרויקט, יחד עם ההודעה שלו. ההיסטוריה דומה לרשימת ההיסטוריה שנבנתה עם ‎`append`‎ בשבוע 4: כל commit הוא תמונת מצב שנוספת לסוף ההיסטוריה, והתמונות הקודמות נשארות בה כפי שהיו.

לפני שיוצרים commit, שינויים עוברים דרך **אזור ה-staging** (ה"מגירת טיוטה") — מקום ביניים שבו בוחרים בדיוק אילו שינויים ייכנסו ל-snapshot הבא. זה מאפשר, למשל, לבחור רק חלק מהקבצים ששיניתם לצילום מצב אחד, ולהשאיר את השאר לצילום הבא.

שלוש הפקודות הבסיסיות:

- **‎`git init`‎** — הופך תיקייה רגילה ל-repository: תיקייה שבה Git עוקב אחרי היסטוריית הקבצים. מריצים פעם אחת בתחילת הפרויקט.
- **‎`git add <קובץ>`‎** — מעביר שינויים באותו קובץ לאזור ה-staging (המגירה).
- **‎`git commit -m "הודעה"`‎** — לוקח את כל מה שנמצא כרגע במגירה, ושומר אותו כתמונת מצב קבועה בהיסטוריה, עם ההודעה שכתבתם.

```{note}
כל דוגמאות ה-Git בשבוע הזה רצות בתוך תיקייה **זמנית**, שנוצרת מחדש בכל הרצה של המחברת (‎`tempfile.mkdtemp()`‎) — כדי לא לגעת בפרויקט האמיתי שלכם. כשתתרגלו על המחשב שלכם, תריצו את אותן פקודות בדיוק בתוך תיקיית הפרויקט האמיתית.
```

### דוגמה: לעטוף repo סביב סקריפט אמיתי

ניקח את סקריפט טווח הזריקה שכבר בנינו בסעיף 7.9 (‎`R = v0**2 * sin(2*theta) / g`‎), ונבנה אותו **בהדרגה, קומיט אחרי קומיט** — בדיוק כמו שהייתם עושים תוך כדי עבודה אמיתית: קודם השלד, אחר כך החישוב, ולבסוף ההדפסה.

``````python
import tempfile, os

workdir = tempfile.mkdtemp(prefix="git_demo_")
os.chdir(workdir)
print("עובדים בתיקייה זמנית:", workdir)
``````

``````python
!git init -q -b main
!git config user.email "student@example.com"
!git config user.name "Student"
!git config color.ui false
!git status
``````

`git status` מדווח: repo ריק, בלי קבצים עוקבים. עכשיו ניצור את השלד של הסקריפט, ונבצע קומיט ראשון.

``````python
%%writefile projectile.py
import numpy as np
from scipy import constants

v0 = 20.0
theta_deg = 45
``````

``````python
!git add projectile.py
!git commit -q -m "שלד: הגדרת המשתנים v0 ו-theta_deg"
!git log --oneline
``````

עכשיו נוסיף את חישוב הטווח עצמו — **אותו קובץ**, שינוי חדש, קומיט חדש.

``````python
%%writefile -a projectile.py

theta = np.radians(theta_deg)
R = v0**2 * np.sin(2 * theta) / constants.g
``````

``````python
!git add projectile.py
!git commit -q -m "הוספת חישוב טווח הזריקה R"
!git log --oneline
``````

ולבסוף, שורת הדפסה — קומיט שלישי.

``````python
%%writefile -a projectile.py

print(f"R = {R:.2f} m")
``````

``````python
!git add projectile.py
!git commit -q -m "הוספת הדפסת התוצאה"
!git log --oneline
!echo "---"
!python3 projectile.py
``````

שלושה קומיטים, כל אחד **גרסה מלאה** של ‎`projectile.py`‎ באותו רגע. אפשר לחזור לכל אחת מהן; כיצד קוראים ומשווים ביניהן נלמד בסעיף הבא.
