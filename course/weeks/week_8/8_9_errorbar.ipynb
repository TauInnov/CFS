## errorbar: נקודות עם שגיאות

זה אולי הכלי החשוב ביותר לגרף של דוח מעבדה: כל מדידה אמיתית מגיעה עם **שגיאה**, וזו חייבת להופיע על הגרף לצד הנקודה עצמה. הפונקציה:

```python
ax.errorbar(x, y, yerr=..., xerr=..., fmt="o", capsize=4)
```

- **‎`yerr`‎** / **‎`xerr`‎** — גודל השגיאה בכל ציר (מספר בודד, או מערך — שגיאה שונה לכל נקודה).
- **‎`fmt="o"`‎** — קובע שהנקודות עצמן מוצגות כעיגולים **בלי** קו מחבר ביניהן (כפי שמתאים לנקודות מדידה בדידות).
- **‎`capsize`‎** — אורך ה"כובעים" הקטנים בקצות קווי השגיאה.

### דוגמה: טווח נמדד מול תיאוריה, לפי זווית

בכל אחת מחמש זוויות שיגור נמדד הטווח כמה פעמים. במערך ‎`means`‎ נמצא הטווח הממוצע בכל זווית, ובמערך ‎`errors`‎ הפיזור של המדידות סביב הממוצע (בסעיף 11.2 נלמד לחשב את שניהם מהמדידות עצמן). נציג אותם כנקודות עם קווי שגיאה, לצד עקומת התיאוריה.

``````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
angles_deg = np.array([15, 30, 45, 60, 75])
means = np.array([20.6, 35.0, 41.2, 35.6, 20.1])   # mean measured range (m)
errors = np.array([2.1, 3.4, 4.2, 3.6, 2.0])       # spread of the throws (m)

angles_theory = np.linspace(0, 90, 200)
R_theory = v0**2 * np.sin(np.radians(2 * angles_theory)) / g

fig, ax = plt.subplots()
ax.plot(angles_theory, R_theory, color="gray", label="תיאוריה")
ax.errorbar(angles_deg, means, yerr=errors, fmt="o", color="tab:red",
            capsize=4, label='"מדידה" ± שגיאה')
ax.set_xlabel("Launch angle (degrees)")
ax.set_ylabel("Range R (m)")
ax.legend()
plt.show()
``````

### באג נפוץ: קו שמחבר בין נקודות המדידה

אם קוראים ל-‎`errorbar`‎ עם ‎`fmt="-"`‎ (או משאירים את ברירת המחדל שמצפה לקו), Matplotlib **יחבר את הנקודות בקו רציף** — מה שנראה כאילו יש מודל רציף בין נקודות המדידה, למרות שאין. עבור נקודות מדידה בדידות, כמעט תמיד רוצים ‎`fmt="o"`‎ (או סמל אחר) בלי קו מחבר.
