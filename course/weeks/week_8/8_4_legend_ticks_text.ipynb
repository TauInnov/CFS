## תיוג הגרף: legend, גבולות וסימוני ציר, טקסט וחצים

כלים נוספים לתיוג גרף מדעי:

- **‎`ax.legend(loc=...)`‎** — מקרא, בונה את עצמו מתוך ‎`label=`‎ שהעברתם ל-‎`plot`‎/‎`scatter`‎.
- **‎`ax.set_xlim(xmin, xmax)`‎**, **‎`ax.set_ylim(ymin, ymax)`‎** — קביעת תחום הצירים ידנית.
- **‎`ax.set_xticks([...])`‎** — קביעת מיקומי הסימונים על הציר במפורש.
- **‎`ax.text(x, y, "...")`‎** — טקסט חופשי בנקודה מסוימת על הגרף.
- **‎`ax.annotate("...", xy=..., xytext=..., arrowprops=...)`‎** — טקסט **עם חץ** שמצביע על נקודה ספציפית.

### דוגמה: סימון נקודת השיא במסלול

הגובה המרבי של הזריקה מתקבל ב-$t_{peak} = v_0\sin\theta / g$, ושווה $y_{max} = (v_0\sin\theta)^2 / (2g)$. נסמן את הנקודה הזו על הגרף עם חץ וטקסט. ב-‎`ax.annotate`‎, הארגומנט ‎`xy`‎ הוא הנקודה שהחץ מצביע עליה, ו-‎`xytext`‎ הוא המקום שבו נכתב הטקסט (ממנו יוצא החץ). ‎`arrowprops`‎ הוא מילון שקובע כיצד החץ נראה.

```````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
theta = np.radians(45)
t_flight = 2 * v0 * np.sin(theta) / g
t = np.linspace(0, t_flight, 200)
x = v0 * np.cos(theta) * t
y = v0 * np.sin(theta) * t - 0.5 * g * t**2

t_peak = v0 * np.sin(theta) / g
x_peak = v0 * np.cos(theta) * t_peak
y_peak = (v0 * np.sin(theta))**2 / (2 * g)

fig, ax = plt.subplots()
ax.plot(x, y, label=r"$\theta=45^\circ$")
ax.annotate("Maximum height",
            xy=(x_peak, y_peak), xytext=(x_peak + 3, y_peak - 2),
            arrowprops={"arrowstyle": "->", "color": "gray"})
ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
ax.legend(loc="upper right")
plt.show()
```````

### דוגמה נוספת: גבולות, סימוני ציר וטקסט

בגרף הקירור מסעיף 8.1 נוסיף קו מקווקו של טמפרטורת החדר, נקבע את תחום הצירים ואת הסימונים על ציר הזמן, ונכתוב משפט על הגרף עם ‎`ax.text`‎.

```````python
T0, k = 80.0, 0.1                # initial temperature (°C), cooling constant
minutes = np.arange(0, 31)       # 0, 1, ..., 30 min
T = 20 + (T0 - 20) * (1 - k)**minutes   # temperature (°C) after each minute

fig, ax = plt.subplots()
ax.plot(minutes, T, label="Body")
ax.plot(minutes, 20 * np.ones(minutes.shape), linestyle="--",
        color="gray", label="Room")
ax.text(2, 8, "The gap to the room shrinks by 10% every minute")
ax.set_xlim(0, 30)
ax.set_ylim(0, 90)
ax.set_xticks([0, 10, 20, 30])
ax.set_xlabel("t (min)")
ax.set_ylabel("T (°C)")
ax.legend(loc="upper right")
plt.show()
```````

### נסו בעצמכם

בקוד הבא $v_0 = 20$ מ'/ש', הזווית ‎45°‎ ו-$g = 10$ מ'/ש², מספרים עגולים שאפשר לחשב איתם בראש. שורה 9 חסרה: השלימו אותה כך שחץ יצביע על **נקודת הנחיתה**, והמילה "Landing" תיכתב בנקודה ‎`(32, 3)`‎. חשבו קודם את הטווח $R = v_0^2\sin(2\theta)/g$: באיזו נקודה החץ צריך לגעת בקרקע?

```````python
g, v0 = 10.0, 20.0
theta = np.radians(45)
t = np.linspace(0, 2 * v0 * np.sin(theta) / g, 100)
x = v0 * np.cos(theta) * t
y = v0 * np.sin(theta) * t - 0.5 * g * t**2

fig, ax = plt.subplots()
ax.plot(x, y)
# line 9: ax.annotate(...)
ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
plt.show()
```````

`````{admonition} פתרון
:class: dropdown, tip

```python
ax.annotate("Landing", xy=(40, 0), xytext=(32, 3),

```
`````
