## subplots

לפעמים רוצים כמה גרפים **נפרדים** באותו figure, לא כמה עקומות על אותו ציר. לכך נועד ‎`plt.subplots(nrows, ncols)`‎: הוא מחזיר ‎`fig`‎ יחיד, ומערך ‎`axs`‎ של צירים בגודל ‎`nrows × ncols`‎.

```python
fig, axs = plt.subplots(1, 3, figsize=(12, 4))
axs[0].plot(...)   # first axes
axs[1].plot(...)   # second axes
```

```{note}
כאשר ‎`nrows=1`‎ (או ‎`ncols=1`‎), ‎`axs`‎ הוא מערך **חד-ממדי** — גישה עם ‎`axs[i]`‎. כאשר גם ‎`nrows>1`‎ וגם ‎`ncols>1`‎, ‎`axs`‎ הוא מערך **דו-ממדי** — גישה עם ‎`axs[row, col]`‎: קודם מספר השורה ואחריו מספר העמודה, שניהם מתחילים מ-0. למשל ‎`axs[0, 0]`‎ הוא הציר השמאלי העליון, ו-‎`axs[1, 0]`‎ הציר שמתחתיו.
```

### דוגמה: פאנל נפרד לכל זווית שיגור

במקום לצייר את שלוש הזוויות על אותו גרף (כמו בסעיף 8.5), הפעם ניתן לכל זווית ציר משלה, עם ‎`sharey=True`‎ כדי שיהיה קל להשוות גבהים. הלולאה עוברת על האינדקסים 0, 1 ו-2: בסיבוב ‎`i`‎ היא לוקחת את הזווית ‎`angles_deg[i]`‎ ומציירת אותה על הציר ‎`axs[i]`‎.

```````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
angles_deg = [30, 45, 60]

fig, axs = plt.subplots(1, 3, figsize=(12, 4), sharey=True)
for i in range(len(angles_deg)):
    angle = angles_deg[i]
    theta = np.radians(angle)
    t_flight = 2 * v0 * np.sin(theta) / g
    t = np.linspace(0, t_flight, 200)
    x = v0 * np.cos(theta) * t
    y = v0 * np.sin(theta) * t - 0.5 * g * t**2
    axs[i].plot(x, y)
    axs[i].set_title(fr"$\theta={angle}^\circ$")
    axs[i].set_xlabel("x (m)")

axs[0].set_ylabel("y (m)")
plt.show()
```````

### דוגמה נוספת: שני גדלים, שני פאנלים

לגוף המתקרר מסעיף 8.1 נצייר גם את הירידה בטמפרטורה בדקה הבאה, ‎`k*(T - 20)`‎, שהיא בדיוק מה ש-‎`cool`‎ מחסירה. לטמפרטורה ולירידה יש סקאלות שונות, ולכן כל אחת מקבלת פאנל משלה, זו מעל זו. ‎`sharex=True`‎ נותן לשני הפאנלים אותו ציר זמן.

```````python
T0, k = 80.0, 0.1                # initial temperature (°C), cooling constant
minutes = np.arange(0, 31)       # 0, 1, ..., 30 min
T = 20 + (T0 - 20) * (1 - k)**minutes   # temperature (°C) after each minute
drop = k * (T - 20)              # temperature drop in the next minute (°C)

fig, axs = plt.subplots(2, 1, figsize=(6, 6), sharex=True)
axs[0].plot(minutes, T)
axs[0].set_ylabel("T (°C)")
axs[1].plot(minutes, drop, color="tab:red")
axs[1].set_ylabel("Drop per minute (°C)")
axs[1].set_xlabel("t (min)")
plt.show()
```````
