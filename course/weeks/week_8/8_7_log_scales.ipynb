## סקאלות לוגריתמיות ושימושיהן

מסלול הזריקה שלנו לא דורש סולם לוגריתמי — הערכים נמצאים כולם באותו סדר גודל. אבל כשגודל פיזיקלי משתנה על פני **כמה סדרי גודל**, סולם ליניארי "דוחס" את הערכים הקטנים לכדי קו אחד צמוד לציר, והגרף הופך חסר תועלת. הפתרון:

```python
ax.set_xscale("log")
ax.set_yscale("log")
```

אפשר להפעיל את שניהם יחד (log-log), רק אחד מהם, או אף אחד — תלוי בטווח הערכים בכל ציר בנפרד.

### דוגמה: טווח הזריקה כתלות במהירות ההתחלתית

הטווח נתון על ידי $R(v_0) = v_0^2 \sin(2\theta)/g$ — תלות **ריבועית** ב-$v_0$. אם נסרוק $v_0$ על פני טווח רחב (למשל בין 1 ל-1000 מ/ש, כתרחיל מתמטי גרידא, לא ערכים פיזיקליים סבירים לזריקת יד), $R$ ישתנה על פני כמה סדרי גודל, ונשווה תצוגה ליניארית מול log-log.

```````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
theta = np.radians(45)
v0_range = 10**np.linspace(0, 3, 200)   # 1 ... 1000, log-spaced
R_range = v0_range**2 * np.sin(2 * theta) / g

fig, axs = plt.subplots(1, 2, figsize=(10, 4))

axs[0].plot(v0_range, R_range)
axs[0].set_title("Linear scale")
axs[0].set_xlabel(r"$v_0$ (m/s)")
axs[0].set_ylabel("R (m)")

axs[1].plot(v0_range, R_range)
axs[1].set_xscale("log")
axs[1].set_yscale("log")
axs[1].set_title("Log-log scale")
axs[1].set_xlabel(r"$v_0$ (m/s)")
axs[1].set_ylabel("R (m)")

plt.show()
```````

בגרף הליניארי (משמאל) כמעט כל המידע על ערכי $v_0$ הקטנים "נעלם" צמוד לציר. בגרף ה-log-log (מימין) רואים קו ישר — תוצאה של חוק חזקה ($R\propto v_0^2$). השיפוע של הקו הישר הוא החזקה, 2: כשה-$v_0$ גדל פי 10, הטווח גדל פי $10^2 = 100$.

חשוב: הביטוי ‎`10**np.linspace(0, 3, 200)`‎ פורש 200 **מעריכים** במרווחים שווים בין 0 ל-3, והחזקה הופכת אותם לערכים בין $10^0$ ל-$10^3$. כך הנקודות פרוסות באופן אחיד **בסולם הלוגריתמי**, ולא באופן אחיד בין 1 ל-1000.

### דוגמה נוספת: דעיכה בסולם לוגריתמי בציר אחד

בגוף המתקרר מסעיף 8.1, ההפרש מטמפרטורת החדר, $T - 20$, קטן פי $(1-k)$ בכל דקה. בסולם לוגריתמי **בציר y בלבד** דעיכה כזו נראית כקו ישר. זה ההבדל מהדוגמה הקודמת: חוק חזקה נראה ישר בסולם log-log, ודעיכה מעריכית נראית ישרה כשרק ציר y לוגריתמי.

```````python
T0, k = 80.0, 0.1                # initial temperature (°C), cooling constant
minutes = np.arange(0, 61)       # 0, 1, ..., 60 min
gap = (T0 - 20) * (1 - k)**minutes   # T - 20: distance from the room (°C)

fig, axs = plt.subplots(1, 2, figsize=(10, 4))
axs[0].plot(minutes, gap)
axs[0].set_title("Linear scale")
axs[0].set_xlabel("t (min)")
axs[0].set_ylabel("T - 20 (°C)")
axs[1].plot(minutes, gap)
axs[1].set_yscale("log")
axs[1].set_title("Log scale on y only")
axs[1].set_xlabel("t (min)")
axs[1].set_ylabel("T - 20 (°C)")
plt.show()
```````
