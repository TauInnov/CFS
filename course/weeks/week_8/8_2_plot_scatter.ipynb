## plot ו-scatter: קווים, נקודות וסגנון

ב-Matplotlib יש שתי פונקציות ציור עיקריות שרלוונטיות כמעט לכל דוח מעבדה:

- **‎`ax.plot(x, y)`‎** — מתאימה ל**עקומה רציפה**: מודל תיאורטי, פונקציה מחושבת על רשת צפופה של נקודות.
- **‎`ax.scatter(x, y)`‎** — מתאימה ל**נקודות בדידות**: מדידות בפועל, שאין ביניהן קשר רציף מובנה.

ההבחנה הזו אינה טכנית בלבד — היא משקפת הבדל פיזיקלי: בדוח מעבדה מקובל לצייר את המודל התיאורטי כקו רציף, ואת המדידות כנקודות נפרדות עליו.

### דוגמה: מודל תיאורטי מול "מדידות" עם רעש

ניצור את מסלול הזריקה $y(x)$ (גובה כתלות במרחק אופקי) עבור $\theta=45^\circ$ כעקומה רציפה, ולצידה כמה נקודות "מדידה" על אותו מסלול, עם רעש אקראי קטן שמדמה אי-דיוק של מכשיר מדידה.

את הרעש הכנו מראש: המערך ‎`noise`‎ מכיל 8 סטיות אקראיות קטנות סביב 0, בפיזור (סטיית תקן) של 0.3 מטר. מוסיפים אותו לגובה המחושב, וכך כל "מדידה" זזה מעט, באקראי, מהמסלול. כיצד מגרילים מספרים כאלה נלמד בסעיף 11.3.

```````python
import numpy as np

# measurement noise for the example: 8 small random deviations (m), given ready
noise = np.random.default_rng(0).normal(0, 0.3, 8)
```````

```````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
theta = np.radians(45)
t_flight = 2 * v0 * np.sin(theta) / g

t_full = np.linspace(0, t_flight, 200)
x_full = v0 * np.cos(theta) * t_full
y_full = v0 * np.sin(theta) * t_full - 0.5 * g * t_full**2

t_meas = np.linspace(0, t_flight, 8)
x_meas = v0 * np.cos(theta) * t_meas
y_meas = v0 * np.sin(theta) * t_meas - 0.5 * g * t_meas**2 + noise

fig, ax = plt.subplots()
ax.plot(x_full, y_full, color="tab:blue", label="Model")
ax.scatter(x_meas, y_meas, color="tab:red", marker="o",
           label='"Measurements" (with noise)')
ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
ax.legend()
plt.show()
```````

פרמטרי הסגנון הנפוצים: ב-‎`plot`‎ — ‎`color`‎, ‎`linestyle`‎ (‎`"-"`‎, ‎`"--"`‎, ‎`":"`‎), ‎`linewidth`‎; ב-‎`scatter`‎ — ‎`color`‎, ‎`marker`‎ (‎`"o"`‎, ‎`"^"`‎, ‎`"s"`‎...), ‎`s`‎ (גודל הנקודה).

### דוגמה נוספת: מודל הקירור מול קריאות של מדחום

הגוף המתקרר מסעיף 8.1: המודל מחושב בכל דקה ומצויר כקו, ושבע קריאות של מדחום, אחת כל 5 דקות, מצוירות כנקודות. הקריאות נתונות כמערך.

```````python
T0, k = 80.0, 0.1                # initial temperature (°C), cooling constant
minutes = np.arange(0, 31)       # 0, 1, ..., 30 min
T = 20 + (T0 - 20) * (1 - k)**minutes   # temperature (°C) after each minute
t_meas = np.array([0, 5, 10, 15, 20, 25, 30])                 # min
T_meas = np.array([80.4, 55.1, 40.6, 32.7, 27.4, 24.3, 22.4])  # readings (°C)

fig, ax = plt.subplots()
ax.plot(minutes, T, color="tab:blue", label="Model")
ax.scatter(t_meas, T_meas, color="tab:red", label="Thermometer")
ax.set_xlabel("t (min)")
ax.set_ylabel("T (°C)")
ax.legend()
plt.show()
```````

### נסו בעצמכם: מצאו את שלוש הטעויות

הקוד הבא אמור לצייר את המודל כ**קו מקווקו** ואת 8 המדידות כ**משולשים**, עם מקרא שמבדיל ביניהם. הוא רץ בלי שגיאה, אבל בגרף שמתקבל המודל מופיע כ-200 נקודות נפרדות, המדידות מחוברות בקו, ואין מקרא.

**לפני שמריצים**: מצאו את שלוש הטעויות לפי מספרי השורות, תקנו כל אחת והסבירו במשפט אחד. רק אחר כך הריצו את הגרסה המתוקנת.

```````python
fig, ax = plt.subplots()
ax.scatter(x_full, y_full, color="tab:blue", label="Model")
ax.plot(x_meas, y_meas, color="tab:red", marker="^", label='"Measurements"')
ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
plt.show()
```````
