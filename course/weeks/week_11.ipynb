# שבוע 11 — pandas

השבוע הזה שונה מהשבועות הקודמים: המטרה היא **אוריינות, לא הפקה**. לא נדרש לכתוב שאילתת pandas מהזיכרון — נדרש להבין קוד pandas קיים, לחזות מה הוא מחזיר, ולזהות בו טעות. זו גם ההכנה לשבוע 12: pandas משמשת כשכבת הקלט של נתוני הניסוי, לפני שהניתוח חוזר למערכי NumPy (סעיף 11.7).

גם כאן הכלים הם תרגום של מבנים מוכרים: **מילון** מהחצי הראשון ← עמודות של טבלה, ו**צמצום** משבועות 7 ו-9 ← צמצום לפי קבוצה עם ‎`groupby`‎.

**הדוגמה המובילה:** יומן מדידות ‎`launch_log.csv`‎ — 30 זריקות (חמש זוויות, שש חזרות בכל זווית), המשך ישיר לניסוי של השבועות הקודמים.

בסוף השבוע תדעו:
- להסביר מדוע טבלה (DataFrame) מתאימה לנתוני מעבדה יותר ממערכי NumPy נפרדים
- להבדיל בין ‎`DataFrame`‎ ל-‎`Series`‎, ולקרוא ‎`columns`‎/‎`index`‎/‎`dtypes`‎
- לטעון קובץ CSV ולבצע עליו בדיקות שפיות בסיסיות
- לבחור עמודות, לסנן שורות לפי תנאי בוליאני, ולמיין
- לזהות ערכים חסרים (‎`NaN`‎) ולהסיר אותם עם ‎`dropna`‎
- לקבץ נתונים לפי עמודה (‎`groupby`‎) ולחשב סטטיסטיקות על כל קבוצה
- להפיק תמונת מצב מהירה על טבלה שלמה עם ‎`describe()`‎
- לעבור בחזרה למערכי NumPy עם ‎`to_numpy()`‎, ולהבין מדוע זה נחוץ
- **לקרוא קוד pandas שנכתב על ידי סוכן AI ולזהות בו באגים אופייניים** — המיומנות המרכזית של השבוע
