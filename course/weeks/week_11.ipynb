# שבוע 11 — סטטיסטיקה והתאמת קווים

השבוע הזה מאחד את כל מה שנבנה עד כה: NumPy (שבועות 7–9), Matplotlib (שבוע 8), ו-pandas (שבוע 10) -- כולם משרתים כאן מטרה אחת: להסיק מסקנה כמותית ואמינה מתוך נתוני מעבדה אמיתיים, עם רעש. זהו השבוע המורכב והארוך ביותר בקורס.

**הדוגמה המובילה:** `lab_measurements.csv` -- 40 מדידות (חמש זוויות, שמונה חזרות בכל זווית) של אותה זריקה שמלווה אותנו מאז שבוע 7, הפעם עם ערכים חסרים (`NaN`) ונקודה חריגה אחת, בדיוק כמו במעבדה אמיתית. השאלה המרכזית של השבוע: **האם המדידות תואמות את המודל התאורטי `R = v0²/g · sin(2θ)`, ובאיזו רמת ביטחון?**

בסוף השבוע תדעו:
- לנקות נתונים אמיתיים: `NaN`, וזיהוי נקודות חריגות בשיטה עמידה (median/MAD)
- לחשב ממוצע, סטיית תקן, ושגיאת הממוצע (SEM), ולהבין למה הם שונים זה מזה
- לדמות רעש מדידה עם `default_rng`, ולעבור בין `groupby` (pandas) לצמצום לפי ציר (NumPy) על אותם נתונים
- להוליך שגיאת מדידה למסקנה מחושבת, אנליטית ובמונטה-קרלו
- לכתוב מאפס פונקציית התאמה ליניארית (least squares), בלי `polyfit`/`curve_fit`
- לחשב שאריות, שגיאות מקדמים, `chi²`, `chi²_ν`, ו-`R²` -- ולדעת מה כל אחד מהם באמת אומר
- להתאים בעזרת משקלים, לזהות השפעת נקודה חריגה, ולבצע לינאריזציה (`ln`) להתאמת מודלים לא-ליניאריים
- להשוות בין מודלים מתחרים בצורה עניינית, לא רק לפי `R²`
- להרכיב גרף מסכם שלם (נתונים + התאמה + שאריות) ולשמור טבלת תוצאות עם `savetxt`
