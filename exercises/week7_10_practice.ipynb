## סיכום

- כל מקרה של הנוסחה הופך למסכה: `(t >= 0) & (t < 4)`.
- משתמשים ב-`&` וב-`|` עם סוגריים סביב כל השוואה. `and`, `or` ו-`0 <= t < 4` נכשלים על מערכים.
- בודקים שהמקרים מכסים כל נקודה ואינם חופפים: `np.all(m1 | m2 | m3)`, `np.any(m1 & m2)`.
- בגבולות (`<` או `<=`) קל לטעות בלי שתתקבל שגיאה.
