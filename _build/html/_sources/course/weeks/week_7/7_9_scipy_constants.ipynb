## קבועים פיזיקליים ויחידות (scipy.constants)

עד כה כתבנו את תאוצת הנפילה החופשית כמספר קשיח בקוד: `g = 9.8`. זה עובד, אבל יש לו שני חסרונות: (1) קל לטעות בערך או בדיוק שלו, ו-(2) הקורא לא תמיד יודע מיד מה משמעות המספר. מודול `scipy.constants` מספק קבועים פיזיקליים סטנדרטיים, בדיוק גבוה וביחידות SI מוגדרות היטב.

``````python
from scipy import constants

print("תאוצת הכובד g:", constants.g, "m/s^2")
print("מהירות האור c:", constants.c, "m/s")
print("קבוע פלאנק h:", constants.h, "J·s")
print("מטען היסוד e:", constants.e, "C")
``````

### דוגמה: שכתוב נוסחת הזריקה עם קבוע רשמי

נחזור לנוסחת הזריקה $y(t) = v_0 t - \tfrac{1}{2} g t^2$, הפעם עם `constants.g` במקום 9.8 המוקלד ביד:

``````python
import numpy as np
from scipy import constants

v0 = 15.0
t = np.linspace(0, 3, 10)
y = v0 * t - 0.5 * constants.g * t**2
print(y)
``````

```{note}
`constants.g` הוא **התאוצה הסטנדרטית של כדור הארץ** (9.80665 m/s²) — לא לבלבל עם `constants.G`, **קבוע הכבידה האוניברסלי** ($6.674\times10^{-11}\ \text{N}\cdot\text{m}^2/\text{kg}^2$) מחוק הכבידה של ניוטון. שם דומה, גודל פיזיקלי שונה לגמרי — טעות נפוצה.
```

``````python
print("g (תאוצת נפילה חופשית):", constants.g)
print("G (קבוע הכבידה האוניברסלי):", constants.G)
``````
