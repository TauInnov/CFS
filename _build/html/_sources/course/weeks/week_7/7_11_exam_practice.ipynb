## תרגול מסכם


בלי להריץ, תארו במילים מה מחשב הקוד הבא, ומה יהיה ה-`shape` של `result`:

``````python
import numpy as np

t = np.linspace(0, 5, 50)
v0 = np.array([10.0, 15.0, 20.0])

result = v0[:, None] * t - 0.5 * 9.8 * t**2
``````

`````{admonition} פתרון
:class: dropdown, tip
הקוד מחשב את גובה הזריקה $y(t) = v_0 t - \tfrac{1}{2} g t^2$ עבור **שלוש מהירויות שיגור שונות בו-זמנית**, על פני 50 נקודות זמן. השורה `v0[:, None]` הופכת את `v0` מווקטור באורך 3 למערך בצורת `(3, 1)`, כך שכל מהירות "מוכפלת" כנגד כל נקודות הזמן.

`result.shape` הוא `(3, 50)`: שורה לכל מהירות שיגור, עמודה לכל נקודת זמן — בדיוק כמו התבנית שראינו בסעיף 7.3.
`````

הקוד הבא נועד לחשב את זמן הטיסה הכולל של כדור שנזרק אנכית כלפי מעלה במהירות $v_0$ וחוזר לאותו גובה: $t_{flight} = 2v_0/g$, וכן את גובהו המרבי $h_{max} = v_0^2/(2g)$. יש בו באג. אתרו ותקנו:

``````python
from scipy import constants

v0 = 12.0   # m/s

t_flight_buggy = 2 * v0 / constants.G     # שימו לב לאות
h_max_buggy = v0**2 / 2 * constants.g     # ולכאן

print(t_flight_buggy, h_max_buggy)
``````

`````{admonition} פתרון
:class: dropdown, tip
שני באגים:
1. `constants.G` (קבוע הכבידה האוניברסלי) נכתב במקום `constants.g` (תאוצת הנפילה החופשית) — טעות שראינו בסעיף 7.9.
2. סדר הפעולות ב-`h_max_buggy` שגוי: `v0**2 / 2 * constants.g` מחשב $(v_0^2/2)\cdot g$ במקום $v_0^2/(2g)$.

```python
t_flight_fixed = 2 * v0 / constants.g
h_max_fixed = v0**2 / (2 * constants.g)
print(t_flight_fixed, h_max_fixed)
`````
