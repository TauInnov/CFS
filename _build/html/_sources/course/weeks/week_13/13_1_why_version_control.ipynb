## למה בקרת גרסאות (Version Control)

עד עכשיו, כשרציתם לחזור אחורה בקוד, השתמשתם ב-Ctrl+Z, או שמרתם עותקים ידניים של קובץ עם שמות כמו `sim_v2.py`, `sim_v2_final.py`, `sim_v2_final_REAL.py`. שתי השיטות האלה נשברות בדיוק כשהכי צריך אותן:

- **Ctrl+Z** נעלם ברגע שסוגרים את הקובץ, ולא עוזר כלל אם השינוי הבעייתי בוצע לפני שבוע.
- **תיקיות `v1`/`v2`/`final`** לא מתעדות *למה* כל גרסה נוצרה, קל לאבד עקבות אחרי איזה קובץ באמת "הכי מעודכן", ואי אפשר בקלות לראות מה בדיוק השתנה בין שתי גרסאות.

**בקרת גרסאות** (Version Control) פותרת את זה אחרת: היא שומרת **היסטוריה** של תמונות מצב (snapshots) של הקוד שלכם, כל אחת מתויגת בהודעה שמסבירה מה השתנה ולמה. אפשר לחזור לכל נקודה בהיסטוריה, להשוות בין שתי נקודות, ולדעת בדיוק מה קרה שם — גם אחרי שנה. **Git** היא התוכנה הנפוצה ביותר לבקרת גרסאות, וזו המערכת שנלמד השבוע.

### תרחיש: שני מקרים שכל פיזיקאי מכיר

**מקרה א׳ — הניתוח שאי אפשר לשחזר.** ניתחתם נתוני מעבדה, הקוד עבד ונתן תוצאה סבירה. שיניתם פרמטר אחד "רק לבדוק", בלי לשמור עותק גיבוי קודם. השינוי שיבש הכל, ואתם כבר לא זוכרים בדיוק מה היה הערך המקורי — התוצאה הטובה, שאולי כבר הספקתם לדווח עליה, לא ניתנת לשחזור.

**מקרה ב׳ — שני שותפי מעבדה, קובץ אחד.** אתם ובן/בת הזוג למעבדה עורכים את אותו סקריפט ניתוח, כל אחד בנפרד על המחשב שלו. בסוף היום שולחים קבצים ב-WhatsApp ומנסים "לשלב ידנית" — ומגלים ששניכם שיניתם את אותה פונקציה, בדרכים שונות.

בשני המקרים הבעיה זהה: **אין תיעוד של היסטוריית השינויים**, ואין דרך מהימנה לחזור אחורה או להשוות גרסאות.
