## שמירה לדוח: savefig, dpi, פורמט

צילום מסך של הגרף מהמחברת **אינו** דרך מקובלת לצרף גרף לדוח מעבדה — הרזולוציה נמוכה, ולרוב נחתכות תוויות. הדרך הנכונה: `fig.savefig(...)`.

```python
fig.savefig("range_vs_angle.png", dpi=300, bbox_inches="tight")
```

- **`dpi`** — רזולוציית השמירה (נקודות לאינץ'); 300 היא רזולוציית הדפסה סבירה, לעומת ברירת המחדל הנמוכה יותר להצגה על מסך.
- **`bbox_inches="tight"`** — חותך שוליים מיותרים ומוודא שתוויות שלמות (לא נחתכות בקצה).
- **פורמט הקובץ** נקבע לפי הסיומת: `.png` (רסטר, טוב לתצוגה), `.pdf` או `.svg` (וקטורי — חד גם בהגדלה, מועדף למסמכי LaTeX/Word).

### דוגמה: שמירת גרף ההשוואה כקובץ לדוח

נבנה מחדש את גרף הטווח-מול-זווית עם קווי השגיאה (כמו בסעיף 8.9), והפעם נשמור אותו לקובץ.

``````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
rng = np.random.default_rng(2)
angles_deg = np.array([15, 30, 45, 60, 75])

means, stds = [], []
for angle in angles_deg:
    theta_i = np.radians(angle)
    v0_trials = rng.normal(20.0, 1.0, size=300)
    R_trials = v0_trials**2 * np.sin(2 * theta_i) / g
    means.append(R_trials.mean())
    stds.append(R_trials.std())
means, stds = np.array(means), np.array(stds)

theta_theory = np.radians(np.linspace(0, 90, 200))
R_theory = v0**2 * np.sin(2 * theta_theory) / g

fig, ax = plt.subplots()
ax.plot(np.degrees(theta_theory), R_theory, color="gray", label="תיאוריה")
ax.errorbar(angles_deg, means, yerr=stds, fmt="o", color="tab:red", capsize=4, label='"מדידה" ± שגיאה')
ax.set_xlabel("Launch angle (degrees)")
ax.set_ylabel("Range R (m)")
ax.legend()

fig.savefig("range_vs_angle.png", dpi=300, bbox_inches="tight")
plt.show()
``````
