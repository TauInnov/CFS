בדוח מעבדה, גרף בלי תוויות צירים עם יחידות הוא גרף לא קביל — גם אם הנתונים שבו נכונים. שתי הפקודות המרכזיות:

```python
ax.set_xlabel("x (m)")

```

הכלל הפשוט: **כל תווית ציר כוללת גם את היחידה בסוגריים**, לא רק את שם המשתנה.

### דוגמה

``````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
theta = np.radians(45)
t_flight = 2 * v0 * np.sin(theta) / g
t = np.linspace(0, t_flight, 200)
x = v0 * np.cos(theta) * t
y = v0 * np.sin(theta) * t - 0.5 * g * t**2

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
ax.set_title(r"מסלול הזריקה ($\theta=45^\circ$, $v_0=20\,$m/s)")
plt.show()
``````

### באג נפוץ: תווית בלי יחידה, או עם יחידה שגויה

תווית כמו `"y"` בלבד אינה אומרת לקורא הדוח אם $y$ נמדד במטרים, בסנטימטרים, או ביחידות שרירותיות. גרוע מזה — תווית עם **יחידה שגויה** (למשל `"y (cm)"` כשהערכים בפועל במטרים) מטעה יותר מאשר תווית חסרה, כי היא נראית תקינה במבט ראשון.

### נסו בעצמכם

בקוד הבא יש טעות: הערכים מחושבים במטרים, אבל התווית כתובה בסנטימטרים. אתרו ותקנו.

``````python
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel("x (cm)")   # <- יש כאן טעות
ax.set_ylabel("y (m)")
plt.show()
``````
