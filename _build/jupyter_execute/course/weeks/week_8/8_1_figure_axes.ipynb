## figure ו-axes: המודל המנטלי

מעכשיו נלמד *לראות* את מה שחישבנו בשבוע 7. הספרייה שבה נשתמש היא **Matplotlib**, ובכל השבוע נעבוד עם **הממשק האובייקטי** שלה בלבד:

```python
fig, ax = plt.subplots()
ax.plot(x, y)
```

`fig` הוא כל הקנבס (התמונה כולה, כולל אפשרות למספר צירים), ו-`ax` הוא מערכת הצירים הבודדת שבתוכה באמת מציירים. כל הפקודות שקובעות איך הגרף נראה — כותרת, תוויות, תחומים — מופעלות על ה-`ax`, עם קידומת `set_`: `ax.set_title(...)`, `ax.set_xlabel(...)`.

יש גם ממשק אחר, "מבוסס-מצב" (`plt.plot(...)` ישירות, בלי `ax`), שנפוץ בהדרכות באינטרנט. הוא עובד היטב כשיש figure אחד בלבד, אבל הופך מבלבל ברגע שיש כמה צירים על אותו figure (סעיף 8.6) — ולכן בקורס הזה נדבוק תמיד ב-`()fig, ax = plt.subplots`.

### דוגמה: מסלול הזריקה מזווית אחת

בשבוע 7 חישבנו את גובה הזריקה כפונקציה של הזמן:

$$y(t) = v_0 \sin\theta \, t - \tfrac{1}{2} g t^2$$

עבור מהירות התחלתית $v_0$ וזווית שיגור $\theta$. עכשיו ניצור figure אחד וציר (axes) אחד, ונצייר עליו את הגובה כתלות בזמן, עבור $\theta = 45^\circ$.

``````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
theta = np.radians(45)

t_flight = 2 * v0 * np.sin(theta) / g
t = np.linspace(0, t_flight, 100)
y = v0 * np.sin(theta) * t - 0.5 * g * t**2

fig, ax = plt.subplots()
ax.plot(t, y)
ax.set_title(r"גובה הזריקה כפונקציה של הזמן ($\theta=45^\circ$)")
ax.set_xlabel("t (s)")
ax.set_ylabel("y (m)")
plt.show()
``````

### באג נפוץ: `plt.plot` במקום `ax.plot`

כשיש figure אחד בלבד, אין הבדל נראה-לעין בין `plt.plot(...)` ל-`ax.plot(...)`. אבל `()plt.plot` תמיד מצייר לתוך "הציר הפעיל האחרון" — וברגע שיש כמה צירים על אותו figure (כמו בסעיף 8.6), קל מאוד שהגרף "יברח" לציר הלא נכון בלי שום שגיאת ריצה. ההרגל של `()ax.plot` מהיום הראשון חוסך את הבאג הזה מראש.
