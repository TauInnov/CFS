## תיוג הגרף: legend, גבולות וסימוני ציר, טקסט וחצים

כלים נוספים לתיוג גרף מדעי:

- **`ax.legend(loc=...)`** — מקרא, בונה את עצמו מתוך `label=` שהעברתם ל-`plot`/`scatter`.
- **`ax.set_xlim(xmin, xmax)`**, **`ax.set_ylim(ymin, ymax)`** — קביעת תחום הצירים ידנית.
- **`ax.set_xticks([...])`** — קביעת מיקומי הסימונים על הציר במפורש.
- **`ax.text(x, y, "...")`** — טקסט חופשי בנקודה מסוימת על הגרף.
- **`ax.annotate("...", xy=..., xytext=..., arrowprops=...)`** — טקסט **עם חץ** שמצביע על נקודה ספציפית.

### דוגמה: סימון נקודת השיא במסלול

הגובה המרבי של הזריקה מתקבל ב-$t_{peak} = v_0\sin\theta / g$, ושווה $y_{max} = (v_0\sin\theta)^2 / (2g)$. נסמן את הנקודה הזו על הגרף עם חץ וטקסט.

``````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
theta = np.radians(45)
t_flight = 2 * v0 * np.sin(theta) / g
t = np.linspace(0, t_flight, 200)
x = v0 * np.cos(theta) * t
y = v0 * np.sin(theta) * t - 0.5 * g * t**2

t_peak = v0 * np.sin(theta) / g
x_peak = v0 * np.cos(theta) * t_peak
y_peak = (v0 * np.sin(theta))**2 / (2 * g)

fig, ax = plt.subplots()
ax.plot(x, y, label=r"$\theta=45^\circ$")
ax.annotate("גובה מרבי",
            xy=(x_peak, y_peak), xytext=(x_peak + 3, y_peak - 2),
            arrowprops=dict(arrowstyle="->", color="gray"))
ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
ax.legend(loc="upper right")
plt.show()
``````
