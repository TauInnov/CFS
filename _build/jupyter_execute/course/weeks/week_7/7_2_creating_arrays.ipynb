## יצירת מערכים: array, zeros, ones, arange, linspace

מלבד בנייה מתוך רשימה קיימת (`np.array`), ל-NumPy יש כמה פונקציות ליצירת מערכים ישירות:

- `np.zeros(n)` — מערך של n אפסים, שימושי כ"מכל" ריק שממלאים בהמשך.
- `np.ones(n)` — מערך של n אחדים.
- `np.arange(start, stop, step)` — כמו `range` בפייתון, אך מחזיר מערך. **הקצה העליון לא נכלל**.
- `np.linspace(start, stop, num)` — num נקודות במרווחים שווים בין start ל-stop, **כולל שתי הקצוות**.

ההבדל בין `arange` ל-`linspace` הוא מקור נפוץ לטעויות "בגובה אחד" (off-by-one).

### דוגמה: וקטור זמן לזריקה

בסעיף הקודם בנינו את וקטור הזמן של הזריקה כרשימה קבועה מראש. הפעם נבנה אותו ישירות עם `linspace`, ונכין גם מערך ריק לאחסון תוצאה שנחשב בהמשך.

``````python
import numpy as np
import matplotlib.pyplot as plt

g = 9.8
y0, v0 = 0.0, 15.0

t_arr = np.linspace(0, 3, 10)   # 10 נקודות בין 0 ל-3, כולל שתי הקצוות
y_arr = np.zeros(t_arr.size)    # מכל ריק לתוצאה

y_arr = y0 + v0 * t_arr - 0.5 * g * t_arr**2
print(t_arr)
print(y_arr)
``````

### arange מול linspace

נשווה ישירות בין שתי הפונקציות על אותו טווח, כדי לראות את ההבדל בהתנהגות בקצה העליון:

``````python
a = np.arange(0, 10, 2)     # לא כולל 10
b = np.linspace(0, 10, 5)    # כולל 10

fig, ax = plt.subplots(figsize=(6, 1.5))
ax.plot(a, np.zeros_like(a), 'o', label="arange(0,10,2)")
ax.plot(b, np.ones_like(b), 'o', label="linspace(0,10,5)")
ax.set_yticks([0, 1])
ax.set_yticklabels(["arange", "linspace"])
ax.set_xlabel("Value")
ax.legend(loc="upper right", bbox_to_anchor=(1.3, 1))
plt.show()
``````
