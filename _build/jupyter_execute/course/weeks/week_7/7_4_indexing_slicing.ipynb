## אינדוקס ופריסה בציר אחד

אינדוקס ופריסה (slicing) של מערך חד-ממדי כמעט זהים לאלה של רשימה: `a[i]` לאיבר בודד, `a[start:stop]` לטווח, `a[start:stop:step]` עם קפיצה, ואינדקסים שליליים סופרים מהסוף. ההבדל המרכזי בין השניים יידון בסעיף הבא (views מול copies).

### דוגמה: חיתוך מסלול הזריקה

נחזור למסלול הזריקה $y(t) = v_0 t - \tfrac{1}{2} g t^2$ ונשלוף ממנו חלקים שונים.

``````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 15.0
t = np.linspace(0, 3, 16)
y = v0 * t - 0.5 * g * t**2

first_half = y[:8]          # מחצית ראשונה של הטיסה
every_third = y[::3]        # כל נקודה שלישית — קצב דגימה נמוך יותר
last_three = y[-3:]         # שלוש הנקודות האחרונות

print("מחצית ראשונה:", first_half)
print("כל שלישית:", every_third)
print("שלוש אחרונות:", last_three)
``````

נסמן על גבי הגרף אילו נקודות נכללות במחצית הראשונה, כדי לחדד את המשמעות של הפריסה:

``````python
fig, ax = plt.subplots()
ax.plot(t, y, "o-", color="lightgray", label="כל המסלול")
ax.plot(t[:8], y[:8], "o", color="tab:blue", label="y[:8]")
ax.set_xlabel("t (s)")
ax.set_ylabel("y (m)")
ax.legend()
plt.show()
``````
