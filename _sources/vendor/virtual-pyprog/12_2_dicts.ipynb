# מילונים (`dict`)

## מוטיבציה למילונים

ברשימה ניגשים לאיבר לפי המיקום שלו. זה נוח כשלסדר יש משמעות, למשל בסדרת מדידות לאורך זמן. אבל לפעמים טבעי יותר לגשת לערך **לפי שם**: הציון של סטודנט לפי שמו, הצפיפות של חומר לפי שם החומר, טמפרטורת החדר לפי שם החדר. ברשימה היינו צריכים לזכור באיזה אינדקס נמצא כל ערך, או לחפש אותו איבר אחרי איבר.

מבנה הנתונים **מילון** (dictionary, או `dict`) בנוי בדיוק בשביל זה.

## מילונים בפייתון

<span style="color: red;">TODO: סרטון חגי לגבי מבנה הנתונים מילון. עם התוכן הזה</span>

מילון שומר התאמה בין **מפתחות** (keys) ל**ערכים** (values). בניגוד לרשימה, שבה ניגשים לאיבר לפי המיקום שלו, במילון ניגשים לערך לפי המפתח שלו.

לכל מפתח יש ערך אחד בלבד: המפתח חייב להיות **ייחודי**. אם נכניס למילון מפתח שכבר קיים בו, הערך הקודם יוחלף. לכן בוחרים מפתחות שמזהים את הערך באופן חד-משמעי. למשל, אם בכיתה יש שני סטודנטים בשם משה, שם פרטי לא יספיק כמפתח, וצריך שם מלא או מספר תעודת זהות.

מילון הוא mutable, כמו רשימה: אפשר להוסיף לו מפתחות ולשנות את הערכים שבו אחרי שנוצר.

```{admonition} **כדאי לדעת!**
:class: info

מילון שומר על סדר ההכנסה: כשמדפיסים מילון או עוברים עליו בלולאה, המפתחות מופיעים לפי הסדר שבו הוכנסו אליו. עם זאת, אי אפשר לגשת לערך במילון לפי המיקום שלו, אלא רק לפי המפתח.
```

בדומה לרשימות וטאפלים, גם מילונים עטופים בסוגריים - במקרה של מילון מדובר בסוגריים מסולסלים.
לדוגמא, מילון ריק יראה כך: ‎`d={}`‎.

אך בניגוד לרשימות וטאפלים, **כל אלמנט במילון הוא למעשה צמד של מפתח וערך**.   
המפתח והערך מופרדים ע"י נקודותיים (`:`) וכל זוג של מפתח וערך מופרד באמצעות פסיק (`,`).  


לכן המימוש של המילון המופיע באילוסטרציה ייראה כך:

`````python
studs={"Avi": 90, "Moshe": 80, "Adam": 70}
`````

כדי לקבל את הערך של מפתח, כותבים את המפתח בסוגריים מרובעים אחרי שם המילון. למשל, הציון של משה:

`````python
print(studs["Moshe"])
`````

הסוגריים המרובעים נראים כמו גישה לאיבר ברשימה, אבל בתוכם כותבים מפתח ולא אינדקס.

ומה היה קורה אם היינו מנסים לגשת למפתח שלא קיים במילון?  
בדומה לשגיאת ‎`index out of range`‎ שקיבלנו ברשימות באינדקסים שלא קיימים, במקרה של מפתח לא קיים נקבל ‎`key error`‎.

`````python
studs = {"Avi": 90, "Moshe": 80, "Adam": 70}
print(studs["Haim"])
`````

מאחר שגישה למפתח לא קיים תגרום לשגיאה, היינו רוצים לוודא שמפתח קיים במילון לפני שננסה לגשת לערך שלו.
למזלנו, יש כזו אפשרות, על-ידי המילה השמורה `in` שכבר ראינו ברשימות ומחרוזות.

במקרה של מילונים, הפקודה ‎`a in dict`‎ תחזיר `True` אם `a` הוא **מפתח** במילון, ו`False` אחרת.

`````python
studs = {"Avi": 90, "Moshe": 80, "Adam": 70}

print("Johnny" in studs)
print("Avi" in studs)
print(80 in studs) # 80 is a value, not a key, so that will be False.
`````

דרך נוספת לגשת לערכים בצורה בטוחה היא באמצעות המתודה `get`. מתודה זו מאפשרת לנו לגשת לערך של מפתח במילון מבלי לגרום לשגיאה אם המפתח לא קיים.

אם המפתח קיים במילון - המתודה תחזיר לנו את הערך שמקושר לkey.  
אם המפתח לא קיים במילון, המתודה תחזיר לנו None.  
שימו לב כי ניתן לשנות את הערך שחוזר במקרים בהם המפתח לא קיים במילון לכל ערך אלטרנטיבי שנרצה ע"י העברתו בארגומנט השני בפונקציה `get`.

`````python
studs = {"Avi": 90, "Moshe": 80, "Adam": 70}

print(studs.get("Avi")) # The key exists, so we'll get the value

print(studs.get("Johnny")) # The key doesn't exist, so we'll get the default - None

print(studs.get("Johnny", "Student not found")) # The key doesn't exist, so we'll get the default - "not found"
`````

כעת נרצה להוסיף סטודנט חדש למילון.
אין צורך ליצור את המילון מחדש - ניתן להוסיף למילון ערך נוסף על ידי הצבת הערך במפתח המתאים:
``` python
d[key] = value
```
נגיד, נוסיף את ג'וני (Johnny) למילון הסטודנטים שלנו.

`````python
studs["Johnny"] = 95
print(studs)
`````

באופן דומה, אם נבצע השמה ל‎`d[key]`‎ עבור מפתח שכבר היה במילון, אנחנו **נשנה** את הערך שהיה קיים כבר - נדרוס אותו ונכניס ערך חדש.
לדוגמה, אם נבצע השמה מחדש לציון של משה:

`````python
studs["Moshe"] = 90
print(studs)
`````

### לולאה על מילון

לולאת `for` על מילון עוברת על **המפתחות** שלו, לפי הסדר שבו הוכנסו. בעזרת המפתח אפשר לקבל בתוך הלולאה גם את הערך:

`````python
studs = {"Avi": 90, "Moshe": 80, "Adam": 70}
`````

`````python
for name in studs:
    print(name, studs[name])
`````

כשצריך בכל איטרציה גם את המפתח וגם את הערך, נוח להשתמש במתודה `items`. היא נותנת בכל איטרציה זוג של מפתח וערך, והזוג מתפרק לשני משתנים, כמו שראינו בדף על טאפלים:

`````python
for key, value in studs.items():
    print(key, ":", value)
`````

קיימות גם המתודות `keys` ו-`values`, שנותנות רק את המפתחות או רק את הערכים. למשל, כך מחשבים את ממוצע הציונים (הפונקציה `len` של מילון מחזירה את מספר המפתחות שבו):

`````python
print(sum(studs.values()) / len(studs))
`````

````{admonition} **להרחבה: הסרה ומיזוג של מילונים**
:class: dropdown, note

המתודה `pop` מסירה מפתח מהמילון ומחזירה את הערך שהיה לו. אם המפתח לא קיים במילון, מתקבלת השגיאה `KeyError`.

המתודה `update` מקבלת מילון נוסף ומכניסה את כל הזוגות שלו למילון: מפתח חדש מתווסף, ומפתח שכבר קיים מקבל את הערך החדש. שתי המתודות משנות את המילון עצמו, כמו `append` ו-`remove` ברשימה.

```python
studs = {"Avi": 90, "Moshe": 80, "Adam": 70}

print(studs.pop("Moshe"))   # 80
print(studs)                # {'Avi': 90, 'Adam': 70}

studs.update({"Adam": 75, "Johnny": 60})
print(studs)                # {'Avi': 90, 'Adam': 75, 'Johnny': 60}
```
````

````{admonition} **להרחבה: מפתח שמורכב מכמה ערכים**
:class: dropdown, note

מפתח במילון חייב להיות ערך immutable: מספר, מחרוזת או `tuple`. רשימה אינה יכולה לשמש מפתח, וניסיון להשתמש בה כמפתח מסתיים בשגיאה `TypeError`.

כשערך אחד לא מספיק כדי לזהות את מה שרוצים לשמור, למשל כשיש שני סטודנטים עם אותו שם פרטי, אפשר להשתמש ב-`tuple` כמפתח:

```python
studs = {("Moshe", "Cohen"): 90, ("Moshe", "Levi"): 85}

print(studs[("Moshe", "Levi")])   # 85
```

באותה דרך אפשר לשמור ערך לכל נקודה ברשת, כשהמפתח הוא `tuple` של שתי הקואורדינטות שלה.
````

### סיכום: פעולות על מילון

| פעולה            | תיאור                                                                 |
|---------------------|------------------------------------------------------------------------|
| ‎`D = {}`‎            | יוצר מילון ריק                                                         |
| ‎`D[k] = value`‎      | קובע את הערך של המפתח `k` (ומוסיף את המפתח אם הוא לא קיים)            |
| ‎`D[k]`‎              | מחזיר את הערך של המפתח `k` (שגיאת KeyError אם `k` לא קיים)           |
| ‎`k in D`‎            | מחזיר `True` אם `k` הוא מפתח במילון, אחרת `False`                    |
| ‎`D.get(k, [d])`‎     | מחזיר את ‎`D[k]`‎ אם קיים, אחרת את `d` (ברירת מחדל: `None`)             |
| ‎`for k in D`‎        | לולאה על המפתחות, לפי סדר ההכנסה                                      |
| ‎`D.items()`‎         | זוגות של מפתח וערך, לשימוש בלולאה                                     |
| ‎`D.keys()`‎          | כל המפתחות במילון                                                     |
| ‎`D.values()`‎        | כל הערכים במילון                                                      |
| ‎`len(D)`‎            | מספר המפתחות במילון                                                   |

- ארגומנט שמסומן כ־<span style="color:red"><i>[d]</i></span> הוא אופציונלי  
    - אם לא מציינים אותו, ייעשה שימוש בערך ברירת המחדל  
- מידע נוסף על מתודות של מילון נמצא [כאן](https://docs.python.org/3/library/stdtypes.html#dict)

```python
def get_char_count(text):
    char_count = {}
    for char in text:
        if not char in char_count:
            char_count[char] = 1
        else: 
            char_count[char] += 1
    return char_count
```

```python
def get_char_count(text):
    char_count={}
    for char in text:
        char_count[char] = char_count.get(char, 0) + 1
    return char_count
```

```{admonition} **מה בעצם קורה פה?**
:class: dropdown, info

עד כה ביצענו לכל תו במחרוזת:
אם התו במילון - תוסיף לו 1.
אם התו לא במילון - תכניס אותו מחדש עם הערך 1.

במקום זאת, אנחנו מחליפים את הכל בפקודת `get` אחת!
אם התו היה במילון - הפקודה תחזיר את הערך שהיה לו, תוסיף לו 1, ותשמור אותו חזרה באותו המפתח.
אם התו לא היה במילון - הפקודה תחזיר 0, תוסיף לו 1, ותשמור אותו (לראשונה) במילון תחת המפתח המתאים.

```
