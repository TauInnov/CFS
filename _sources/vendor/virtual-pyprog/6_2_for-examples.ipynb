````{admonition} **לחצו כאן כדי לצפות בפתרון**
```python
def step(T1, T2, k, dt):
    flow = k * (T1 - T2) * dt
    return [T1 - flow, T2 + flow]


def simulate(T1, T2, k, dt, n):
    history1 = [T1]
    history2 = [T2]
    for i in range(n):
        T = step(history1[-1], history2[-1], k, dt)
        history1.append(T[0])
        history2.append(T[1])
    return history1, history2


history1, history2 = simulate(100.0, 20.0, 0.25, 1, 3)
print([round(t, 2) for t in history1])   # [100.0, 80.0, 70.0, 65.0]
print([round(t, 2) for t in history2])   # [20.0, 40.0, 50.0, 55.0]
```
````

7. **מצאו את 3 השגיאות.** סטודנט כתב את `simulate` מתרגיל 5 בלי הפונקציה `step`. בהרצה שבתחתית החלונית הוא ציפה לקבל את הפלט:

```
[100.0, 80.0, 70.0, 65.0]
[20.0, 40.0, 50.0, 55.0]
```

קראו את הקוד **לפני** שאתם מריצים אותו. מצאו את שלוש השגיאות, תקנו אותן, והסבירו במשפט אחד מה הבעיה בכל אחת. איזו שגיאה תראו ראשונה כשתריצו את הקוד, ואילו שגיאות היא מסתירה?

`````python
def simulate(T1, T2, k, dt, n):
    for i in range(n):
        history1 = [T1]
        history2 = [T2]
        T1 = T1 - k * (T1 - T2) * dt
        T2 = T2 + k * (T1 - T2) * dt
        history1.append(T1)
        history2.append(T2)
        return history1, history2


history1, history2 = simulate(100.0, 20.0, 0.25, 1, 3)
print(history1)
print(history2)
`````

````{admonition} **לחצו כאן כדי לצפות בפתרון**
```python
def simulate(T1, T2, k, dt, n):
    history1 = [T1]
    history2 = [T2]
    for i in range(n):
        flow = k * (T1 - T2) * dt
        T1 = T1 - flow
        T2 = T2 + flow
        history1.append(T1)
        history2.append(T2)
    return history1, history2


history1, history2 = simulate(100.0, 20.0, 0.25, 1, 3)
print(history1)   # [100.0, 80.0, 70.0, 65.0]
print(history2)   # [20.0, 40.0, 50.0, 55.0]

1. **`return` בתוך הלולאה.** הפונקציה מסתיימת כבר בסוף האיטרציה הראשונה. צריך להוציא את `return` מההזחה של הלולאה.
2. **הרשימות נוצרות מחדש בכל איטרציה.** השורות `history1 = [T1]` ו-`history2 = [T2]` נמצאות בתוך הלולאה, ולכן בכל צעד הרשימות מתחילות מחדש וההיסטוריה הקודמת נמחקת. צריך ליצור אותן פעם אחת, לפני הלולאה.
3. **`T2` מחושב מ-`T1` החדש.** בשורה השנייה `T1` כבר עודכן, ולכן הגוף הקר מקבל פחות חום ממה שהגוף החם איבד, והסכום לא נשמר. מחשבים קודם את `flow` מהערכים הישנים ומשתמשים בו בשני העדכונים.

**מה רואים בהרצה?** הקוד המקורי מדפיס `[100.0, 80.0]` ו-`[20.0, 35.0]`: רק צעד אחד, בגלל שגיאה 1. היא מסתירה את שגיאה 2, שאפשר לראות רק אחרי התיקון הראשון: אז מודפסות רשימות באורך 2 בלבד, עם ערכים "משונים" כמו `62.421875`. גם שגיאה 3 גלויה כבר בהרצה הראשונה למי שבודק את הסכום: `80.0 + 35.0` הוא 115 ולא 120.
````
