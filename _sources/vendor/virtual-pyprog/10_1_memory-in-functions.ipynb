## Scope

כפי שחלקכם אולי שמתם לב, לא כל משתנה שנגדיר יהיה זמין בכל מקום בתוכנית. הscope של משתנה מתאר את "תחום הראייה" שלו - כלומר, באילו חלקים של התוכנית ניתן לגשת אליו או להשתמש בו. 

בכל פעם שמריצים פונקציה, נוצר עבורה scope חדש ונפרד. בתוך הscope הזה חיים המשתנים המקומיים שלה - למשל, הפרמטרים שהוגדרו לפונקציה או משתנים שנוצרו בתוכה. משתנים אלו קיימים רק בזמן הריצה של הפונקציה, וברגע שהפונקציה מסתיימת - הם "נעלמים". מחוץ לפונקציה אנחנו לא יכולים לגשת אליהם ישירות, אלא רק לראות את התוצאה שהפונקציה מחזירה.

משתנים השייכים לפונקציה מסוימת נקראים משתנים מקומיים (local).
משתנים אשר לא שייכים לאף פונקציה נקראים משתנים גלובליים (global).



בואו נראה דוגמה לכך. 
הנה פונקציה שמגדירה משתנה בתוך הפונקציה.

`````python
def my_func():
    n=5
`````

כעת נגדיר משתנה מחוץ לפונקציה בשם n:

`````python
n=6
`````

`````python
n = 6
my_func()
print(n) # we will get 6 because we didn't change the global n.
`````

## משתנים מקומיים וגלובליים

שלוש דוגמאות קצרות מסכמות את מה שראינו. משתנה מקומי (local) קיים רק בתוך הפונקציה שבה הוא הוגדר, ולכן אי אפשר לגשת אליו מבחוץ:

`````python
def my_func():
    local_var = 6

print(local_var) # Error
`````

לעומת זאת, משתנה גלובלי (global) מוגדר מחוץ לפונקציות, ולכן אפשר להשתמש בו מכל מקום בתוכנית – גם מתוך פונקציות. 

`````python
global_var = 6
def my_func():
    print(global_var)

my_func()
`````

שימו לב! אם בתוך פונקציה נגדיר משתנה בשם זהה למשתנה גלובלי, המשתנה המקומי "מסתיר" את הגלובלי, והקוד בפונקציה יתייחס אך ורק אליו.

`````python
var = 6
def my_func():
    var = 5
    print(var)

my_func()
`````

## העברת מידע לפונקציות וחזרה


כפי שראינו, כאשר אנחנו מעבירים משתנה לפונקציה, אנחנו בעצם משייכים את הערך של המשתנה האקטואלי אל המשתנה הפורמלי של הפונקציה. כלומר, שני השמות מצביעים לאותו אובייקט בזיכרון. 

השאלה המעניינת היא – **האם הפונקציה יכולה לשנות את האובייקט שהיא מקבלת כפרמטר?**
מדובר בדיוק במקרה שראינו בפרק הקודם - שני שמות מצביעים לאותו אובייקט.

כלומר, היכולת לשנות תלויה בסוג האובייקט: אם מדובר באובייקט immutable, הפונקציה לא תוכל לשנות אותו. לעומת זאת, אם האובייקט שהועבר כקלט הוא mutable, הפונקציה כן יכולה לבצע עליו שינויים, ו**שינויים אלו יישארו בתוקף גם לאחר שהפונקציה סיימה לרוץ**.

`````python
def func(x):
    x.append(9)
    
a=[4,5,6,7,8]

print(a)
func(a)
print(a)
`````

אז מה קרה כאן בעצם?
כשהעברנו את הרשימה `a` לפונקציה, נוצרה השמה של הערך של `a` במשתנה הפורמלי `x`.
כלומר, `x` ו`a` כעת מצביעים לתחילת הרשימה.

מאחר שרשימה היא אובייקט mutable ו`append` מבצע שינוי של הרשימה עצמה, אז השינוי יופיע גם בa.

```{admonition} **לסיכום**  באילו דרכים ניתן "להוציא" ערכים מתוך פונקציה?
:class: dropdown, note
אנחנו ראינו שתי דרכים להחזיר מידע מתוך פונקציות:  
1. שימוש במילה השמורה `return`.
2. עריכה של קלטים שהם mutable.

לשם השלמות - ניתן גם להשתמש במשתנים גלובליים על מנת לשמור ערכים, אך הדרך הזו אינה בחומר הקורס (וגם נחשבת לא הכי אלגנטית בקרב מתכנתים).
```

## בחנו את עצמכם: גוף מתקרר

נחזור לקירור משבוע 4. גוף בטמפרטורה `T` נמצא בחדר בטמפרטורה 20 מעלות. בכל צעד הוא מתקרב לטמפרטורת החדר, והפונקציה `cool` מחזירה את הטמפרטורה החדשה. נכתוב שתי פונקציות שמקררות רשימה של גופים: `cool_all` משנה את הרשימה שהיא מקבלת, ו-`cooled` מחזירה רשימה חדשה ולא נוגעת ברשימה המקורית.

```python
def cool(T, k):
    return T - k * (T - 20)

def cool_all(temps, k):      # changes the list it gets
    for i in range(len(temps)):
        temps[i] = cool(temps[i], k)

def cooled(temps, k):        # returns a new list
    new = []
    for T in temps:
        new.append(cool(T, k))
    return new

a = [80.0, 50.0]
b = cooled(a, 0.5)
print(a)
print(b)
cool_all(a, 0.5)
print(a)
cooled(a, 0.5)
print(a)
```

### מצאו את 3 השגיאות

סטודנט רצה ש-`cool_all` תקרר את כל הגופים ברשימה, ושהקוד ידפיס:

```
[50.0, 35.0]
```

קראו את הקוד **לפני** שאתם מריצים אותו. מצאו את שלוש השגיאות, תקנו אותן, והסבירו במשפט אחד מה הבעיה בכל אחת. איזו שגיאה תראו ראשונה כשתריצו את הקוד, ואילו שגיאות היא מסתירה?

`````python
def cool(T, k):
    T = T - k * (T - 20)


def cool_all(temps, k):
    for T in temps:
        T = cool(T, k)


temps = [80.0, 50.0]
temps = cool_all(temps, 0.5)
print(temps)
`````

````{admonition} **לחצו כאן כדי לצפות בפתרון**
:class: dropdown, tip

```python
def cool(T, k):
    return T - k * (T - 20)


def cool_all(temps, k):
    for i in range(len(temps)):
        temps[i] = cool(temps[i], k)


temps = [80.0, 50.0]
cool_all(temps, 0.5)
print(temps)   # [50.0, 35.0]
```

1. **`cool` לא מחזירה ערך.** ההשמה ‎`T = ...`‎ משנה רק את המשתנה המקומי `T`, והפונקציה מחזירה `None`. צריך `return`.
2. **השמה למשתנה הלולאה לא משנה את הרשימה.** בלולאה ‎`for T in temps`‎, השורה ‎`T = ...`‎ מצמידה את השם `T` לערך חדש, אבל הרשימה עצמה נשארת כמו שהייתה. כדי לכתוב לתוך הרשימה צריך לעבור על האינדקסים ולכתוב ‎`temps[i] = ...`‎.
3. **‎`temps = cool_all(...)`‎.** הפונקציה `cool_all` לא מחזירה ערך, ולכן ההשמה שמה ב-`temps` את `None`. מספיק לקרוא לפונקציה: היא משנה את הרשימה עצמה.

**מה רואים בהרצה?** הקוד המקורי מדפיס `None`, בגלל שגיאה 3. היא מסתירה את השאר: אחרי שמתקנים אותה מודפס ‎`[80.0, 50.0]`‎, כלומר הרשימה לא השתנתה (שגיאה 2). אחרי התיקון השני מודפס ‎`[None, None]`‎, כי `cool` עדיין לא מחזירה ערך (שגיאה 1). רק אחרי שלושת התיקונים מודפס ‎`[50.0, 35.0]`‎.
````
