# שבוע 9 — NumPy מתקדם ואינטגרציה

רב-ממד וצירים, ופעולות חשבון אינפיניטסימלי כקריאה לפונקציה. השבוע ממשיך ישירות את דוגמת מסלול הזריקה משבועות 7–8, הפעם עם **כמה ריצות של אותה זריקה** — בדיוק הדוגמה שמארגנת את ארבעת הכללים לעבודה עם מערכים רב-ממדיים: מערך בצורה (ריצה, זמן, קואורדינטה).

```{admonition} ארבעת הכללים לעבודה עם מערכים רב-ממדיים
:class: tip
1. `shape` היא רשימת הצירים, ולכל מיקום יש שם.
2. `axis=k` הוא הציר שנעלם בפעולת צמצום.
3. שני מערכים מתחברים אם, מימין לשמאל, כל זוג צירים שווה או שאחד מהם 1.
4. `None` מוסיף ציר בגודל 1.
```

בסוף השבוע תדעו:
- לקרוא ולפרש `shape` של מערך רב-ממדי לפי המשמעות הפיזיקלית של כל ציר
- לפרוס מערכים לפי צירים (`a[0]`, `a[:,2]`, `a[...,0]`)
- להבדיל בין `reshape` (סידור מחדש) ל-`transpose` (החלפת צירים)
- לאחד מערכים עם `stack` ו-`concatenate`
- להתאים צירים בין מערכים בעזרת broadcasting ו-`None`
- לבנות רשתות דו-ממדיות עם `meshgrid`
- לצמצם מערך לפי ציר נבחר (`mean`, `std`, `sum`, `argmax`...)
- לקרב נגזרת ואינטגרל עם `diff`, `cumsum`, `trapezoid` ו-`quad`
- לזהות מתי לא ניתן לוקטר (יחס נסיגה), ולכתוב לולאת עדכון נכונה

## Materials
- {doc}`9.1 • מבנה נתונים רב-ממדי: דו-ממד ותלת-ממד <../../course/weeks/week_9/9_1_multidim_arrays>`
- {doc}`9.2 • פריסה לפי צירים <../../course/weeks/week_9/9_2_axis_indexing>`
- {doc}`9.3 • reshape ו-transpose <../../course/weeks/week_9/9_3_reshape_transpose>`
- {doc}`9.4 • stack ו-concatenate — לאחד ריצות <../../course/weeks/week_9/9_4_stack_concatenate>`
- {doc}`9.5 • התאמת צירים בין שני מערכים (Broadcasting) <../../course/weeks/week_9/9_5_broadcasting>`
- {doc}`9.6 • None ו-meshgrid <../../course/weeks/week_9/9_6_none_meshgrid>`
- {doc}`9.7 • צמצום לפי ציר <../../course/weeks/week_9/9_7_axis_reductions>`
- {doc}`9.8 • diff ו-cumsum — נגזרת וסכום מצטבר <../../course/weeks/week_9/9_8_diff_cumsum>`
- {doc}`9.9 • אינטגרציה: trapezoid ו-quad <../../course/weeks/week_9/9_9_trapezoid_quad>`
- {doc}`9.10 • יחס נסיגה ולולאת עדכון <../../course/weeks/week_9/9_10_recurrence_loop>`
- {doc}`9.11 • תרגול מסכם <../../course/weeks/week_9/9_11_practice>`
