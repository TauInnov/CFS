## errorbar: נקודות עם שגיאות

זה אולי הכלי החשוב ביותר לגרף של דוח מעבדה: כל מדידה אמיתית מגיעה עם **שגיאה**, וזו חייבת להופיע על הגרף לצד הנקודה עצמה. הפונקציה:

```python
ax.errorbar(x, y, yerr=..., xerr=..., fmt="o", capsize=4)
```

- **`yerr`** / **`xerr`** — גודל השגיאה בכל ציר (מספר בודד, או מערך — שגיאה שונה לכל נקודה).
- **`fmt="o"`** — קובע שהנקודות עצמן מוצגות כעיגולים **בלי** קו מחבר ביניהן (כפי שמתאים לנקודות מדידה בדידות).
- **`capsize`** — אורך ה"כובעים" הקטנים בקצות קווי השגיאה.

### דוגמה: טווח נמדד מול תיאוריה, לפי זווית

עבור כמה זוויות שיגור, נדמה כמה חזרות עם רעש במהירות ההתחלתית (כמו בסעיף 8.8), נחשב את **הממוצע וסטיית התקן** של הטווח בכל זווית, ונציג אותם כנקודות עם קווי שגיאה — לצד עקומת התיאוריה.

``````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
rng = np.random.default_rng(2)
angles_deg = np.array([15, 30, 45, 60, 75])

means, stds = [], []
for angle in angles_deg:
    theta_i = np.radians(angle)
    v0_trials = rng.normal(20.0, 1.0, size=300)
    R_trials = v0_trials**2 * np.sin(2 * theta_i) / g
    means.append(R_trials.mean())
    stds.append(R_trials.std())

means = np.array(means)
stds = np.array(stds)

theta_theory = np.radians(np.linspace(0, 90, 200))
R_theory = 20.0**2 * np.sin(2 * theta_theory) / g

fig, ax = plt.subplots()
ax.plot(np.degrees(theta_theory), R_theory, color="gray", label="תיאוריה")
ax.errorbar(angles_deg, means, yerr=stds, fmt="o", color="tab:red", capsize=4, label='"מדידה" ± שגיאה')
ax.set_xlabel("זווית שיגור (מעלות)")
ax.set_ylabel("טווח R (m)")
ax.legend()
plt.show()
``````

### באג נפוץ: `fmt="-"` במקום `fmt="o"`

אם קוראים ל-`errorbar` עם `fmt="-"` (או משאירים את ברירת המחדל שמצפה לקו), Matplotlib **יחבר את הנקודות בקו רציף** — מה שנראה כאילו יש מודל רציף בין נקודות המדידה, למרות שאין. עבור נקודות מדידה בדידות, כמעט תמיד רוצים `fmt="o"` (או סמל אחר) בלי קו מחבר.
