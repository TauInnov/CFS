## היסטוגרמה: bins, density

היסטוגרמה מציגה את **התפלגות** ערכי מדגם — כמה תוצאות נפלו בכל תחום ערכים. הפונקציה המרכזית: ‎`ax.hist(data, bins=, density=)`‎.

- **‎`bins`‎** — מספר התאים (או גבולותיהם המפורשים) שאליהם מחלקים את טווח הנתונים.
- **‎`density`‎** — כשהיא ‎`True`‎, ציר ה-$y$ מציג **צפיפות הסתברות** (השטח הכולל מתחת להיסטוגרמה שווה 1); כשהיא ‎`False`‎ (ברירת המחדל), ציר ה-$y$ מציג **מספר מדידות גולמי** בכל תא.

בדוגמה מופיע גם ‎`edgecolor`‎ (צבע קו המתאר של כל עמודה), ובתרגול הנוסף ‎`alpha`‎ (שקיפות). אלה פרמטרי תצוגה, כמו אלה שבסעיף 8.2.

### דוגמה: פיזור הטווח עקב אי-ודאות במהירות השיגור

בפועל, מהירות השיגור $v_0$ בכל זריקה אינה קבועה לגמרי — יש רעש מדידה/ביצוע. נדמה 500 "זריקות" עם $v_0$ מפוזר סביב $20\pm 1$ מ/ש, ונציג את התפלגות הטווח המתקבל. גם כאן הנתונים מוכנים מראש: המערך ‎`v0_trials`‎ מכיל 500 מהירויות שיגור מדומות סביב 20 מ/ש, בפיזור (סטיית תקן) של 1.

```````python
import numpy as np

# 500 simulated launch speeds (m/s) around 20 with spread 1, given ready
v0_trials = np.random.default_rng(1).normal(20.0, 1.0, 500)
```````

```````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
theta = np.radians(45)

R_trials = v0_trials**2 * np.sin(2 * theta) / g   # v0_trials: given ready

fig, ax = plt.subplots()
ax.hist(R_trials, bins=25, density=True, color="tab:blue", edgecolor="white")
ax.set_xlabel("Range R (m)")
ax.set_ylabel("Probability density")
ax.set_title("Range distribution from launch-speed uncertainty")
plt.show()
```````

### נסו בעצמכם: מספר העמודות וגובהן

עשר זריקות נמדדו, והטווחים (במטרים) נשמרו במערך ‎`R_meas`‎. **בלי להריץ**, ענו:

1. כמה עמודים תצייר ההיסטוגרמה?
2. מה הגובה של כל עמוד?
3. מה יהיה הגובה של כל עמוד עם ‎`density=True`‎? (עם ‎`density=True`‎ גובה העמוד הוא מספר המדידות בתא, חלקי מספר כל המדידות, חלקי רוחב התא. כך השטח הכולל של העמודים הוא 1.)

אחר כך הריצו ובדקו.

```````python
# ranges of 10 throws (m)
R_meas = np.array([36.5, 38.2, 39.0, 39.4, 40.1, 40.6, 41.3, 41.9, 42.5, 44.0])

fig, ax = plt.subplots()
ax.hist(R_meas, bins=[36, 38, 40, 42, 44], edgecolor="white")
ax.set_xlabel("Range R (m)")
ax.set_ylabel("Number of throws")
plt.show()
```````

### באג נפוץ: תווית ‎`"צפיפות"`‎ כש-‎`density=False`‎

אם משאירים את התווית ‎`"צפיפות הסתברות"`‎ על ציר $y$ בזמן שקוראים ל-‎`ax.hist`‎ עם ‎`density=False`‎ (ברירת המחדל) — התווית שגויה: הערכים בפועל הם ספירה גולמית, לא צפיפות. יש לוודא תמיד שהתווית תואמת את הפרמטר בפועל.
