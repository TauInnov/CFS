- ‎`plt.subplots(...)`‎ קובע כמה צירים יש וכיצד הם מסודרים.
- כל קריאה ל-‎`plot`‎, ‎`scatter`‎, ‎`errorbar`‎ או ‎`hist`‎ מוסיפה לציר שלה דבר אחד: קו, נקודות נפרדות, נקודות עם קווי שגיאה או עמודות. הארגומנטים ‎`linestyle`‎, ‎`marker`‎ ו-‎`fmt`‎ קובעים כיצד הוא נראה.

### משימה 1: הגרף שהקוד מצייר

הקוד הבא משווה את הטווח שהמודל $R = v_0^2 \sin(2\theta)/g$ מנבא לטווחים שנמדדו בשלוש זוויות. קראו אותו בלי להריץ וענו:

א. מה תדפיס הפקודה ‎`print(angles)`‎ אחרי שורה 5? מהו הערך הגדול ביותר במערך ‎`R_model`‎, ובאיזו זווית?

ב. מה כתוב על כל ציר, מה מצויר על הגרף, ומה יופיע במקרא? בין אילו ערכים נמתח קו השגיאה של המדידה ב-45°?

ג. איזה מארבעת השרטוטים שמתחת לקוד הוא הגרף שהקוד מצייר?

ד. משנים את שורה 5 ל-‎`angles = np.linspace(0, 90, 4)`‎. מהו עכשיו הערך הגדול ביותר בקו המקווקו, ובאילו זוויות?

``````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 10.0, 20.0
angles = np.linspace(0, 90, 7)
R_model = v0**2 * np.sin(np.radians(2 * angles)) / g

angles_meas = np.array([15.0, 45.0, 75.0])
R_meas = np.array([21.0, 39.0, 19.0])
R_err = np.array([1.0, 2.0, 1.0])

fig, ax = plt.subplots()
ax.plot(angles, R_model, linestyle="--", label="model")
ax.errorbar(angles_meas, R_meas, yerr=R_err, fmt="o", label="measured")
ax.set_xlabel("launch angle (degrees)")
ax.set_ylabel("range (m)")
ax.legend()
plt.show()
``````

``````python
# Four sketches for task 1, part c (this code is hidden on the page).
fig, axs = plt.subplots(2, 2, figsize=(9, 6.5), sharex=True, sharey=True,
                        constrained_layout=True)
for i in range(4):
    ax = axs[i // 2, i % 2]
    if i == 2:
        R_wrong = v0**2 * np.sin(np.radians(angles)) / g
        ax.plot(angles, R_wrong, linestyle="--", label="model")
    else:
        style = "-" if i == 0 else "--"
        ax.plot(angles, R_model, linestyle=style, label="model")
    if i == 3:
        ax.plot(angles_meas, R_meas, marker="o", label="measured")
    else:
        ax.errorbar(angles_meas, R_meas, yerr=R_err, fmt="o", label="measured")
    ax.set_title("ABCD"[i])
    ax.legend(loc="lower center")
for c in range(2):
    axs[1, c].set_xlabel("launch angle (degrees)")
for r in range(2):
    axs[r, 0].set_ylabel("range (m)")
plt.show()
``````

### משימה 3: מצאו את שלוש הטעויות

הקוד הבא אמור לצייר את מסלולי הזריקה בשלוש זוויות **על גרף אחד**, עם מקרא שמבדיל בין הזוויות. הוא רץ בלי שגיאה, אבל מתקבלות שלוש תמונות נפרדות, מקרא מופיע רק באחרונה, והעקומה של 30° נוחתת ב-x שלילי, משמאל לנקודת הזריקה.

**לפני שמריצים**: מצאו את שלוש הטעויות לפי מספרי השורות, תקנו כל אחת והסבירו במשפט אחד. שימו לב שטעות אחת יכולה להסתיר טעות אחרת.

``````python
g, v0 = 10.0, 20.0
angles_deg = [30.0, 45.0, 60.0]

for angle in angles_deg:
    fig, ax = plt.subplots()
    theta = angle
    t = np.linspace(0, 2 * v0 * np.sin(theta) / g, 50)
    x = v0 * np.cos(theta) * t
    y = v0 * np.sin(theta) * t - 0.5 * g * t**2
    ax.plot(x, y, label="trajectory")

ax.legend()
plt.show()
``````
