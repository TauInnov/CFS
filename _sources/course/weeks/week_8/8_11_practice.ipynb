## 8.11 — תרגול מסכם

בלי להריץ, תארו במילים מה מצייר הקוד הבא, וכמה עקומות יופיעו על כל אחד משני הצירים:

``````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0
angles_deg = [30, 60]

fig, axs = plt.subplots(1, 2, figsize=(10, 4), sharey=True)
for ax, angle in zip(axs, angles_deg):
    theta = np.radians(angle)
    t_flight = 2 * v0 * np.sin(theta) / g
    t = np.linspace(0, t_flight, 100)
    x = v0 * np.cos(theta) * t
    y = v0 * np.sin(theta) * t - 0.5 * g * t**2
    ax.plot(x, y, label=fr"$\theta={angle}^\circ$")
    ax.legend()
``````

`````{admonition} פתרון
:class: dropdown, tip
הקוד יוצר figure עם **שני צירים זה לצד זה** (`1×2`), שחולקים את אותו ציר $y$ (`sharey=True`). על כל ציר מצוירת **עקומה אחת** — מסלול הזריקה עבור זווית אחת (30° בציר השמאלי, 60° בציר הימני), עם מקרא שמציג את הזווית באמצעות $\theta$. שני הצירים לא מוצגים על אותו גרף — זו טעות נפוצה לבלבל בין `subplots(1, 2)` (שני גרפים נפרדים) לבין לולאה שמציירת הכול על `ax` יחיד (סעיף 8.5).
`````

בקוד הבא יש כוונה לצייר את הטווח הנמדד עם קווי שגיאה מול התיאוריה. יש בו באג — אתרו ותקנו:

``````python
theta = np.radians(45)
angles_deg = np.array([15, 30, 45, 60, 75])

rng = np.random.default_rng(3)
means, stds = [], []
for angle in angles_deg:
    theta_i = np.radians(angle)
    v0_trials = rng.normal(20.0, 1.0, size=200)
    R_trials = v0_trials**2 * np.sin(2 * theta_i) / g
    means.append(R_trials.mean())
    stds.append(R_trials.std())

fig, ax = plt.subplots()
ax.errorbar(angles_deg, means, yerr=stds, fmt="-", color="tab:red")   # <- שימו לב כאן
ax.set_xlabel("זווית שיגור (מעלות)")
ax.set_ylabel("טווח R (m)")
plt.show()
``````

`````{admonition} פתרון
:class: dropdown, tip
הבאג הוא ב-`fmt="-"`: זה מחבר את נקודות המדידה בקו רציף, כאילו יש מודל רציף בין הזוויות שנמדדו בפועל — בניגוד לעקרון שראינו בסעיף 8.9, שלפיו נקודות מדידה בדידות מוצגות כסמלים בלי קו מחבר.
`````
