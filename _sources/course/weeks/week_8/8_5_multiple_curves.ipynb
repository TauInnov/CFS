## כמה עקומות באותו גרף

השוואה בין כמה תרחישים פיזיקליים נעשית בדרך כלל על אותו גרף, לא בגרפים נפרדים. הדפוס הנפוץ: לולאה שבכל איטרציה מוסיפה עקומה נוספת לאותו `ax`, כשלכל עקומה `label` משלה — כדי שהמקרא יבדיל ביניהן.

### דוגמה מובילה: מסלול הזריקה עבור שלוש זוויות שיגור

זו הדוגמה שתלווה את שאר השבוע: אותה מהירות התחלתית $v_0$, שלוש זוויות שיגור שונות — כולן על אותו גרף.

``````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
angles_deg = [30, 45, 60]

fig, ax = plt.subplots()
for angle in angles_deg:
    theta = np.radians(angle)
    t_flight = 2 * v0 * np.sin(theta) / g
    t = np.linspace(0, t_flight, 200)
    x = v0 * np.cos(theta) * t
    y = v0 * np.sin(theta) * t - 0.5 * g * t**2
    ax.plot(x, y, label=fr"$\theta={angle}^\circ$")

ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
ax.set_title("מסלול הזריקה עבור שלוש זוויות שיגור")
ax.legend()
plt.show()
``````

### באג נפוץ: label קבוע בתוך לולאה

אם כותבים `ax.plot(x, y, label="זריקה")` בתוך הלולאה — כלומר אותו `label` בכל איטרציה — המקרא יציג את אותה תווית שלוש פעמים, במקום להבדיל בין הזוויות. הפתרון: לבנות את המחרוזת דינמית מתוך משתנה הלולאה, כמו ב-`f"..." / fr"..."` שלמעלה.
