## plot ו-scatter: קווים, נקודות וסגנון

ב-Matplotlib יש שתי פונקציות ציור עיקריות שרלוונטיות כמעט לכל דוח מעבדה:

- **`ax.plot(x, y)`** — מתאימה ל**עקומה רציפה**: מודל תיאורטי, פונקציה מחושבת על רשת צפופה של נקודות.
- **`ax.scatter(x, y)`** — מתאימה ל**נקודות בדידות**: מדידות בפועל, שאין ביניהן קשר רציף מובנה.

ההבחנה הזו אינה טכנית בלבד — היא משקפת הבדל פיזיקלי: בדוח מעבדה מקובל לצייר את המודל התיאורטי כקו רציף, ואת המדידות כנקודות נפרדות עליו.

### דוגמה: מודל תיאורטי מול "מדידות" עם רעש

ניצור את מסלול הזריקה $y(x)$ (גובה כתלות במרחק אופקי) עבור $\theta=45^\circ$ כעקומה רציפה, ולצידה כמה נקודות "מדידה" על אותו מסלול, עם רעש אקראי קטן שמדמה אי-דיוק של מכשיר מדידה.

``````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
rng = np.random.default_rng(0)
theta = np.radians(45)
t_flight = 2 * v0 * np.sin(theta) / g

t_full = np.linspace(0, t_flight, 200)
x_full = v0 * np.cos(theta) * t_full
y_full = v0 * np.sin(theta) * t_full - 0.5 * g * t_full**2

t_meas = np.linspace(0, t_flight, 8)
x_meas = v0 * np.cos(theta) * t_meas
y_meas = v0 * np.sin(theta) * t_meas - 0.5 * g * t_meas**2 + rng.normal(0, 0.3, size=t_meas.size)

fig, ax = plt.subplots()
ax.plot(x_full, y_full, color="tab:blue", label="מודל תיאורטי")
ax.scatter(x_meas, y_meas, color="tab:red", marker="o", label='"מדידות" (עם רעש)')
ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")
ax.legend()
plt.show()
``````

פרמטרי הסגנון הנפוצים: ב-`plot` — `color`, `linestyle` (`"-"`, `"--"`, `":"`), `linewidth`; ב-`scatter` — `color`, `marker` (`"o"`, `"^"`, `"s"`...), `s` (גודל הנקודה).
