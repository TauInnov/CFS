## השוואות ומסכות בוליאניות, where, any, all

אופרטורי השוואה (‎`>`‎, ‎`<`‎, ‎`==`‎ וכו') על מערך NumPy אינם מחזירים ערך יחיד — הם מחזירים **מערך בוליאני** באותו גודל, עם ‎`True`‎/‎`False`‎ לכל איבר. מערך כזה נקרא **מסכה (mask)**, ואפשר להשתמש בו כדי לסנן את המערך המקורי: ‎`a[mask]`‎ מחזיר רק את האיברים שבהם המסכה ‎`True`‎.

ארבע פעולות שימושיות על מסכות:
- ‎`np.where(mask)`‎ — מחזיר את **האינדקסים** שבהם המסכה ‎`True`‎.
- ‎`np.any(mask)`‎ — ‎`True`‎ אם **לפחות** איבר אחד במסכה הוא ‎`True`‎.
- ‎`np.all(mask)`‎ — ‎`True`‎ רק אם **כל** האיברים במסכה הם ‎`True`‎.
- ‎`mask.sum()`‎ — **כמה** איברים במסכה הם ‎`True`‎: כל ‎`True`‎ נספר כ-1, וכל ‎`False`‎ כ-0.

סרטון: **"מסכות בוליאניות עם & ו-|: נוסחה שמוגדרת בקטעים"** — הקלטת מסך של המחברת עם קול.

### עיקרי הסרטון

- כל מקרה של הנוסחה הופך למסכה: ‎`(t >= 0) & (t < 4)`‎.
- משתמשים ב-‎`&`‎ וב-‎`|`‎ עם סוגריים סביב כל השוואה. ‎`and`‎, ‎`or`‎ ו-‎`0 <= t < 4`‎ נכשלים על מערכים.
- בודקים שהמקרים מכסים כל נקודה ואינם חופפים: ‎`np.all(m1 | m2 | m3)`‎, ‎`np.any(m1 & m2)`‎.
- בגבולות (‎`<`‎ או ‎`<=`‎) קל לטעות בלי שתתקבל שגיאה.

### דוגמה: מציאת רגע הנחיתה

עבור מסלול הזריקה $y(t) = v_0 t - \tfrac{1}{2} g t^2$, הביטוי הזה ממשיך להיות מוגדר מתמטית גם לאחר שהכדור פגע בקרקע ($y<0$) — אבל פיזיקלית זה חסר משמעות. נשתמש במסכה כדי לזהות ולחתוך את החלק הרלוונטי בלבד.

``````python
import numpy as np

g, v0 = 9.8, 15.0
t = np.linspace(0, 4, 40)
y = v0 * t - 0.5 * g * t**2

above_ground = y >= 0
print("מסכה (10 ראשונים):", above_ground[:10])
print("יש נקודות מתחת לקרקע?", np.any(y < 0))
print("כל הנקודות מעל הקרקע?", np.all(above_ground))
``````

``````python
below = np.where(y < 0)[0]   # indices where the ball is below ground
landing_idx = below[0]       # the first of them
print("אינדקס הנחיתה:", landing_idx, " t =", t[landing_idx])

t_flight = t[above_ground]
y_flight = y[above_ground]
``````

``````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot(t, y, "--", color="lightgray", label="Formula (all t)")
ax.plot(t_flight, y_flight, color="tab:blue",
        label="Physical trajectory (y ≥ 0)")
ax.axhline(0, color="black", linewidth=0.8)
ax.set_xlabel("t (s)")
ax.set_ylabel("y (m)")
ax.legend()
plt.show()
``````

### שני תנאים יחד

כדי לבחור נקודות שמקיימות שני תנאים בבת אחת, מחברים שתי מסכות באופרטור ‎`&`‎ ("וגם"), וכל השוואה עוטפים בסוגריים: ‎`(y > 5) & (t < 2)`‎. בלי הסוגריים פייתון מחשב את ‎`&`‎ לפני ההשוואות, והקוד נכשל; המילה ‎`and`‎ לא עובדת על מערכים. באותו אופן, האופרטור ‎`|`‎ ("או") מחבר שתי מסכות ובוחר את הנקודות שמקיימות לפחות אחד מהתנאים. כמה נקודות במסלול נמצאות מעל 5 מטרים בשתי השניות הראשונות?

``````python
rising_high = (y > 5) & (t < 2)   # above 5 m and before t = 2 s
print(rising_high.sum())          # each True counts as 1
``````
