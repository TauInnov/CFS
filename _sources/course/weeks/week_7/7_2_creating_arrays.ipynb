## יצירת מערכים: array, zeros, ones, arange, linspace

מלבד בנייה מתוך רשימה קיימת (‎`np.array`‎), ל-NumPy יש כמה פונקציות ליצירת מערכים ישירות:

- ‎`np.zeros(n)`‎ — מערך של n אפסים, שימושי כ"מכל" ריק שממלאים בהמשך.
- ‎`np.ones(n)`‎ — מערך של n אחדים.
- ‎`np.arange(start, stop, step)`‎ — כמו ‎`range`‎ בפייתון, אך מחזיר מערך. **הקצה העליון לא נכלל**.
- ‎`np.linspace(start, stop, num)`‎ — num נקודות במרווחים שווים בין start ל-stop, **כולל שתי הקצוות**.

ההבדל בין ‎`arange`‎ ל-‎`linspace`‎ הוא מקור נפוץ לטעויות "בגובה אחד" (off-by-one).

### דוגמה: וקטור זמן לזריקה

בסעיף הקודם בנינו את וקטור הזמן של הזריקה כרשימה קבועה מראש. הפעם נבנה אותו ישירות עם ‎`linspace`‎, ונחשב את הגובה בכל רגע בביטוי אחד על המערך.

```````python
import numpy as np

g = 9.8
y0, v0 = 0.0, 15.0

t_arr = np.linspace(0, 3, 10)   # 10 points from 0 to 3, both ends included

y_arr = y0 + v0 * t_arr - 0.5 * g * t_arr**2
print(t_arr)
print(y_arr)
```````

### הדפסת מערך ב-NumPy

הפלט של ‎`print`‎ על מערך נראה אחרת מהפלט על רשימה, וכדאי להכיר אותו: בשאלות "מה יודפס" צריך לכתוב אותו בדיוק.

- **אין פסיקים** בין האיברים, רק רווחים.
- מערך של מספרים **שלמים** מודפס בלי נקודה עשרונית.
- במערך של מספרים **עשרוניים** (‎`float`‎) כל מספר מודפס עם נקודה, גם מספר שלם: 1 מודפס כ-‎`1.`‎, עם נקודה ובלי אפס אחריה.
- NumPy מוסיף רווחים כדי שכל המספרים יתפסו אותו רוחב, ולכן לפעמים יש שני רווחים בין מספרים, או רווח אחרי הסוגר הפותח.

שלוש דוגמאות:

```````python
a = np.array([1, 2, 3])          # integers
b = np.array([0.5, 1.0, 2.25])   # floats
c = np.array([1, 2, 3])          # integers again
c[0] = 2.7                       # a float stored in an integer array
print(a)
print(b)
print(c)
```````

השורה השלישית מפתיעה: ‎`c`‎ נוצר ממספרים שלמים, ולכן הוא נשאר מערך של שלמים גם אחרי ההשמה, ו-2.7 נשמר בו כ-2. נחזור לזה בסעיף הבא.

### arange מול linspace

נשווה ישירות בין שתי הפונקציות על אותו טווח, כדי לראות את ההבדל בהתנהגות בקצה העליון:

```````python
a = np.arange(0, 10, 2)     # 10 is not included
b = np.linspace(0, 10, 5)    # 10 is included
print(a)
print(b)
```````

```````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 1.5))
ax.plot(a, np.zeros(a.shape), 'o', label="arange(0,10,2)")
ax.plot(b, np.ones(b.shape), 'o', label="linspace(0,10,5)")
ax.set_yticks([0, 1])
ax.set_yticklabels(["arange", "linspace"])
ax.set_xlabel("Value")
ax.legend(loc="upper right", bbox_to_anchor=(1.3, 1))
plt.show()
```````
