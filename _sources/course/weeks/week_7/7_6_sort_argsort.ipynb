## sort ו-argsort

`np.sort(a)` מחזיר עותק ממוין של המערך. לעיתים קרובות יש לנו כמה מערכים "מקבילים" — לכל איבר במערך אחד יש איבר מתאים באחר (למשל: זמן מדידה וטמפרטורה, לפי אותו סדר ניסויים) — ורוצים למיין את כולם לפי אותו סדר. לשם כך משמש `np.argsort(a)`, שמחזיר לא את הערכים הממוינים אלא את **האינדקסים** שיסדרו את המערך.

### דוגמה: גבהי קפיצה במספר ניסויים

מדדנו את הגובה המרבי שכדור הגיע אליו בחמישה ניסויי זריקה, וגם רשמנו את חותמת הזמן (בשניות מתחילת היום) שבה כל ניסוי בוצע. רוצים להציג את הניסויים ממוינים לפי גובה — אך לשמור על ההתאמה לחותמות הזמן.

``````python
import numpy as np
import matplotlib.pyplot as plt

heights = np.array([11.2, 8.9, 14.7, 9.8, 12.5])
timestamps = np.array([540, 555, 561, 570, 588])   # דקה מתחילת המדידות

order = np.argsort(heights)
print("סדר האינדקסים לפי גובה עולה:", order)

heights_sorted = heights[order]
timestamps_sorted = timestamps[order]

print("גבהים ממוינים:", heights_sorted)
print("חותמות זמן בהתאמה:", timestamps_sorted)
``````

שימו לב: `timestamps[order]` ולא `np.sort(timestamps)` — אילו מיינו כל מערך בנפרד היינו מאבדים את ההתאמה בין ניסוי לניסוי. נציג את זה גרפית:

``````python
fig, ax = plt.subplots()
ax.bar(range(len(heights_sorted)), heights_sorted,
       tick_label=[f"t={ts}" for ts in timestamps_sorted])
ax.set_ylabel("Max height (m)")
ax.set_xlabel("Trial (by timestamp)")
ax.set_title("גבהי קפיצה, ממוינים מהנמוך לגבוה")
plt.show()
``````
