`````{admonition} פתרון
```python
def model(T0, k, minutes):
    """Temperature (deg C) after minutes; minutes can be an array."""
    return 20 + (T0 - 20) * (1 - k)**minutes


def chi2_of(k, T0, minutes, T_meas, dT):
    """chi2 of the model with T0 and k against T_meas with errors dT."""
    return np.sum(((T_meas - model(T0, k, minutes)) / dT)**2)


def best_k(k_values, T0, minutes, T_meas, dT):
    """The value in k_values with the smallest chi2."""
    best = k_values[0]
    best_chi2 = chi2_of(best, T0, minutes, T_meas, dT)
    for k in k_values:
        c = chi2_of(k, T0, minutes, T_meas, dT)
        if c < best_chi2:
            best = k
            best_chi2 = c
    return best


print(model(60.0, 0.5, minutes_test))   # [60. 40. 30.]
print(chi2_of(0.5, 60.0, minutes_test, T_test, 1.0))   # 2.0
print(best_k([0.4, 0.5, 0.6], 60.0, minutes_test, T_test, 1.0))   # 0.5
```
- **א.** עם ‎`k = 0.5`‎ ההפרש מטמפרטורת החדר, 40 בהתחלה, מוכפל ב-0.5 בכל דקה: $20 + 40 = 60$, אחר כך $20 + 20 = 40$ ואחר כך $20 + 10 = 30$. החזקה ‎`(1 - k)**minutes`‎ פועלת על כל איבר של המערך ‎`minutes`‎, ולכן אותה שורה עובדת גם על מספר וגם על מערך.

- **ג.** עבור ‎`k = 0.4`‎ מתקבל χ² של 38.16, עבור ‎`k = 0.5`‎ של 2, ועבור ‎`k = 0.6`‎ של 31.76; הקטן ביותר הוא של ‎`k = 0.5`‎. הלולאה שומרת ב-‎`best`‎ את הערך עם χ² הקטן ביותר עד כה, כמו הסריקה בסעיף 12.2. דרך אחרת היא לבנות מערך של כל ערכי χ² ולבחור בו את המיקום של הקטן ביותר עם ‎`argmin`‎, כמו בסריקה במערך בסעיף 12.2:

```python
def best_k(k_values, T0, minutes, T_meas, dT):
    """The value in k_values with the smallest chi2."""
    chi2_all = np.array([chi2_of(k, T0, minutes, T_meas, dT)
                         for k in k_values])
    return k_values[chi2_all.argmin()]
`````
