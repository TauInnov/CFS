## סקאלה לוגריתמית ולינאריזציה

מסלול הזריקה אינו דורש סולם לוגריתמי — הערכים בו נמצאים כולם באותו סדר גודל. אבל כשגודל פיזיקלי משתנה על פני **כמה סדרי גודל**, סולם ליניארי "דוחס" את הערכים הקטנים לכדי קו אחד צמוד לציר, והגרף הופך חסר תועלת. הפתרון הוא סולם לוגריתמי:

```python
ax.set_xscale("log")
ax.set_yscale("log")
```

אפשר להפעיל את שניהם יחד (log-log), רק אחד מהם, או אף אחד — תלוי בטווח הערכים בכל ציר בנפרד.

בחלקו השני של העמוד הרעיון של סולם לוגריתמי מוביל ל**לינאריזציה**: הפיכת קשר לא-ליניארי לקו ישר, שאליו אפשר להתאים קו עם ‎`linear_fit`‎.

### דוגמה: טווח הזריקה כתלות במהירות ההתחלתית

הטווח נתון על ידי $R(v_0) = v_0^2 \sin(2\theta)/g$ — תלות **ריבועית** ב-$v_0$. אם נסרוק $v_0$ על פני טווח רחב (למשל בין 1 ל-1000 מ/ש, כתרחיל מתמטי גרידא, לא ערכים פיזיקליים סבירים לזריקת יד), $R$ ישתנה על פני כמה סדרי גודל, ונשווה תצוגה ליניארית מול log-log.

```````python
import numpy as np
import matplotlib.pyplot as plt

g, v0 = 9.8, 20.0          # m/s^2, m/s
theta = np.radians(45)
v0_range = 10**np.linspace(0, 3, 200)   # 1 ... 1000, log-spaced
R_range = v0_range**2 * np.sin(2 * theta) / g

fig, axs = plt.subplots(1, 2, figsize=(10, 4))

axs[0].plot(v0_range, R_range)
axs[0].set_title("Linear scale")
axs[0].set_xlabel(r"$v_0$ (m/s)")
axs[0].set_ylabel("R (m)")

axs[1].plot(v0_range, R_range)
axs[1].set_xscale("log")
axs[1].set_yscale("log")
axs[1].set_title("Log-log scale")
axs[1].set_xlabel(r"$v_0$ (m/s)")
axs[1].set_ylabel("R (m)")

plt.show()
```````

בגרף הליניארי (משמאל) כמעט כל המידע על ערכי $v_0$ הקטנים "נעלם" צמוד לציר. בגרף ה-log-log (מימין) רואים קו ישר — תוצאה של חוק חזקה ($R\propto v_0^2$). השיפוע של הקו הישר הוא החזקה, 2: כשה-$v_0$ גדל פי 10, הטווח גדל פי $10^2 = 100$.

חשוב: הביטוי ‎`10**np.linspace(0, 3, 200)`‎ פורש 200 **מעריכים** במרווחים שווים בין 0 ל-3, והחזקה הופכת אותם לערכים בין $10^0$ ל-$10^3$. כך הנקודות פרוסות באופן אחיד **בסולם הלוגריתמי**, ולא באופן אחיד בין 1 ל-1000.

### דוגמה נוספת: דעיכה בסולם לוגריתמי בציר אחד

בגוף המתקרר מסעיף 8.1, ההפרש מטמפרטורת החדר, $T - 20$, מוכפל ב-$(1-k)$ בכל דקה. בסולם לוגריתמי **בציר y בלבד** דעיכה כזו נראית כקו ישר. זה ההבדל מהדוגמה הקודמת: חוק חזקה נראה ישר בסולם log-log, ודעיכה מעריכית נראית ישרה כשרק ציר y לוגריתמי.

```````python
T0, k = 80.0, 0.1                # initial temperature (°C), cooling constant
minutes = np.arange(0, 61)       # 0, 1, ..., 60 min
gap = (T0 - 20) * (1 - k)**minutes   # T - 20: distance from the room (°C)

fig, axs = plt.subplots(1, 2, figsize=(10, 4))
axs[0].plot(minutes, gap)
axs[0].set_title("Linear scale")
axs[0].set_xlabel("t (min)")
axs[0].set_ylabel("T - 20 (°C)")
axs[1].plot(minutes, gap)
axs[1].set_yscale("log")
axs[1].set_title("Log scale on y only")
axs[1].set_xlabel("t (min)")
axs[1].set_ylabel("T - 20 (°C)")
plt.show()
```````

### לינאריזציה: הגוף המתקרר דרך ln

‎`linear_fit`‎ מתאימה רק **קווים ישרים**, והטמפרטורה של הגוף המתקרר אינה קו ישר בזמן:

$$T_n = 20 + (T_0 - 20)(1-k)^n$$

מעבירים את 20 לאגף שמאל ומפעילים $\ln$ על שני האגפים:

$$\ln(T_n - 20) = \ln(T_0 - 20) + n\ln(1-k)$$

זהו **קו ישר** ב-$n$: החיתוך הוא $a_0 = \ln(T_0 - 20)$ והשיפוע הוא $a_1 = \ln(1-k)$. מהחיתוך ומהשיפוע מחזירים את הפרמטרים הפיזיקליים: $k = 1 - e^{a_1}$ ו-$T_0 = 20 + e^{a_0}$. הפיכת קשר לא-ליניארי לקו ישר על ידי שינוי משתנה נקראת **לינאריזציה**. מאותה סיבה הדעיכה נראתה ישרה בסולם לוגריתמי בציר y: סולם כזה ממקם כל ערך לפי הלוגריתם שלו.

הנתונים: הטמפרטורה נמדדה כל 2 דקות, מ-0 עד 30 דקות. המערך ‎`minutes`‎ מכיל את זמני המדידה בדקות, והמערך ‎`T_meas`‎ את הטמפרטורות שנמדדו. הנתונים נוצרו מהמודל עם ‎`T0 = 60`‎ ו-‎`k = 0.1`‎, ועם רעש מדידה קטן, כדי שאפשר יהיה להשוות את תוצאת ההתאמה לערכים שמהם נוצרו.

```````python
# Measured temperatures: the model 20 + (T0 - 20)*(1 - k)**minutes
# with T0 = 60, k = 0.1, plus small fixed deviations (measurement noise)
minutes = np.arange(0, 31, 2)   # 0, 2, ..., 30 min
noise = np.array([0.2, -0.3, 0.1, 0.3, -0.2, 0.1, -0.1, 0.2,
                  -0.2, 0.1, 0.2, -0.1, 0.1, -0.2, 0.1, -0.1])
T_meas = 20 + (60 - 20) * (1 - 0.1)**minutes + noise
```````

```````python
T0_true = 60.0   # the values used to make the data
k_true = 0.1
print("minutes:", minutes)
print("T_meas: ", np.round(T_meas, 1))
```````

### לינאריזציה והתאמה

```````python
def linear_fit(x, y):
    x_bar, y_bar = x.mean(), y.mean()
    a1 = ((x - x_bar) * (y - y_bar)).sum() / ((x - x_bar)**2).sum()
    a0 = y_bar - a1 * x_bar
    return a0, a1

y_lin = np.log(T_meas - 20)   # ln(T - 20)
a0, a1 = linear_fit(minutes, y_lin)

k_fit = 1 - np.exp(a1)
T0_fit = 20 + np.exp(a0)
print(f"k  = {k_fit:.3f}   (true: {k_true})")
print(f"T0 = {T0_fit:.1f}   (true: {T0_true})")
```````

```````python
fig, ax = plt.subplots()
ax.plot(minutes, y_lin, "o", label="Measurement")
ax.plot(minutes, a0 + a1*minutes, "-", label="Linear fit")
ax.set_xlabel("t (min)")
ax.set_ylabel("ln(T - 20)")
ax.legend()
plt.show()
```````

הנקודות מסודרות סביב הקו המותאם. בסוף המדידה ההפרש ‎`T - 20`‎ קטן, ולכן אותו רעש של עשיריות מעלה משנה את ‎`ln(T - 20)`‎ יותר מאשר בתחילתה.

### באג נפוץ: ln של אפס או של ערך שלילי

קרוב לטמפרטורת החדר ההפרש ‎`T - 20`‎ קטן, ורעש המדידה יכול להוציא אותו אפס או שלילי. על ערכים כאלה ‎`np.log`‎ **אינו זורק שגיאה**: עבור ערך שלילי הוא מחזיר ‎`nan`‎, ועבור אפס הוא מחזיר ‎`-inf`‎, עם ‎`RuntimeWarning`‎ שקל לפספס, והריצה ממשיכה. ערך כזה מזהם את ‎`linear_fit`‎ כולה: שני הפרמטרים ‎`a0`‎ ו-‎`a1`‎ יוצאים ‎`nan`‎, בלי שגיאה שמסבירה מדוע. בדוגמה הבאה הערך האחרון, בעותק של הנתונים, נמוך מטמפרטורת החדר.

```````python
T_bad = T_meas.copy()
T_bad[-1] = 19.8   # the last value came out below the room temperature

y_lin_bad = np.log(T_bad - 20)   # RuntimeWarning, not an error
print(np.round(y_lin_bad, 2))

a0_bad, a1_bad = linear_fit(minutes, y_lin_bad)
print("a0 =", a0_bad, "  a1 =", a1_bad)   # nan spreads to the whole fit
```````
