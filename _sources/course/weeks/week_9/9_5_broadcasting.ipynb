## התאמת צירים בין שני מערכים (כלל 3)

בסעיף 7.8 ראינו broadcasting בין מערך לסקלר בודד. עכשיו מכללים את זה לכלל המלא — **כלל 3** מארבעת הכללים:

```{admonition} כלל 3 — Broadcasting
:class: tip
שני מערכים מתחברים אם, **מימין לשמאל**, כל זוג צירים מתאימים או **שווה בגודלו**, או **שאחד מהם 1**.
```

הכלל הזה מאפשר לבנות בדיוק את `Y` שבנינו בלולאה בסעיף 9.1 — בלי לולאה בכלל.

``````python
import numpy as np

g = 9.8
theta = np.radians(45)
v0_nom = 20.0
rng = np.random.default_rng(0)
n_runs, n_t = 5, 60
v0_runs = rng.normal(v0_nom, 1.0, size=n_runs)     # shape (5,)
t_flight_nom = 2 * v0_nom * np.sin(theta) / g
t = np.linspace(0, t_flight_nom, n_t)              # shape (60,)
``````

### דוגמה: בניית `Y` בפעולה וקטורית אחת

`v0_runs` בצורה `(5,)` ו-`t` בצורה `(60,)` — צורות שונות, אי אפשר לחבר ישירות. אבל אם מוסיפים ל-`v0_runs` ציר בגודל 1 (`v0_runs[:, None]`, צורה `(5, 1)`), הכלל מתקיים: מימין לשמאל, `60` מול `1` (אחד מהם 1 ✓), ו-`5` מול `1` הנסתר של `t` (אחד מהם 1 ✓).

``````python
Y = v0_runs[:, None] * np.sin(theta) * t - 0.5 * g * t**2
print("Y.shape:", Y.shape)   # (5, 60) - בלי אף לולאה

# בדיקת שפיות מול הבנייה בלולאה מ-9.1:
Y_loop = np.empty((n_runs, n_t))
for i in range(n_runs):
    Y_loop[i] = v0_runs[i] * np.sin(theta) * t - 0.5 * g * t**2

print(np.allclose(Y, Y_loop))
``````

### באג נפוץ: לשכוח את `[:, None]`

בלי `[:, None]`, `v0_runs` נשאר בצורה `(5,)` — ו-NumPy מנסה להתאים אותו ל-`t` בצורה `(60,)` ישירות: `5` מול `60`, שניהם שונים מ-1. זה **לא** באג שקט — NumPy יזרוק `ValueError` באופן מיידי, מה שדווקא נוח: השגיאה מצביעה בדיוק על הבעיה.

``````python
try:
    bad = v0_runs * np.sin(theta) * t
except ValueError as e:
    print("שגיאה כצפוי:", e)
``````
