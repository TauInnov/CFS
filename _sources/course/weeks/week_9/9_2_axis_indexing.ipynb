## 9.2 — פריסה לפי צירים: `a[0]`, `a[:,2]`, `a[...,0]`

(המשך הסרטון מ-9.1.) עכשיו שיש למערכים כמה צירים, צריך דרך לבחור פרוסה מתוכם:

- **`a[0]`** — כל מה שנמצא במיקום 0 של **הציר הראשון**, שאר הצירים נשארים במלואם.
- **`a[:, 2]`** — כל הערכים בציר הראשון, אבל רק מיקום 2 בציר השני.
- **`a[..., 0]`** — שלוש נקודות (`...`) מסמנות "כל שאר הצירים כמו שהם", ובוחרות מיקום 0 רק בציר **האחרון**. נוח כשלא רוצים לספור כמה צירים יש לפני האחרון.

``````python
import numpy as np
import matplotlib.pyplot as plt

g = 9.8
theta = np.radians(45)
v0_nom = 20.0
rng = np.random.default_rng(0)
n_runs, n_t = 5, 60
v0_runs = rng.normal(v0_nom, 1.0, size=n_runs)
t_flight_nom = 2 * v0_nom * np.sin(theta) / g
t = np.linspace(0, t_flight_nom, n_t)

Y = np.empty((n_runs, n_t))
X = np.empty((n_runs, n_t))
for i in range(n_runs):
    Y[i] = v0_runs[i] * np.sin(theta) * t - 0.5 * g * t**2
    X[i] = v0_runs[i] * np.cos(theta) * t
XY = np.stack([X, Y], axis=-1)   # shape (n_runs, n_t, 2)
``````

### דוגמה

``````python
run2_full_trajectory = Y[1]          # מסלול הריצה השנייה (index 1), כל הזמן
all_runs_at_t2 = Y[:, 2]             # הגובה בכל הריצות, בנקודת הזמן השלישית (index 2)
all_x_coords = XY[..., 0]            # קואורדינטת ה-x בלבד, מכל ריצה ומכל זמן

print("run2_full_trajectory.shape:", run2_full_trajectory.shape)
print("all_runs_at_t2.shape:", all_runs_at_t2.shape)
print("all_x_coords.shape:", all_x_coords.shape)
``````

### באג נפוץ: `a[0]` מול `a[:, 0]`

`Y[0]` ו-`Y[:, 0]` נראים דומים אך שונים לגמרי פיזיקלית: `Y[0]` הוא **כל מסלול הריצה הראשונה**, בעוד `Y[:, 0]` הוא **הגובה ההתחלתי של כל הריצות** (שכולן אמורות להתחיל מ-$y=0$). זו בדיוק דוגמה לבאג "ציר שגוי" ממאגר הבאגים של סעיף 7.10 — הקוד רץ בלי שגיאה בשני המקרים, אבל מחזיר משהו אחר לגמרי.
