``````python
import numpy as np
import matplotlib.pyplot as plt

g = 9.8
``````

### דוגמה: בניית רשת דו-ממדית עם `meshgrid`

לפעמים רוצים להעריך גודל פיזיקלי על פני **רשת שלמה** של שני פרמטרים בבת אחת — למשל הטווח $R(v_0,\theta)=v_0^2\sin(2\theta)/g$, על פני טווח של מהירויות **וגם** טווח של זוויות. `np.meshgrid` בונה משתי רשימות חד-ממדיות שתי רשתות דו-ממדיות תואמות.

``````python
v0_vals = np.linspace(5, 30, 50)          # 50 ערכי מהירות
theta_vals = np.radians(np.linspace(10, 80, 40))  # 40 ערכי זווית

V0, THETA = np.meshgrid(v0_vals, theta_vals, indexing="ij")
print("V0.shape:", V0.shape, " THETA.shape:", THETA.shape)   # שניהם (50, 40)

R = V0**2 * np.sin(2 * THETA) / g
print("R.shape:", R.shape)   # (50, 40) - טווח לכל צירוף (v0, theta)
``````

`meshgrid` בעצם עושה את מה ש-broadcasting היה עושה לבד (`v0_vals[:, None]` ו-`theta_vals[None, :]`) — אבל **בונה את שני המערכים המלאים במפורש**, מה שנוח כשרוצים גם לצייר אותם (heatmap/contour בשבוע 12) ולא רק לחשב איתם.

```{note}
`indexing="ij"` אומר שהציר הראשון של `V0`/`THETA` תואם לרשימה הראשונה (`v0_vals`). ברירת המחדל של NumPy היא `indexing="xy"`, שמחליפה את סדר הצירים — מקור נפוץ לבלבול. בקורס הזה נשתמש תמיד ב-`"ij"` כדי שהציר הראשון יתאים תמיד למשתנה הראשון שהעברנו.
```
