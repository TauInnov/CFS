### פריט 1: חיזוי הפלט

כדור נזרק ישר למעלה פעמיים, במהירויות שונות. **בלי להריץ**, כתבו מה תדפיס כל אחת משלוש שורות ההדפסה.

``````python
import numpy as np

v0_runs = np.array([10.0, 20.0])   # launch speed of each throw (m/s)
t = np.array([0.0, 1.0, 2.0])      # times (s)
g = 10.0                           # m/s^2

Y = v0_runs[:, None] * t - 0.5 * g * t**2   # height (m): (throw, time)
print(Y)
print(Y.mean(axis=0))
print(Y.max(axis=1))
``````

`````{admonition} פתרון
- ‎`Y`‎ בצורה ‎`(2, 3)`‎: שורה לכל זריקה ועמודה לכל זמן. בזריקה הראשונה הגובה בזמן 1 הוא ‎`10 - 5 = 5`‎, ובזמן 2 הוא ‎`20 - 20 = 0`‎.
- ‎`mean(axis=0)`‎ מעלים את ציר הזריקות: ממוצע לכל זמן, למשל ‎`(0 + 20) / 2 = 10`‎ בזמן 2.
- ‎`max(axis=1)`‎ מעלים את ציר הזמן: הגובה המרבי של כל זריקה.
`````

### פריט 3: שאלה מלאה — מדידות חוזרות

הטווח נמדד בשתי זוויות, ארבע חזרות בכל זווית. **בלי להריץ**, כתבו מה תדפיס כל אחת משורות ההדפסה A עד E. את כל החישובים אפשר לעשות בראש.

``````python
import numpy as np

R = np.array([[ 9.0,  9.0,  9.0, 13.0],
              [17.0, 21.0, 21.0, 21.0]])   # range (m): (angle, repetition)
N = R.shape[1]
print(R.shape, N)                 # line A
means = R.mean(axis=1)
print(means)                      # line B
s = R.std(axis=1, ddof=1)
print(s)                          # line C
print(s / np.sqrt(N))             # line D
print(R - means[:, None])         # line E
``````

`````{admonition} פתרון
```text

[10. 20.]

[1. 1.]
[[-1. -1. -1.  3.]
 [-3.  1.  1.  1.]]
```
- **A:** שתי זוויות וארבע חזרות, ולכן הצורה ‎`(2, 4)`‎, ו-‎`N`‎ הוא מספר החזרות, 4.
- **B:** ‎`axis=1`‎ מעלים את ציר החזרות: $(9+9+9+13)/4 = 10$ ו-$(17+21+21+21)/4 = 20$.
- **C:** בשורה הראשונה הסטיות הן ‎`-1, -1, -1, 3`‎, ובשנייה ‎`-3, 1, 1, 1`‎. בשתיהן סכום הריבועים 12, ו-$\sqrt{12/3} = 2$.
- **D:** שגיאת הממוצע היא $2/\sqrt{4} = 1$ בכל זווית.
- **E:** ‎`means[:, None]`‎ בצורה ‎`(2, 1)`‎, ולכן כל ממוצע מוחסר מכל השורה שלו (כלל 3): אלה הסטיות מסעיף C.
`````
