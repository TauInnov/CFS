## concatenate — לאחד ריצות

בפועל, הנתונים מגיעים לעתים בחלקים נפרדים: ריצה שנמדדה בנפרד, או מדידה אחת שנשמרה בשני קבצים. **‎`np.concatenate([a1, a2, ...])`‎** מחבר כמה מערכים למערך אחד לאורך ציר **קיים**, בלי ליצור ציר חדש: שני מערכים בצורה ‎`(30,)`‎ נותנים מערך אחד בצורה ‎`(60,)`‎.

התא הבא בונה את ‎`Y`‎, בצורה ‎`(5, 60)`‎, כמו בסעיף 9.1: מערך אפסים ולולאה על הריצות.

``````python
import numpy as np

g = 9.8
theta = np.radians(45)
v0_nom = 20.0
v0_runs = np.array([18.0, 19.5, 20.0, 20.5, 22.0])   # launch speeds (m/s)
n_runs, n_t = 5, 60
t_flight_nom = 2 * v0_nom * np.sin(theta) / g
t = np.linspace(0, t_flight_nom, n_t)   # the same time grid for every run

def height(v0, t):
    """Height (m) of one run with launch speed v0 (m/s), at the times t (s)."""
    return v0 * np.sin(theta) * t - 0.5 * g * t**2

Y = np.zeros((n_runs, n_t))   # one row per run, one column per time
for i in range(n_runs):
    Y[i] = height(v0_runs[i], t)
print("Y.shape:", Y.shape)
``````

### דוגמה: חיבור מערכים חד-ממדיים זה אחר זה

נניח שהגבהים של הריצה השלישית נשמרו בשני קבצים: 30 הזמנים הראשונים בקובץ אחד, ו-30 האחרונים בקובץ שני. ‎`concatenate`‎ מצרף את החלק השני לסוף החלק הראשון. ההדפסה האחרונה בודקת שהתוצאה זהה לשורה ‎`Y[2]`‎.

``````python
y_part1 = height(v0_runs[2], t[:30])   # the first 30 times (one file)
y_part2 = height(v0_runs[2], t[30:])   # the last 30 times (another file)

y_full = np.concatenate([y_part1, y_part2])
print(y_part1.shape, y_part2.shape, y_full.shape)
print(np.all(y_full == Y[2]))
``````

### דוגמה: הוספת ריצה חדשה עם ‎`concatenate`‎

בוצעה ריצה שישית. במקום לבנות הכול מחדש, מוסיפים אותה לאורך ציר הריצות הקיים. ‎`concatenate`‎ מחבר רק מערכים עם אותו מספר צירים: ל-‎`Y`‎ יש שני צירים, ולכן הריצה החדשה צריכה להיות מערך עם שורה אחת, בצורה ‎`(1, 60)`‎. לפי כלל 4 מסעיף 9.1, ‎`None`‎ בתוך סוגרי האינדקס מוסיף ציר חדש בגודל 1 במקום שבו הוא כתוב. לכן ‎`y_new[None, :]`‎ נותן בדיוק את הצורה הזו: ציר ריצות בגודל 1, ואחריו ציר הזמן. בסעיף 9.5 נחזור לכלל הזה בפירוט.

``````python
v0_new = 21.0                 # the launch speed of the sixth run (m/s)
y_new = height(v0_new, t)

# y_new[None, :] has shape (1, 60); None is explained in 9.5
Y_extended = np.concatenate([Y, y_new[None, :]])
print("Y_extended.shape:", Y_extended.shape)       # (6, 60)
``````

### באג נפוץ: לשכוח להוסיף ציר לריצה החדשה

אם כותבים ‎`np.concatenate([Y, y_new])`‎, בלי ‎`[None, :]`‎, הריצה ‎`y_new`‎ נשארת בצורה ‎`(60,)`‎, עם ציר אחד, ואילו ל-‎`Y`‎ יש שני צירים. ‎`concatenate`‎ אינו מוסיף ציר בעצמו, ולכן השורה נעצרת בשגיאה:

```text
ValueError: all the input arrays must have same number of dimensions, but the array at index 0 has 2 dimension(s) and the array at index 1 has 1 dimension(s)
```

המערך ב-‎`index 0`‎ הוא הראשון ברשימה, ‎`Y`‎, עם שני צירים, והמערך ב-‎`index 1`‎ הוא ‎`y_new`‎, עם ציר אחד. השגיאה כאן **כן** מופיעה, בניגוד לבאגי הצירים השקטים שראינו, והפתרון הוא ‎`y_new[None, :]`‎, כמו בדוגמה.

### תרגול עצמי

בתא הבא ‎`Y2`‎ מכיל שתי ריצות, ו-‎`c`‎ ריצה שלישית.

**לפני שמריצים:** מה תדפיס השורה הראשונה? ומה יקרה בשורה האחרונה?

``````python
Y2 = np.array([[0.0, 1.5, 2.0],
               [0.0, 2.5, 4.0]])      # two runs: shape (2, 3)
c = np.array([0.0, 3.5, 6.0])         # a third run
print(np.concatenate([Y2, c[None, :]]))
print(np.concatenate([Y2, c]))
``````

`````{admonition} פתרון
:class: dropdown, tip
השורה הראשונה מדפיסה שלוש ריצות, אחת בכל שורה:
```text
[[0.  1.5 2. ]
 [0.  2.5 4. ]
 [0.  3.5 6. ]]
```
‎`c[None, :]`‎ הופך את ‎`c`‎ לצורה ‎`(1, 3)`‎, ו-‎`concatenate`‎ מוסיף אותו לאורך ציר הריצות הקיים. השורה האחרונה נעצרת בשגיאה:
```text
ValueError: all the input arrays must have same number of dimensions, but the array at index 0 has 2 dimension(s) and the array at index 1 has 1 dimension(s)
```
ל-‎`Y2`‎ יש שני צירים, בצורה ‎`(2, 3)`‎, ול-‎`c`‎ ציר אחד, בצורה ‎`(3,)`‎. זה הבאג הנפוץ של הסעיף: חסר ‎`[None, :]`‎.
`````
