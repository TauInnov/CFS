# שבוע 5 — זיכרון ופונקציות מתקדמות

בשבוע החמישי נעמיק אל מאחורי הקלעים של פייתון ונבין כיצד היא **שומרת מידע בזיכרון**.  
נלמד על ההבדל החשוב בין אובייקטים *ניתנים לשינוי* (mutable) לבין *בלתי־ניתנים לשינוי* (immutable),  
ונראה כיצד זה משפיע על משתנים, על העתקות, ועל התנהגות פונקציות.

נמשיך לעסוק בפונקציות: מה קורה למשתנים שמועברים לפונקציה, פרמטרים עם ערכי ברירת מחדל,  
ופונקציה שמקבלת פונקציה אחרת כפרמטר.

בסוף השבוע תדעו:
- להבחין בין אובייקטים ניתנים ובלתי־ניתנים לשינוי  
- להבין מה פונקציה יכולה לשנות בערכים שהיא מקבלת, ומה לא  
- לזהות מתי צריך עותק של רשימה ולא שם נוסף לאותה רשימה  
- להשתמש בפרמטרים עם ערכי ברירת מחדל ולהעביר פונקציה כפרמטר
