# שבוע 8 — Matplotlib: לראות את התוצאה

עד עכשיו תרגמנו משוואה פיזיקלית למספרים (שבוע 7). השבוע לומדים **לראות** אותם: הספרייה **Matplotlib**, מהמודל המנטלי הבסיסי (`fig`, `ax`) ועד לגרף בסגנון דוח מעבדה עם קווי שגיאה.

השבוע מוקדם בכוונה בסדר הקורס — מעבדה א' דורשת גרפים לאורך הסמסטר, לא רק בסופו. **הדוגמה המובילה לאורך כל השבוע: מסלול הזריקה משבוע 7, עבור שלוש זוויות שיגור.**

בסוף השבוע תדעו:
- לבנות גרף עם המודל האובייקטי של Matplotlib (`fig, ax = plt.subplots()`)
- להבדיל בין עקומה רציפה (`plot`) לנקודות מדידה בדידות (`scatter`)
- לתייג גרף כראוי: צירים עם יחידות, מקרא, סימוני נקודה וחצים
- לצייר כמה עקומות על אותו גרף, ולפרוס כמה גרפים נפרדים באותו figure (`subplots`)
- לבחור בין סולם ליניארי ללוגריתמי, ולדעת מתי כל אחד מתאים
- לבנות היסטוגרמה ולהציג קווי שגיאה על נקודות מדידה — שני הכלים המרכזיים לניתוח נתוני מעבדה
- לשמור גרף כקובץ באיכות המתאימה לצירוף לדוח
