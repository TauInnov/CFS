## read_csv

עד עכשיו בנינו טבלאות ידנית בקוד. בפועל, נתוני מעבדה מגיעים מקובץ — לרוב CSV (Comma-Separated Values, טבלה כטקסט פשוט). `pd.read_csv("path/to/file.csv")` קורא אותו ישירות ל-`DataFrame`.

לצד המחברת הזו יש קובץ `launch_log.csv` — יומן של 30 זריקות (חוזרות על הניסוי של שבועות 7–9), בחמש זוויות שונות, שש חזרות בכל זווית.

### דוגמה

``````python
import pandas as pd

df = pd.read_csv("launch_log.csv")
df.head()
``````

``````python
print(df.shape)
print(df.dtypes)
``````

### באג נפוץ: עמודה מספרית שהתקבלה כטקסט

אם באחד התאים בקובץ ה-CSV יש טקסט חורג (למשל `"20.3?"` במקום `20.3`, או תא ריק שמסומן ב-`"N/A"` בלי שpandas מזהה אותו), כל העמודה הופכת לטיפוס טקסטואלי (`object` או `str`, תלוי בגרסת pandas) — כלומר pandas מתייחס לערכים כמחרוזות, לא כמספרים. הקוד **ימשיך לרוץ**, אבל כל חישוב מספרי על העמודה הזו (`()mean`, `()describe`) ייכשל או יתנהג מוזר. `df.dtypes` אחרי כל `read_csv` הוא בדיוק הבדיקה שתופסת את זה.

``````python
import io

bad_csv = io.StringIO("angle_deg,v0_measured\n15,20.3\n15,19.0?\n15,20.8\n")
df_bad = pd.read_csv(bad_csv)
print(df_bad.dtypes)   # v0_measured יצא כטיפוס טקסטואלי, לא float64 - בגלל ה-"19.0?"
``````
