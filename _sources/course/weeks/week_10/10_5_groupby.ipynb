## groupby(...).mean()

בסעיף 9.7 חישבנו ממוצע וסטיית תקן **על פני ריצות**, בעזרת `axis=0` על מערך NumPy — כי כל הריצות ישבו יפה בציר קבוע. ביומן המדידות שלנו, הריצות **מקובצות לפי זווית**, לא לפי מיקום קבוע במערך. בדיוק לשם כך נועד `groupby`:

```python
df.groupby("angle_deg")["range_measured"].mean()
```

"קבצו את כל השורות לפי ערך `angle_deg`, ובכל קבוצה חשבו את הממוצע של `range_measured`." — זו בדיוק אותה שאלה פיזיקלית כמו `Y.mean(axis=0)` בשבוע 9, רק שהקיבוץ הוא לפי **תווית בעמודה**, לא לפי מיקום בציר.

``````python
import pandas as pd
df = pd.read_csv("launch_log.csv")
``````

### דוגמה: טווח ממוצע וסטיית תקן, לפי זווית

``````python
mean_range_by_angle = df.groupby("angle_deg")["range_measured"].mean()
print(mean_range_by_angle)

# כמה עמודות סטטיסטיות בבת אחת:
summary = df.groupby("angle_deg")["range_measured"].agg(["mean", "std", "count"])
summary
``````

זו המסגרת שנשתמש בה גם בשבוע 11: `groupby` על תנאי הניסוי (כאן — הזווית), וממוצע/סטיית תקן על התוצאה הנמדדת (כאן — הטווח) בכל קבוצה.

### באג נפוץ: `groupby` בלי פעולת צבירה

`df.groupby("angle_deg")` **לבדו**, בלי `().mean`/`().sum`/וכו' בסוף, לא מחזיר טבלה או מספרים — הוא מחזיר אובייקט `DataFrameGroupBy`, "הבטחה לחישוב" שעדיין לא בוצע. קוד שנעצר שם (בטעות, או כי שכחו את הפעולה הבאה) לא יזרוק שגיאה, אבל גם לא יציג שום תוצאה שימושית.

``````python
grouped = df.groupby("angle_deg")
print(type(grouped))   # DataFrameGroupBy - עדיין לא חושב שום דבר
print(grouped)          # הדפסה לא שימושית - רק כתובת האובייקט בזיכרון
``````
