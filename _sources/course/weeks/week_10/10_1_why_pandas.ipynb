### דוגמה: אותם נתונים, שתי דרכים לייצג אותם

בשבוע 9 בנינו מערך `Y` בצורה `(ריצה, זמן)`. הפעם, במקום מסלול שלם לכל ריצה, יש לנו **סיכום** אחד לכל ריצה: הזווית שבה שוגרה, המהירות שנמדדה, והטווח שנמדד. אם ננסה לייצג את זה כמערכי NumPy נפרדים, נצטרך לזכור בעצמנו אילו שלושה מערכים "הולכים ביחד":

``````python
import numpy as np

run_id = np.array([1, 2, 3, 4, 5, 6])
angle_deg = np.array([15, 15, 15, 15, 15, 15])
v0_measured = np.array([20.3, 19.0, 20.8, 20.9, 18.0, 18.7])
range_measured = np.array([21.1, 18.2, 22.0, 22.1, 16.9, 18.1])

# כדי לדעת את כל הנתונים על ריצה 3, צריך לזכור שהאינדקס שלה הוא 2 בכל אחד מארבעת המערכים:
print(run_id[2], angle_deg[2], v0_measured[2], range_measured[2])
``````

זה עובד, אבל שביר: צריך לשמור ידנית על סנכרון בין ארבעה מערכים נפרדים, ואין שום דבר שמונע מהם "להיפרד" (למשל, אם ממיינים מערך אחד ושוכחים למיין את השאר). pandas פותר את זה — **טבלה אחת**, שבה כל שורה היא ריצה, וכל עמודה שומרת את השם והטיפוס שלה.

``````python
import pandas as pd

df = pd.DataFrame({
    "run_id": run_id,
    "angle_deg": angle_deg,
    "v0_measured": v0_measured,
    "range_measured": range_measured,
})
df
``````

כעת `df.loc[2]` (או, כפי שנרחיב בסעיף 10.4, `df[df["run_id"] == 3]`) מחזיר את **כל השורה** בבת אחת, עם שמות העמודות — אין צורך לזכור אינדקסים בין כמה מערכים.
