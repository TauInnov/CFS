### פריט 1: באגים בלולאת עדכון

הפונקציה ‎`simulate_cooling`‎ אמורה לחשב, דקה אחר דקה, את הטמפרטורה של גוף מתקרר, לפי הכלל מסעיף 10.3: $T_{n+1} = T_n - k\,(T_n - T_{room})$. הפונקציה ‎`first_below`‎ אמורה להחזיר את הדקה הראשונה שבה הטמפרטורה נמוכה מ-‎`T_limit`‎, או ‎`-1`‎ אם אין כזו. בקוד ארבעה באגים: שלושה ב-‎`simulate_cooling`‎ ואחד ב-‎`first_below`‎.

הקוד, כפי שקיבלתם אותו:

``````python
import numpy as np

def simulate_cooling(T0, k, T_room, n_steps):
    """Temperature (°C) of a cooling body, minute by minute."""
    T = np.zeros(n_steps)
    for i in range(n_steps):
        T[i + 1] = T[i] - k * (T[i + 1] - T_room)
    return T

def first_below(T, T_limit):
    """First minute with T below T_limit, or -1 if there is none."""
    for i in range(len(T)):
        if T[i] < T_limit:
            return i
        return -1

T = simulate_cooling(80.0, 0.1, 20.0, 4)
print(T)
print(first_below(np.array([80.0, 74.0, 68.6, 63.74]), 70.0))
``````

**א.** לפני שמחפשים באגים: מה **צריכה** להחזיר ‎`simulate_cooling(80.0, 0.1, 20.0, 4)`‎? מלאו טבלה של ‎`i`‎ ו-‎`T[i]`‎, כמו בסעיף 10.3.

**ג.** אחרי התיקון, הריצה השנייה מדפיסה:

השוו לטבלה מסעיף א. מהו הערך הראשון ששונה, ומה התיקון?

**ד.** אחרי התיקון השני, הריצה השלישית מדפיסה:

הגוף מתחמם במקום להתקרר. מהו הערך הראשון ששונה, איזו שורה מחשבת אותו, ומה התיקון?

**ה.** ‎`first_below`‎ מחזירה ‎`-1`‎, אף שהטמפרטורות בשורה 19 יורדות מתחת ל-70 בדקה 2. מצאו את הבאג ותקנו אותו. מה תחזיר הפונקציה המתוקנת על הטמפרטורות מסעיף א, עם גבול של 60 מעלות?

### פריט 2: שאלה מלאה — מצאו את שלוש הטעויות

גוף נע לאורך קו ישר, ומיקומו ‎`x`‎ נמדד בזמנים ‎`t`‎. הפונקציה ‎`velocity`‎ אמורה להחזיר את המהירות בין כל שתי מדידות סמוכות, והפונקציה ‎`distance_so_far`‎ את המרחק שהגוף עבר מההתחלה ועד כל מדידה. הפלט הצפוי כתוב בהערות של שתי שורות ההדפסה.

בקוד שלוש טעויות. **קראו אותו בלי להריץ**: מצאו כל טעות לפי מספר השורה, תקנו אותה, והסבירו במשפט אחד מה היה לא נכון. ציינו גם איזו טעות מתגלה ראשונה כשמריצים את הקוד.

``````python
import numpy as np

def velocity(t, x):
    """Velocity (m/s) between consecutive measurements."""
    return np.diff(t) / np.diff(x)

def distance_so_far(t, x):
    """Distance (m) traveled from the start up to each measurement."""
    v = velocity(t, x)
    dt = t[1] - t[0]
    steps = abs(v) * dt
    return np.concatenate([0.0, np.cumsum(steps)])

t = np.array([0.0, 1.0, 2.0, 4.0])   # time (s)
x = np.array([0.0, 4.0, 6.0, 8.0])   # position (m)
print(velocity(t, x))          # expected: [4. 2. 1.]
print(distance_so_far(t, x))   # expected: [0. 4. 6. 8.]
``````

`````{admonition} פתרון
:class: dropdown, tip
1. **שורה 5:** המהירות היא הפרש המיקומים חלקי הפרש הזמנים, ולא להפך. התיקון: ‎`return np.diff(x) / np.diff(t)`‎.
2. **שורה 12:** ‎`np.concatenate`‎ מחבר מערכים, ו-‎`0.0`‎ הוא מספר בודד ולא מערך. התיקון: ‎`np.concatenate([[0.0], np.cumsum(steps)])`‎.
3. **שורות 10–11:** הזמנים לא במרווחים שווים (המרווח האחרון הוא 2 שניות), ו-‎`t[1] - t[0]`‎ הוא רק המרווח הראשון. כל צעד צריך את המרווח שלו. התיקון: ‎`steps = abs(v) * np.diff(t)`‎, ואז השורה של ‎`dt`‎ מיותרת.

**הסדר שבו הטעויות מתגלות:** שורת ההדפסה הראשונה מדפיסה ‎`[0.25 0.5  1.  ]`‎ במקום ‎`[4. 2. 1.]`‎; זו טעות 1. השורה השנייה נעצרת בשגיאה של טעות 2:

```text
ValueError: zero-dimensional arrays cannot be concatenated
```

השגיאה הזו מסתירה את טעות 3: רק אחרי שמתקנים את טעויות 1 ו-2 הפלט הוא ‎`[0. 4. 6. 7.]`‎, והמספר האחרון עדיין שגוי.

הקוד המתוקן:

```python
`````

### פריט 3: תחזית מול מדידה (דוגמה פתורה)

כדור נזרק ישר למעלה במהירות 15 מ'/ש', וגובהו נמדד בסרטון של 20 תמונות בשנייה, במשך 2.5 שניות (‎`t_video`‎ ו-‎`y_video`‎, במטרים). קבוע ההתנגדות ‎`k`‎ לא ידוע, ולכן הקוד מחשב בלולאה של 10.3 תחזית לשלושה ערכים של ‎`k`‎ בבת אחת, ומשווה כל אחת למדידה. צעד הזמן של התחזית קטן פי 10 מהזמן שבין שתי תמונות, כדי ששגיאת הצעד תהיה קטנה, ו-‎`y[:, ::10]`‎ לוקח כל ערך עשירי לאורך ציר הזמן: את התחזית ברגעי הצילום.

``````python
# "measured" height (m) of a ball thrown up at 15 m/s, video at 20 frames/s
rng = np.random.default_rng(0)
t_video = np.round(np.arange(51) * 0.05, 2)
k_video = 0.14
v_end = 9.8 / k_video
y_video = ((15.0 + v_end) / k_video * (1 - np.exp(-k_video * t_video))
           - v_end * t_video)
y_video = np.round(y_video + rng.normal(0, 0.01, t_video.shape), 3)
y_video[0] = 0.0
``````

``````python
g, v0 = 9.8, 15.0
k = np.array([0.0, 0.1, 0.2])          # three guesses for k (1/s)
dt = 0.005                             # 10 steps between two frames
n_steps = 10 * (len(t_video) - 1) + 1

v = np.zeros((3, n_steps))             # (k value, step)
y = np.zeros((3, n_steps))
v[:, 0] = v0
for n in range(n_steps - 1):
    v[:, n + 1] = v[:, n] - (g + k * v[:, n]) * dt
    y[:, n + 1] = y[:, n] + v[:, n] * dt

y_frames = y[:, ::10]                  # the prediction at the frame times
gap = y_frames - y_video               # prediction minus measurement
print(y_frames.shape, gap.shape)
print(gap[:, -1])                      # the gap at the last frame (m)
``````

``````python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(1, 2, figsize=(10, 4))
axs[0].plot(t_video, y_video, "o", markersize=3, color="black",
            label="Measured")
for j in range(3):
    axs[0].plot(t_video, y_frames[j], label=f"k = {k[j]} 1/s")
    axs[1].plot(t_video, gap[j], label=f"k = {k[j]} 1/s")
axs[1].axhline(0, color="gray", linewidth=0.8)
axs[0].set_xlabel("t (s)")
axs[0].set_ylabel("y (m)")
axs[0].set_title("Height: measurement and predictions")
axs[1].set_xlabel("t (s)")
axs[1].set_ylabel("Prediction minus measurement (m)")
axs[1].set_title("The gap along the time axis")
axs[0].legend()
axs[1].legend()
plt.show()
``````

‎`y_video`‎, בצורה ‎`(51,)`‎, מותאם לכל שורה של ‎`y_frames`‎, ולכן ב-‎`gap`‎ יש פער לכל ערך של ‎`k`‎ ולכל רגע. בלי התנגדות התחזית גבוהה מהמדידה, והפער גדל עד 2.6 מטר. עם ‎`k = 0.1`‎ היא גבוהה ב-0.7 מטר בסוף, ועם ‎`k = 0.2`‎ נמוכה ב-0.8 מטר: המדידה נמצאת בין שתיהן.
