## 10.2 — DataFrame ו-Series

שני האובייקטים המרכזיים ב-pandas:

- **`DataFrame`** — הטבלה כולה: כמה עמודות, כל אחת עם שם וטיפוס.
- **`Series`** — **עמודה בודדת**: וקטור של ערכים עם שם ו-index. כל עמודה בתוך `DataFrame` היא בעצם `Series`.

לכל `DataFrame` יש גם `index` — תווית לכל שורה (כברירת מחדל, 0, 1, 2... אבל אפשר לשנות).

``````python
import pandas as pd

df = pd.DataFrame({
    "run_id": [1, 2, 3, 4, 5, 6],
    "angle_deg": [15, 15, 15, 15, 15, 15],
    "v0_measured": [20.3, 19.0, 20.8, 20.9, 18.0, 18.7],
    "range_measured": [21.1, 18.2, 22.0, 22.1, 16.9, 18.1],
})
``````

### דוגמה: `df.columns`, `df.index`, `df.dtypes`, `df.head()`

``````python
print("columns:", list(df.columns))
print("index:", list(df.index))
print()
print(df.dtypes)
print()
df.head(3)   # שלוש השורות הראשונות בלבד
``````

### באג נפוץ: `df["col"]` מול `df[["col"]]`

זה אחד הבלבולים הנפוצים ביותר בקריאת קוד pandas: סוגריים מרובעים בודדים מחזירים **`Series`** (עמודה בודדת, חד-ממדית), וסוגריים מרובעים כפולים מחזירים **`DataFrame`** בעל עמודה אחת (עדיין דו-ממדי). הם נראים דומים בפלט, אבל הם טיפוסים שונים עם התנהגות שונה.

``````python
col_as_series = df["angle_deg"]
col_as_df = df[["angle_deg"]]

print(type(col_as_series))
print(type(col_as_df))
print(col_as_series.shape)   # (6,) - חד-ממדי
print(col_as_df.shape)       # (6, 1) - דו-ממדי
``````

### נסו בעצמכם

נחשו מה יחזיר `df[["angle_deg", "v0_measured"]]` (עם רשימה של שני שמות עמודה) — `Series` או `DataFrame`? ומה יהיה ה-`shape`?

``````python
# נחשו, ואז בדקו:
result = df[["angle_deg", "v0_measured"]]
print(type(result), result.shape)
``````
