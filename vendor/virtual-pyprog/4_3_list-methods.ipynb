# מתודות של רשימות

**תזכורת:**
- mutable (בר שינוי) – ניתן לשנות את האובייקט המקורי לאחר יצירתו
immutable (בלתי־ניתן לשינוי) – לא ניתן לשנות את האובייקט המקורי לאחר יצירתו. כל שינוי ניתן רק על ידי שכפול והחלפה של ערכים.
- שיטה (מתודה) היא פונקציה שמוגדרת **עבור טיפוס מסוים**, והיא "שייכת" לאובייקט שמאותו הטיפוס. האובייקט יכול לקרוא לה כדי לבצע פעולה כלשהי לפי התחביר הבא:

```python
my_object.action()
```

### `append`: הוספת איבר לרשימה

הוספת איבר ל**סוף** הרשימה.

`````python
hello_list = ['h','e','l','l','o']
hello_list_new=hello_list.append('!')
print(hello_list) # This is the original list!!
`````

```{admonition} שימו לב
:class: error

התו `'!'` נוסף **לרשימה המקורית** ללא פעולת השמה.
```

#### אז מה מחזיר `append`?

`````python
hello_list = ['h','e','l','l','o']
hello_list_new=hello_list.append('!')
print(hello_list)
`````

`````python
print(hello_list_new)
`````

כלומר הפונקציה לא החזירה כלום, וההשמה לאחר `append` מיותרת.

`````python
hello_list = ['h','e','l','l','o']
hello_list.append('!')
print(hello_list)
`````

```{admonition} **טיפ להמשך** 
:class: tip

אם מתודה משנה את האובייקט המקורי, היא כנראה לא תחזיר אותו. למה לעשות פעמיים?
```

מה קורה אם מתעלמים מהכלל הזה? אחרי ההשמה `lst = lst.append(3)` המשתנה `lst` כבר אינו רשימה, ושגיאה תופיע רק בפעם הבאה שמשתמשים בו:

`````python
lst = [1, 2]
lst = lst.append(3)   # lst is now None
lst.append(4)
`````

```{admonition} **טיפ**
:class: tip

כשבהודעת שגיאה מופיעה המילה `NoneType`, כדאי לחפש משתנה שקיבל ערך מפונקציה או ממתודה שלא מחזירה כלום (או ששכחו בה `return`).
```

### `remove`: הסרת איבר מרשימה

המתודה מסירה את האיבר הראשון המתאים לערך שהועבר בקלט

`````python
hello_list = ['h','e','l','l','o']
hello_list.remove('e')
print(hello_list)
`````

מה יקרה אם ננסה להסיר איבר שלא קיים ברשימה?

`````python
hello_list = ['h','e','l','l','o']
hello_list.remove('e')
print(hello_list)
hello_list.remove('e')
print(hello_list)
`````

### `pop`: חילוץ איבר מרשימה

המתודה `pop` מאפשרת לנו להוציא איבר מרשימה לפי אינדקס וגם לקבל אותו חזרה. כלומר, כשאנחנו קוראים ל־`pop(i)`:

1. המתודה שומרת את האיבר שנמצא באינדקס i.
2. היא מוחקת את האיבר הזה מהרשימה.
3. היא מחזירה לנו את הערך שהיה באינדקס i, כך שנוכל להשתמש בו מיד או לשמור אותו במשתנה.

כך אפשר לעבוד עם איברים ברשימה בלי לאבד אותם, וגם לעדכן את הרשימה בו-זמנית.

`````python
hello_list = ['h','e','l','l','o']
res=hello_list.pop(0)
print(hello_list, res)
hello_list.pop() # Note that here we do not assign the popped element to res
print(hello_list, res)
`````

```{admonition} **שימו לב** 
:class: error

פעולה זו לא "שוברת" את כלל האצבע מקודם - המתודה הזו עדיין לא מחזירה את הרשימה המקורית, אלא ערך אחר.

```

### `sort`: מיון רשימה

המתודה ממיינת את **הרשימה המקורית**.

שימו לב שכאשר משתמשים ב**פונקציה המובנית (built-in function)** `sorted` שראינו כבר, ערך ההחזר הוא **רשימה חדשה** ממויינת

`````python
a = [3,7,1]
b = sorted(a)
print(a)
print(b)
`````

לעומת זאת, ה**מתודה** `sort`  ממיינת את **הרשימה המקורית** ואינה מחזירה דבר

`````python
a = [3,7,1]
b = a.sort()
print(a)
print(b)
`````

ראינו פה שני דרכים שונות לפעולה של פונקציות:  
פונקציות שמשנות את האובייקט המקורי נקראות **in-place operation**.  
לעומתן, פונקציות שיוצרות ערך חדש, מבלי לשנות את האובייקט המקורי, נקראות **copy operations**.

```{admonition} **עיצרו וחישבו:** האם קיימים in-place operations למחרוזות? אם כן, אילו? אם לא, מדוע?
:class: dropdown, caution

לא - מחרוזות בפייתון הן immutable (לא ניתנות לשינוי). כל פעולה על מחרוזת יוצרת מחרוזת חדשה במקום לשנות את המקור. לדוגמה, חיתוך או replace מחזירים מחרוזת חדשה, והמקור נשאר כפי שהוא.

```

### רשימה כארגומנט של פונקציה

כשמעבירים רשימה לפונקציה, הפונקציה מקבלת את הרשימה עצמה ולא עותק שלה. לכן פעולה שמשנה את הרשימה, כמו השמה לאיבר (`values[0] = ...`) או מתודה כמו `append`, נראית גם מחוץ לפונקציה. לעומת זאת, השמה לפרמטר עצמו (`x = ...`) משנה רק את המשתנה המקומי של הפונקציה:

`````python
def reset_first(values):
    values[0] = 0.0

def reset(x):
    x = 0.0

data = [3.5, 2.0]
v = 3.5
reset_first(data)
reset(v)
print(data, v)
`````

```{admonition} **עיצרו וחישבו:** הסבירו במשפט אחד למה `data` השתנתה ואילו `v` לא.
:class: dropdown, caution

הפונקציה `reset_first` קיבלה את הרשימה עצמה וכתבה לתוך אחד האיברים שלה, ולכן השינוי נראה גם מחוץ לפונקציה. הפונקציה `reset` קיבלה את הערך 3.5 ושינתה רק את המשתנה המקומי `x`, ולכן `v` נשאר 3.5.
```

### מצאו את השגיאות

כדור מופל ממנוחה מגובה 4.9 מטר, וזמן הנפילה נמדד ארבע פעמים. המדידה השלישית (אינדקס 2) שגויה, ולכן מוציאים אותה מהרשימה, ממצעים את שאר הזמנים ומחשבים את תאוצת הנפילה לפי $g = 2h/t^2$.

הקוד שלפניכם אמור להדפיס `g = 9.8 m/s^2`, אבל יש בו **שלוש שגיאות**. קראו את הקוד לפני שאתם מריצים אותו, מצאו את השגיאות, תקנו אותן והסבירו במשפט אחד מה הייתה הבעיה בכל אחת.

`````python
def remove_outlier(times, bad_index):
    times.remove(bad_index)

def mean(values):
    return sum(values) / len(values)

def free_fall_g(h, t):
    return 2 * h / t^2

h = 4.9
drop_times = [1.01, 0.99, 1.62, 1.00]
drop_times = remove_outlier(drop_times, 2)
t = mean(drop_times)
print("g =", free_fall_g(h, t), "m/s^2")
`````

````{admonition} **לחצו כאן כדי לצפות בפתרון**
:class: dropdown, tip

1. `times.remove(bad_index)` צריך להיות `times.pop(bad_index)`. המתודה `remove` מוחקת לפי **ערך**, ולכן היא מחפשת ברשימה את המספר 2 ומעלה `ValueError`. המתודה `pop` מוחקת לפי **מיקום**.
2. `drop_times = remove_outlier(drop_times, 2)` צריך להיות `remove_outlier(drop_times, 2)`, בלי השמה. הפונקציה משנה את הרשימה ואינה מחזירה כלום, ולכן ההשמה הופכת את `drop_times` ל-`None`.
3. `t^2` צריך להיות `t**2`. בפייתון `^` אינו חזקה.

שימו לב שהשגיאה הראשונה היא שמסתירה את האחרות: הקוד נעצר כבר ב-`ValueError`, ורק אחרי תיקונה מופיעה השגיאה `'NoneType' object is not iterable`, ואחריה השגיאה של `^`.

```python
def remove_outlier(times, bad_index):
    times.pop(bad_index)

def mean(values):
    return sum(values) / len(values)

def free_fall_g(h, t):
    return 2 * h / t**2

```
