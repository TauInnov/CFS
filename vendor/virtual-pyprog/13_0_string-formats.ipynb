# פורמט מחרוזות (f-strings)

## עריכת מחרוזות לפי פורמט (String Formatting)

לעיתים קרובות נרצה להדפיס הודעה שמשלבת טקסט עם ערכים של משתנים, למשל שם של סטודנט והציון שלו. דרך אחת היא לחבר מחרוזות בעזרת `+`, אבל אז צריך להמיר כל מספר למחרוזת בעזרת `str`:

`````python
name="Avi"
grade=90
print(name+"-"+str(grade))
`````

זה עובד, אבל הקוד מסורבל, והוא נעשה קשה יותר לקריאה ככל שמחברים יותר ערכים.

פתרון נוח יותר הוא **f-string**: מחרוזת שלפני המירכאות שלה כותבים את האות f. בתוך המחרוזת אפשר לכתוב סוגריים מסולסלים, ובתוכם ביטוי בפייתון: שם של משתנה, חישוב או קריאה לפונקציה. פייתון מחשבת את הביטוי, ממירה את התוצאה למחרוזת ומשלבת אותה במקום שבו נכתבה.

למשל, כך מדפיסים את כל השמות והציונים שבמילון הסטודנטים:

`````python
studs = {"Avi": 90, "Moshe": 80, "Adam": 70}

for name, grade in studs.items():
    print(f"{name}-{grade}")
`````


שימו לב שבסך הכול שמנו את המשתנים בתוך הסוגריים המסולסלים המתאימים, והString format כבר דאג לחבר אותם ביחד.  

```{admonition} **נסו בעצמכם**
:class:  caution
מה לגבי פעולות קוד יותר מורכבות בתוך הString format? נניח כעת כי נרצה להדפיס את הציון לאחר פקטור של תוספת 10 נקודות. על מנת לבצע זאת, הוסיפו ‎`+10`‎ בתוך הסוגריים המסולסלים המכילים את המשתנה grade. 

אחרי שהצלחתם, נסו "לשחק" עם הפורמט: להוסיף פעולות יותר מורכבות בתוך הסוגריים המסולסלים, להוסיף זוגות של סוגריים מסולסלים נוספים וכו' 
```
  

כשמדפיסים תוצאה של חישוב, בדרך כלל לא רוצים לראות את כל הספרות שאחרי הנקודה העשרונית. ניקח לדוגמה את המספר פאי:

`````python
name = 'Avi'
pi = 3.14159265359
print(f"My name is {name} and I like {pi}")
`````

נרצה כעת "לקצץ" את הדיוק של מספר הפאי ל2 ספרות לאחר הנקודה. על מנת לממש זאת, נוכל להוסיף בסוף הסוגריים המסולסלים נקודותיים (`:`) עם רמת הדיוק הנדרשת (‎`.2`‎) והדרך בה נרצה שמספר יוצג (`f`, כמו `float`/עשרוני) 

`````python
print(f"My name is {name} and I like {pi:.2f}")
`````

באופן כללי, אחרי הערך כותבים נקודתיים, ואחריהן את צורת התצוגה:

```
{value:.2f}
```

- הנקודתיים מפרידות בין הערך לבין צורת התצוגה.
- המספר שאחרי הנקודה קובע כמה ספרות יוצגו אחרי הנקודה העשרונית.
- האות האחרונה קובעת את סגנון התצוגה: f לתצוגה עשרונית רגילה, e לתצוגה מדעית.

המספר מעוגל לצורך התצוגה בלבד, והערך ששמור במשתנה לא משתנה.

כדי לעגל את הערך עצמו, ולא רק את התצוגה שלו, משתמשים בפונקציה `round`. היא מחזירה מספר, ואילו f-string מחזירה מחרוזת. למשל, אחרי שלושה צעדי קירור של גוף מ-80 מעלות בחדר של 20 מעלות:

`````python
T = 80.0
for i in range(3):
    T = T - 0.1 * (T - 20.0)    # one cooling step, k = 0.1

print(T)                        # every digit Python stores
print(round(T, 2), type(round(T, 2)))
print(f"{T:.2f}", type(f"{T:.2f}"))
`````

בפיזיקה פוגשים הרבה מספרים קטנים מאוד או גדולים מאוד. ניקח לדוגמה את קבוע פלאנק, ביחידות של ג'ול כפול שנייה. בקוד כותבים מספר כזה בעזרת האות e, שפירושה "כפול 10 בחזקת". אם נדפיס אותו בתצוגה עשרונית רגילה נקבל אפס, וזה לא שימושי:

`````python
h = 6.62607015e-34
print(f"h = {h:.2f}")
`````

ב**תצוגה מדעית** (האות e, כמו אקספוננט) המספר מוצג כמספר בין 1 ל-10, כפול חזקה של 10:

`````python
print(f"h = {h:.2e}")
`````

הפלט פירושו 6.63 כפול 10 בחזקת מינוס 34. גם כאן המספר שאחרי הנקודה קובע כמה ספרות יוצגו אחרי הנקודה העשרונית.

```{admonition} **נסו בעצמכם**
:class:  caution

נסו להשתמש בתצוגה מדעית על מנת לייצג גם מספרים מאוד גדולים, כמו 31415926535900000 

```
  

לסיכום, תצוגות מספריות שונות בString format מאוד שימושיות בייצוג מספרים ארוכים כמחרוזת. קיימות תצוגות מספריות שימושיות נוספות, כמו תצוגה באחוזים.

אנו מזמינים אתכם לקרוא עוד על String format [בתיעוד הרשמי של פייתון](https://docs.python.org/3/library/string.html#format-examples).

## בחנו את עצמכם

הפונקציה `cool` מחזירה את הטמפרטורה של גוף אחרי צעד קירור אחד בחדר בטמפרטורה `T_room`. המילון `rooms` מהדף הקודם מחזיק את טמפרטורת החדר במקומות שונים. מה ידפיס הקוד הבא? ענו לפני שאתם מריצים.

```python
def cool(T, k, T_room=20.0):
    return T - k * (T - T_room)

rooms = {"lab": 20.0, "fridge": 4.0}
for name, T_room in rooms.items():
    T = cool(80.0, 0.5, T_room)
    print(f"{name}: {T:.1f}")
```

### מצאו את 3 השגיאות

סטודנט ניסה לכתוב את אותו קוד, ובסוף גם להדפיס את טמפרטורת המעבדה. קראו את הקוד **לפני** שאתם מריצים אותו. מצאו את שלוש השגיאות, תקנו אותן, והסבירו במשפט אחד מה הבעיה בכל אחת. איזו שגיאה תראו ראשונה כשתריצו את הקוד?

`````python
def cool(T, k, T_room=20.0):
    return T - k * (T - T_room)


rooms = {"lab": 20.0, "fridge": 4.0}
for name, T_room in rooms:
    T = cool(80.0, 0.5, T_room)
    print("{name}: {T:.1f}")
print(rooms["Lab"])
`````

````{admonition} **לחצו כאן כדי לצפות בפתרון**
:class: dropdown, tip

```python
def cool(T, k, T_room=20.0):
    return T - k * (T - T_room)


rooms = {"lab": 20.0, "fridge": 4.0}
for name, T_room in rooms.items():
    T = cool(80.0, 0.5, T_room)
    print(f"{name}: {T:.1f}")
print(rooms["lab"])
```

1. **‎`for name, T_room in rooms`‎.** לולאה על מילון עוברת רק על המפתחות. כל מפתח הוא מחרוזת אחת, ופייתון מנסה לפרק אותה לשני משתנים, ולכן מתקבלת ‎`ValueError: too many values to unpack`‎. כדי לקבל זוגות של מפתח וערך צריך ‎`rooms.items()`‎.
2. **חסר `f` לפני המחרוזת.** בלי `f`, הסוגריים המסולסלים הם סתם תווים, ויודפס ‎`{name}: {T:.1f}`‎ כמו שהוא.
3. **‎`rooms["Lab"]`‎.** מפתחות במילון רגישים לאותיות גדולות וקטנות: ‎`"Lab"`‎ ו-‎`"lab"`‎ הם מפתחות שונים, ולכן מתקבלת `KeyError`.

**מה רואים בהרצה?** קודם `ValueError`, בשורת הלולאה. היא מסתירה את שתי האחרות: אחרי התיקון הראשון מודפסות השורות בלי ערכים (שגיאה 2), ובסוף מתקבלת `KeyError` (שגיאה 3).
````
